### $\text{Chapter 4 – Metamerism:}$

#### $\text{What is Metamerism?}$
- $\text{Metamerism is a phenomenon in which different combinations of light wavelengths in the visible spectrum can appear to our eyes as the same color.}$
- $\text{In other words, even if two light sources have different spectral compositions} \\ \text{(different sets of wavelengths), they may be perceived by the human eye as the same color.}$
- $\text{This happens because the human visual system interprets light based on how the three types of cone cells in our eyes are stimulated—not} \\  \text{on the exact wavelengths present.}$

#### $\text{Metamers – Spectrally Different Lights That Look the Same}$
- $\text{Metamers are pairs (or sets) of light profiles—each made up of different wavelength combinations—that the eye perceives as identical in color.}$
- $\text{Even though their physical compositions differ, our brains interpret them as the same due to similar cone cell responses.}$

#### $\text{Tristimuli and Color Reproduction}$
- $\text{In an experiment, it was shown that by combining three specific wavelengths (430 nm, 540 nm, and 700 nm) at different intensities,} \\ \text{nearly all visible colors can be produced.}$
- $\text{These three wavelengths are known as tristimuli.}$
- $\text{This fact helps explain why using only three types of color receptors in the eye} \\ \text{(cones sensitive to red, green, and blue) allows us to perceive a full range of colors.}$

#### $\text{Why 3 Cone Types Are Enough}$
- $\text{Because each cone responds to a range of wavelengths and overlaps with the others,} \\ \text{the combination of their responses can uniquely determine the color of light.}$
- $\text{Thus, the human visual system can identify almost any color with just three types of cones.}$

#### $\text{Color Mixing from Other Triplets}$
- $\text{It's not only the red, green, and blue wavelengths that can be used to reproduce colors.}$
- $\text{There are many possible triplets of wavelengths that can be mixed to produce a wide color gamut.}$
- $\text{However, not all triplets are equally effective. Some combinations fail to reproduce the full range of visible colors.}$

#### $\text{Color Spaces Are Usually 3D}$ 
- $\text{Due to the trichromatic nature of human vision, most color spaces (like RGB, CIE) are modeled as three-dimensional.}$
- $\text{These models allow efficient representation and manipulation of colors in digital systems and display technologies.}$

---

### $\text{Contextual Color Perception – Not Just Physical}$
#### $\text{Color Depends on Background and Context}$
- $\text{The same physical color can appear differently depending on its surroundings.}$
- $\text{For example, a blue light and a yellow light might look like different colors because of their background contrast, even if they are physically identical in color.}$

#### $\text{They Are Not Metamers – They Are Truly the Same Color}$
- $\text{This is not metamerism.}$
- $\text{These are cases where the physical color is the same, but our perception shifts due to the surrounding context.}$
- $\text{This is called simultaneous contrast—our eyes interpret colors relatively rather than absolutely.}$

#### $\text{Color Is Interpreted Relatively, Not Absolutely}$
- $\text{Our eyes and brain don't perceive color in isolation—they evaluate it based on surrounding colors and lighting conditions.}$
- $\text{This leads to contextual misinterpretation, where the same wavelength combination can look different depending on adjacent colors.}$

---

### $\text{Conclusion – Metamerism and Color Perception}$
- $\text{Metamerism shows that our color perception is based on how cones are stimulated, not the physical light spectrum.}$
- $\text{It also explains how devices (screens, printers) can reproduce colors using only three primary lights or inks.}$
- $\text{Our perception of color is a complex interplay between physical light and neural interpretation, highly influenced by context.}$
