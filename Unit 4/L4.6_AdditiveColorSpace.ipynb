## $ \text{Display Devices and the RGB Color Triangle} $

#### $ \text{Why Are Only 3 Colors Used in Displays?} $

$ \text{Display devices use three primary colors—red, green, and blue—to represent color.} $
$ \text{These primary colors correspond to specific chromaticities (points) in the CIE xy chromaticity diagram.} $
$ \text{By marking the red, green, and blue of a particular device in the diagram, we can form a triangle.} $
$ \text{All colors within this triangle can be produced by additive combinations of these three primaries.} $

#### $ \text{Not All Visible Colors Are in the Triangle} $

$ \text{Not every color in the visible spectrum is within this RGB triangle.} $
$ \text{Some colors (especially very saturated ones) lie outside, so the device cannot reproduce them.} $
$ \text{This is a limitation of the specific red, green, and blue points used by the device.} $

#### $ \text{How Does a Pixel Emit Color?} $

$ \text{Each pixel on a raster display (like a screen) is made of subpixels that emit red, green, and blue light in varying intensities.} $
$ \text{Combining these intensities allows for the generation of a wide range of perceivable colors.} $
$ \text{Even though only three lights are emitted, the eye perceives many more due to metamerism—different spectral mixtures that appear the same to us.} $

#### $ \text{The Role of White Light} $

$ \text{When a projector shines white light on a white screen, all the incident light is reflected, resulting in white.} $
$ \text{This reflection includes all wavelengths equally and is thus additive in nature.} $
$ \text{That’s why combining red, green, and blue at full intensity gives white: each contributes its part.} $

---

#### $ \text{Comparing Displays Using the CIE Diagram} $

$ \text{Different Devices = Different Triangles} $

$ \text{Different monitors, projectors, and displays use different red, green, and blue primaries.} $
$ \text{This results in different triangles within the CIE diagram, affecting the range of colors (gamut) the device can represent.} $
$ \text{The larger the triangle, the broader the color gamut.} $
$ \text{Comparing the triangles allows us to compare display capabilities.} $

#### $ \text{Colors Outside the Triangle} $

$ \text{No triangle using three RGB points can cover the entire CIE diagram.} $
$ \text{Colors outside the triangle cannot be represented by that device, even though we can see them with our eyes.} $
$\text{This system was chosen since humans are more sensitive to brightness, allowed black-white TVs to show picture (form Y channel) and allowed efficient bandwidth usage}$

### $ \text{Encoding Colors Digitally} $

#### (Y, I, Q) Color System

$\text{Y := Luma (Brightness)}$
$\text{I := Chrominance (Color)}$
$\text{Q := Chrominance (saturation)}$

- $\text{This system was developed by televesion broadcasting, which encodes color images in a way that seperates brightness from color information}$
- $\text{}

#### $ \text{Digital Color Representation} $

$ \text{Each pixel’s color is encoded using three channels—Red (R), Green (G), and Blue (B).} $
$ \text{Each channel is usually represented by 8 bits (256 intensity levels), giving } 256^3 = 16,777,216 \text{ colors.} $
$ \text{These values form a 3D RGB cube, where each point represents a different color.} $
$ \text{At full intensity (255,255,255), the color is white. At zero intensity (0,0,0), it’s black.} $

$ \text{Additive vs. Subtractive Color} $

$ \text{RGB is an additive system, where light is emitted.} $
$ \text{CMY (Cyan, Magenta, Yellow) is a subtractive system, where light is absorbed and reflected—like in printing.} $
$ \text{Each subtractive primary (CMY) complements a pair of RGB primaries:} $
	•	$ \text{Cyan absorbs Red, reflects Green and Blue} $
	•	$ \text{Magenta absorbs Green, reflects Red and Blue} $
	•	$ \text{Yellow absorbs Blue, reflects Red and Green} $

$ \text{These complementary relationships allow us to understand color reproduction across media.} $

--

#### $ \text{Capturing Colors: RGB in Cameras} $

$ \text{Cameras use three color sensors—each sensitive to different parts of the visible spectrum (red, green, blue).} $
$ \text{The resulting image is encoded using the RGB model, allowing consistent reproduction across screens.} $
$ \text{This is why RGB is dominant in both input (cameras) and output (screens).} $
