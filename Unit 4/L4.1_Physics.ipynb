## Physics
#### Objective: 
- $\text{Understand the phyisical definition of color and its basics properties}$

- $\text{ The world we live in is filled electromagnetic radiation, wher it created inside atom that make up everything around us, where radiation is a stream of particles called "Photons"}$
- $\text{Photons can also be viewed as wave traveling through space, where they travel at a fixed speed of light}$
- $\text{There are different types of electromagnetic waves, where the differences is measured in their wavelength := The length between two consecutive peaks of the wave}$
- $\text{Due to the constant speed of these wave, they also determine the frequency := which is the difference in time between two consecutive peaks of the wave}$
- $\text{As the wavelength increase the frequency decreases as it takes longer time to pass two consecutive peaks}$
- $\text{The equation thaat relate wavelength and fequence for electromagnetic waves is: } \lambda v = c \text{ where } \lambda \ v \ c  \text{ are the wavelength, frequency and speed of light respectively}$

- $\text{Wavelengths are measured in units of meters, and we obtain a range of wavelengths known as the electromagnetic spectrum}$
- $\text{The human eye is an electromagnetic sensor, which contain many receptors sensitive to radiation in certain wavelengths ranging betweeen 400nm and 700nm known as the visible light spectrum}$
- $\text{Different wavelengths absorbed by the eye are percieved as different colors}$
$\text{Colors:= Electromagnetic waves of different wavelengths in the spectrum of the visible light}$
