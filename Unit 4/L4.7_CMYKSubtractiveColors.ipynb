## 🎯 Overview

CMYK is a **subtractive color model** used primarily in **color printing**. Unlike RGB (additive, used for screens), CMYK starts with white (paper) and subtracts light by adding ink. Its primary inks are:

- **Cyan (C)**
- **Magenta (M)**
- **Yellow (Y)**
- **Black (K, “Key”)**

Each absorbs a specific range of wavelengths:

| Ink Color | Absorbs | Appears As |
|-----------|---------|------------|
| Cyan      | Red     | Green + Blue |
| Magenta   | Green   | Red + Blue |
| Yellow    | Blue    | Red + Green |
| Black (K) | All     | Black |

### Converting from CYM to RGB to determine color
 - $\text{All values in the CYM are in the range [0,1], representing the percentage of its color used}$
 - $\text{To convert we do the following: }$
  $(R = 1 - C, \ G = 1 - M, \ B = 1 - Y)$
- $\text{With this we can determine the colour by the more common ratios of color combinations}$

#### Example
$\text{Q. Which two colors are represented by the following triplets in CYM color model?}$
1. (1.0, 0.5, 0.5) 
2. (0.5, 1.0, 0.5)

a. 1. Redish Brown    2. Greenish Brown

b. 1. Purple          2. Greenish Blue

c. 1. Greenish Blue   2. Purple

d. 1. Greenish brown  2. Redish brown

#### Answer: D

(1.0, 0.5, 0.5) -> (0, 0.5, 0.5) = Green + Blue = Greenish Blue

(0.5, 1.0, 0.5) -> (0.5, 0, 0.5) = Red + Blue = Purple

## 🟡 Halftoning / Screening

This technique simulates **continuous tones** using **tiny colored dots**.

- Dots of **varying size, density, and spacing** give the illusion of gradients.
- **More ink (larger dots)** = darker color; **less ink (smaller dots)** = lighter color.
- This is how printers simulate shades even with **only four ink colors**.

> ✅ From a distance, our eyes blend these discrete dots into smooth tones – a perceptual trick based on **optical mixing** and **metamerism**.

## 🎨 Why Use CMYK Instead of RGB?

- Printers **cannot modulate dot brightness**, only **dot coverage**.
- CMYK replaces **light emission** with **light absorption**.
- In digital displays, pixels emit **different intensities** of red, green, and blue light (additive).
- In CMYK, paper starts **white**, and color is created by **removing wavelengths** using inks (subtractive).

## 🖨️ The K in CMYK: Black Ink

### Why is black ink needed?

- **CMY combined ideally produces black**, but in reality, it results in a **muddy dark brown** due to ink impurities.
- Enter the **K channel**:
  - Adds **sharp contrast** and **depth**.
  - Reduces ink costs (CMY inks are expensive).
  - Prevents **oversaturation and long drying times** in dark areas.

### Gray Component Replacement (GCR)

- Overlapping CMY produces neutral grays.
- GCR replaces that with black ink for better quality and less ink usage.

---

## 🧱 Dot Mechanics in CMYK Printing

- A printed "pixel" (dot cell) **can’t have partial color** – it's either **covered by ink or not**.
- Each CMYK channel has a **screen pattern** (a halftone grid), usually at a different **angle** to avoid moiré patterns.
- Standard screen angles:
  - Cyan: ~15°
  - Magenta: ~75°
  - Yellow: ~0° (least visible)
  - Black: ~45°

## 🧪 Practical Example: Halftoning

Suppose we want to print 50% gray:

- In RGB: each color is at 128/255 (medium intensity).
- In CMYK: dot patterns are placed so **half the area is covered by black dots**.
- In offset printing, this appears gray to the eye.
