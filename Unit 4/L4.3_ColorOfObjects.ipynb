### $\text{Color of Objects: Emission and Reflection}$

#### $\text{Light and Object Interaction}$
- $\text{Light waves can either penetrate objects, be absorbed, or be reflected.}$
- $\text{Each object absorbs and reflects electromagnetic wavelengths selectively.}$
- $\text{The absorption or reflection depends on the wavelength.}$
- $\text{Every object has a specific spectrum of wavelengths that it absorbs and reflects.}$

#### $\text{Color Perception}$
- $\text{The color of an object, as we see it, is determined by the visible light that is reflected into our eyes.}$
- $\text{Each electromagnetic wave with a different wavelength in the visible spectrum is perceived by the brain as a different color.}$
- $\text{Although each wavelength can correspond to a single perceived color, most colors are not just one wavelength.}$
- $\text{In fact, many colors are composed of several wavelengths.}$
- $\text{Some colors cannot be represented by a single wavelength—such as brown or pink.}$

- $\text{These are called Non-spectral colors.}$



#### $\text{Laser Light}$
- $\text{Laser stands for Light Amplification by Stimulated Emission of Radiation.}$
- $\text{It produces concentrated and focused light through a process of stimulated emission.}$
- $\text{Lasers emit light in a very narrow and focused wavelength range.}$
- $\text{A laser is its own light source, so it does not require external illumination.}$

#### $\text{Light Sources vs. Non-Light Sources}$
- $\text{Every object we see, such as a tomato or a wall, needs to be illuminated to be seen.}$
- $\text{If there's no light source, like in the dark, the object will appear black.}$
- $\text{In contrast, a laser is visible without external illumination.}$
- $\text{Objects that emit visible light are considered light sources (e.g., the sun, a light bulb, or a glowing LED).}$

#### $\text{Light Spectra and Perceived Color}$
- $\text{Most light sources emit a spectrum of wavelengths rather than just one.}$
- $\text{Walls illuminated by different spectra appear to us in different colors.}$
- $\text{White light sources (like the sun) are considered colorless but actually contain all visible wavelengths.}$

#### $\text{Dispersion – Separation of Light}$
- $\text{Light with different wavelengths has different phase velocities when passing through various materials.}$
- $\text{This causes the light to refract (bend) at different angles at material boundaries (e.g., air to glass).}$
- $\text{This phenomenon splits the light based on wavelength—what we perceive as different colors.}$

#### $\text{Prism (מנסרה)}$
- $\text{A prism is an optical tool that uses refraction to separate light into its component wavelengths.}$
- $\text{The operation is based on the fact that the refractive index for glass is higher than that of air.}$
- $\text{Most prisms are made from materials that show dispersion—meaning different wavelengths are refracted differently.}$

### $\text{Summary: Two Ways Objects Appear Colored}$

1. $\text{Objects that are light sources (e.g., lasers) emit radiation with a certain wavelength perceived as color.}$
2. $\text{Objects that are not light sources absorb part of the incident light and reflect the rest—this reflected light is perceived as color.}$

---

### $\text{Example – What Happens If We Shine Blue Light on a Tomato?}$
- $\text{Blue light does not contain red wavelengths.}$
- $\text{Therefore, the tomato absorbs most of the blue light and reflects nothing in red.}$
- $\text{To us, the tomato will appear black.}$
