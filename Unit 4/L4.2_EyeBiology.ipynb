## $\text{The Retina and Photoreceptors}$

$\text{The retina is constructed as a collection of specialized cells called photoreceptors – these are cells that respond to light stimuli and encode it as electrical signals that pass through the nervous system to the brain.}$

---

### $\text{There are 4 types of photoreceptor cells in the retinal layer of the eye:}$

#### 1. $\text{Rod Cells (תאי קנה)}$

- $\text{Rod cells are highly sensitive to the amount and intensity of light, rather than specific wavelengths.}$
- $\text{Therefore, they are not used for distinguishing between different colors.}$
- $\text{They are essential for night vision and function well in dim lighting conditions.}$

$\textbf{➤ Expansion:}$  
$\text{Rod cells are located mostly in the peripheral retina and help detect motion and shapes in darkness, but they do not provide color or fine detail.}$

#### 2. $\text{Three Types of Cone Cells (תאי חרוט)}$

- $\text{Cone cells are less sensitive to light than rod cells, but more effective in bright light.}$
- $\text{They are responsible for color vision and visual acuity (sharp vision).}$
- $\text{There are three types of cone cells, each sensitive to different ranges of light wavelengths in the visible spectrum.}$

### $\text{Each Cone Cell Type Responds to a Different Wavelength Range}$

- $\text{S-cones (Short wavelength): Most sensitive to blue light (כחול)}$
- $\text{M-cones (Medium wavelength): Most sensitive to green light (ירוק)}$
- $\text{L-cones (Long wavelength): Most sensitive to red light (אדום)}$

$\textbf{➤ Expansion:}$  
$\text{Color vision arises from the combined stimulation of these cone types. For example, yellow light stimulates both red- and green-sensitive cones.}$  
$\text{This forms the basis for trichromatic (three-color) color vision in humans.}$

### $\text{Night Vision is Nearly Monochromatic}$

$\text{Because cone cells require high light levels to function, we rely mostly on rod cells at night.}$  
$\text{Rod cells do not detect color, so night vision is essentially black and white.}$

$\textbf{➤ Expansion:}$  
$\text{At night, our surroundings appear grayish due to the dominance of rod cells. Cone cells are inactive in low light, causing the loss of color perception.}$

---

### $\text{Clarification: Cone Cells Respond to a Range of Wavelengths}$

$\text{Each type of cone is not tuned to a single exact wavelength, but rather a spectrum.}$  
$\text{The names “blue”, “green”, and “red” are approximations of the color ranges to which each type is most sensitive.}$
