### 🎨  CIE XYZ Color Space: Overview

$\text{The CIE 1931 XYZ color space was designed to mathematically model human color perception.} \\ \text{It’s based on color matching experiments that define how we perceive color through combinations of three “imaginary primaries” (X, Y, and Z).}$

---

#### Key Features:
•	X is not red, Y is not green, and Z is not blue — they’re imaginary primaries chosen to simplify color matching math. <br>
•	Y corresponds to luminance (brightness). <br>
•	All visible colors can be represented as combinations of X, Y, and Z.

#### $\text{1. Color Representation with } X, Y, Z$
- $\text{Any color can be represented by a triplet of tristimulus values } (X, Y, Z).$
- $\text{Here, }Y\text{ specifically represents luminance (the perceived brightness) of the color.}$
- $\text{Equal amounts of }X, Y, Z\text{ correspond to the color white.}$

#### $\text{2. Positive Combinations and Visual Feasibility}$
- $\text{Any physically possible visible color can be formed by a positive linear combination of }X, Y, Z.$
- $\text{Only those combinations lie within the “color gamut” of the display device or medium.}$
- $\text{Colors that cannot be represented by real light wavelengths alone (like purples or magentas) are “non-spectral”; but they are still representable within CIE XYZ.}$

---

#### $\text{3. Chromaticity Diagram (x,y)}$
- $\text{Chromaticity coordinates }(x, y)\text{ are derived from }X, Y, Z\text{:}$
  $$
    x = \frac{X}{X + Y + Z},\quad
    y = \frac{Y}{X + Y + Z}
  $$
- $\text{These define the color hue and saturation (chrominance), independent of luminance.}$
- $\text{The third component } z = 1 - x - y\text{ is implied.}$

### Example 

$\text{Suppose you have color with XYZ values: } \\  \\ X = 41.24, \ Y = 21.26, \ Z = 1.93$

$\text{To isolate color (Hue + Saturation) from brightness we define Chromaticty coodinates:}$

$ x = \frac{X}{X+Y+Z}, \ y = \frac{Y}{X+Y+Z}$

$ x = \frac{41.24}{41.24+21.26+1.93} = 0.64, \ y = \frac{21.26}{41.24+21.26+1.93}= 0.33$

$\text{This presents color hue and saturation, independent of brightness}$

---


#### $\text{4. Luminance and Perceived Brightness}$
- $\text{The perceived brightness }B\text{ is given by:}$
  $$
    B = X + Y + Z
  $$
- $\text{Alternatively, a color can be described as }(x, y, B)\text{ in xyY space, where }Y = B.$
- $\text{This separation allows us to treat color (hue/saturation) separately from lightness.}$

#### $\text{5. White Point and Centroid in xy Space}$
- $\text{The white point is defined where }x = y = z \text{ (i.e., X = Y = Z).}$
- $\text{It represents “neutral white” in the chromaticity diagram.}$
- $\text{Colors plotted farther from the white point become more saturated (pure) ; closer colors are desaturated.}$

#### $\text{6. Purity and Color Saturation}$
- $\text{Purity indicates how far a chromaticity point lies from the white point towards the edge of the color gamut.}$
- $\text{Higher purity = more saturated (vivid) color.}$
- $\text{Lower purity = closer to white (pastel, desaturated).}$

---

### $\text{Summary: Why CIE XYZ / xyY Matters}$
- $\text{Provides a standardized, perceptually uniform way to represent all visible colors.}$
- $\text{Separates brightness (Y / luminance) from color (x, y / chrominance).}$
- $\text{Allows devices like monitors and printers to map and reproduce colors accurately.}$

### HSB/HSV Color Model

$HSB = \text{(Hue }\in [0, 360], \text{Saturation } \in [0,1],\text{Brightness } \in [0,1]\text{)}$

$HSV = \text{(Hue }\in [0, 360], \text{Saturation } \in [0,1],\text{Value } \in [0,1]\text{)}$

#### Conversion to RGB

1. $\text{Handle the Greyscale}$
- $\text{If saturation = 0 then R = G = B = Brightness/Value}$

2. $\text{Hue Sector Calculation}$
- $\text{Divide hue into (6 sectors) sectors of the color wheel (60 degress per sector)}$
```python
H' = H / 60      # sector index (float)
i = floor(H)    # integer part: 0 to 5
f = H' - i       # fractional part

p = V × (1 - S)
q = V × (1 - f × S)
t = V × (1 - (1 - f) × S)

if i == 0: R, G, B = V, t, p
if i == 1: R, G, B = q, V, p
if i == 2: R, G, B = p, V, t
if i == 3: R, G, B = p, q, V
if i == 4: R, G, B = t, p, V
if i == 5: R, G, B = V, p, q
```
#### Example  HSV = (30, 1, 1)
$\text{H' = 30/60 = 0.5}$

$\text{i = 0}$

$\text{f = 0.5}$

$\text{p = 1 } \cdot \text{ (1-1) = 0}$

$\text{q = 1 } \cdot \text{(1-0.5) = 0.5}$

$\text{t = 1 } \cdot \text{(1 - (1 -0.5)) = 0.5}$

$\text{i = 0 so:}$
$\text{RGB = (1, 0.5, 0)}$


🟠 $\text{Output RGB = (1, 0.5, 0) → Orange}$
