Objective: Learning the Marching Cubes Algorithm which will allow us to draw implicit surfaces thereby overcoming the difficulty of Enumeration on implicit Objects.

### Marching Cube Algorithm

Question: How can we draw iso-surfaces of implicit funciton in our previous examples?  
Answer: We'd need to search the whole domain of the function for $Zero \ Crossings$.

#### Example  
Consider a 2D function: $F(x,y)$:  
If we know that for some two points $(x_1, y_1) \ (x_2, y_2)$ we have $F(x_1, y_1) > 0$ and $F(x_2, y_2) < 0$ (Assuming Continuity)
Then we know for sure that exists a point $(x_0, y_0)$ such that $F(x_0, y_0) = 0$ (Assuming we create an equation of a line connecting these two points in which this would hold via IVT)

### 2D Zero Crossing Algorithm

To sample the zero-crossings on the whole domain of the iso-contour, we will divide a perticular plane into a grid shape, each "corner/intersection" on the grid will be considered a point of classification, by evaluating that point on the function of the surface (plane). If two neighboring points have opposite signs then there exists at least 1 zero-crossing between those neighbors.  

Our next step will be to connect all the zero-crossings (linear time). the result provides an approximation of the object (the shape of the image).

### 3D Zero Crossing Algorithm

The same concept could be applied to 3D, we create a collection of cubes as our point of classification (on the corners of the cubes). 
Based on the arangements on the signes of each corner of the cubes we can approximate the internal shapes of object inside each cubes.
