#### Degrees of Freedom in Transfromations
##### objective: 
1. Discuss different types of Affine Transformations
2. Introduce the notion of Degrees of Freedom of a transformation

$\text{Degree Of Freedom : = The minimum number of independent parameters needed to uniquely define a Transformation}$


$$
\begin{array}{|c|c|}
\hline
\textbf{Purpose} & \textbf{Why DoF Matters} \\
\hline
\text{🔍 Solving for} \\ \text{Transformations} & 
\text{DoF tells you how many point correspondences are needed} \\
& \text{to uniquely solve a transformation.} \\
& \text{E.g., 2D affine has 6 DoF → 3 point pairs (2 constraints each).} \\
\hline
\text{🧠 Understanding} \\ \text{Flexibility} & 
\text{DoF reflects how constrained or flexible a transformation is.} \\
& \text{Higher DoF = more complex possible spatial changes.} \\
\hline
\text{🧮 Parameter} \\ \text{Optimization} & 
\text{In vision/graphics, DoF tells how many parameters} \\
& \text{must be optimized to estimate a transformation.} \\
\hline
\text{📐 Preservation} \\ \text{Properties} & 
\text{Fewer DoF usually means more geometric preservation.} \\
& \text{E.g., rigid (3 DoF) preserves distance/angles;} \\
& \text{affine (6 DoF) does not.} \\
\hline
\end{array}
$$




#### Special Types of Affine Transformations

$$
\begin{array}{|c|c|c|c|c|}
\hline
\textbf{Transformation} & \textbf{DoF} & \textbf{Preserves} & \textbf{Points Needed} & \textbf{Notes} \\
\hline
\text{Translation} & 2 & \text{Angles, Parallelism} & 1 & 
\begin{array}{l}
(x, y) \mapsto (x + v_x, y + v_y) \\
\text{Only } v_x, v_y \text{ required}
\end{array} \\
\hline
\text{Rotation} & 1 & \text{Angles} & 1 & 
\begin{array}{l}
\text{Rotation by angle } \theta \\
\text{1 point gives 2 constraints → overconstrained}
\end{array} \\
\hline
\text{Rigid (Euclidean)} & 3 & \text{Angles, Distances} & 2 & 
\begin{array}{l}
\text{Rotation + Translation} \\
\text{Preserves shape and size} \\
\text{Closed under composition}
\end{array} \\
\hline
\text{Similarity} & 4 & \text{Angles, Parallelism} & 2 & 
\begin{array}{l}
\text{Rigid + Uniform Scaling} \\
\text{Preserves shapes up to scale}
\end{array} \\
\hline
\text{Affine} & 6 & \text{Lines, Ratios, Parallelism} & 3 & 
\begin{array}{l}
\text{Linear + Translation} \\
\text{Doesn't preserve angles or distances}
\end{array} \\
\hline
\text{Projective (Homography)} & 8 & \text{Lines} & 4 & 
\begin{array}{l}
\text{General invertible 3x3 matrix} \\
\text{Doesn't preserve parallelism, distances, angles, or origin} \\
\text{Maps lines to lines}
\end{array} \\
\hline
\end{array}
$$
