#### Special Types of Affine Transformations

##### Translation:
Q. How many degrees of freedom does translation have in 2D?  
A. we only need $v_x$ and $v_y$, from this we conclude that it has 2 degrees of freedom.
- Note: If we are told that the mapping T is a translation then we only need a single point, to find the translation.

##### Rotation:
Q. How many degrees of freedome does translation have in 2D?  
A. **1** the angle $\theta$  
Q. What if we have the mapping of one point in 2D and we look for a rotation that maps the source to the target point can we solve for $\theta$  
A. We saif a single mapping provide **2 Constraints** which is more than the DoF of a Rotation so we are **Over Constriant**.

##### Rigid Transformation
- This is the combination of translations and Rotation. 
- It has 3 DoF.
- **Preserves Distances and angles**
- They are closed uner composition

##### Similarity transformation
- This is the combination of translations and Rotation, and **Uniform-Scaling**
- It has 4 DoF.
- They Preseve Angles.
- They are also Closed under Composition

##### Affine Tranformation

- they have 6 DoF
- **Doesn't preserve Distances or angles**

##### Projective Transformation
- An invertible 3x3 Matrix
- It has 8 DoF
- We need mappings of 4 points to solve/ define the transformation.
- Closed under composition.
- Map lines to Lines.
- **Doesn't preserve: Parallelism, distances, angle, origin point.**
