$\text{Translation transformation, maps the point } (0,0)  \text{ to a different point in 2D space.}$

$\text{So we return to Affine Mapping + Translations}$

$\text{As mentioned in Lesson 6.1 Affine Mapping allows the transformation from one affine space to another (like Local to Global Coordinate systems).}$

$\text{Reminder: To define an affine space we need:}$
1. $\text{Define an origin point}$
2. $\text{A basis of vectors of the space}$

$\text{The Linear mapping from one Affine to another is defined by:}$
$$\vec{\begin{pmatrix} x \\ y \end{pmatrix}} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + \begin{pmatrix} \color{green}e_{11} & \color{yellow}e_{12} \\ \color{green}e_{21} & \color{yellow}e_{22} \end{pmatrix} \cdot \begin{pmatrix} u \\ v \end{pmatrix}$$

$$ = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + u\vec{\begin{pmatrix} \color{green}e_{11} \\ \color{green}e_{21} \end{pmatrix}} + v\vec{\begin{pmatrix} \color{yellow}e_{12} \\ \color{yellow}e_{22} \end{pmatrix}}$$

$$ = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + u\vec{\color{green}e_1} + v\vec{\color{yellow}e_2} $$

$\text{The above describes the mapping (origines) from one space to another by the translation component }  \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} $

$\text{The linear part of the mapping (the matrix) defines the change of base of the axes: } \begin{pmatrix} \color{green}e_{11} & \color{yellow}e_{12} \\ \color{green}e_{21} & \color{yellow}e_{22} \end{pmatrix} $

#### Complex Affine + Translation Transformations: 
$\text{Let } 0 \ne q \in \  \R^2, \  \theta \in \R$
$\text{Suppose we want to rotate a general point around q then we must:}$
1. $\text{Translate q to the origin:} -v = - (q-0) = (-q_x, -q_y)$
2. $\text{Rotate about the origin: } \theta$
3. $\text{Translate back to q: } v = (q-0) = (q_x, q_y)$

$$ p_{new} = R_{\theta} \cdot (p-v) + v $$

## $$\text{Affine Properties}$$ 

$$
\begin{aligned}
\textbf{Property} &\quad \textbf{Description} \\
\text{Line Preservation} &\quad \text{Straight lines remain straight} \\
\text{Parallelism Preservation} &\quad \text{Parallel lines remain parallel} \\
\text{Ratio Preservation} &\quad \text{Preserves ratios along lines (e.g., midpoints)} \\
\text{Distance Preservation} &\quad \textbf{Not preserved} \quad (\text{e.g., scaling distorts length}) \\
\text{Angle Preservation} &\quad \textbf{Not preserved} \quad (\text{e.g., shear distorts angles}) \\
\text{Origin Mapping (Linear)} &\quad \text{Origin maps to origin in linear transforms} \\
\text{Convexity} &\quad \text{Convex sets remain convex} \\
\text{Collinearity} &\quad \text{Collinear points stay collinear} \\
\text{Affine Combinations} &\quad \text{Affine combinations are preserved} \\
\text{Composability} &\quad \text{Composition of affine maps is affine} \\
\text{Invertibility} &\quad \text{Invertible if linear part has } \det A \ne 0 \\
\end{aligned}
$$
