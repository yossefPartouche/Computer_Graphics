#### Affine Mapping with Translation
##### Objective: 
1. We will learn about the transformation that doesn't keep the origin in its place - Translation

The translation transformation, maps the point $(0,0)$ to a different point in 2D space. So we return to Affine Mapping with Translations

As mentioned in Lesson 6.1 Affine Mapping allows the transformation from one affine space to another (like Local to Global Coordinate systems).

Translation operation maps:
- $x$ to $x + t_x$
- $y$ to $y + t_y$

Also viewed as adding a vector to a point which is another point (by Definition).

Q. What can we do with an affine mapping that cannot be done with a linear mapping?  
A. Suppose We wanted to roate p around some point q which isn't the origin. focusing on q, we'll:
- Translate q to the origin
- Apply the rotation by $\theta$ degrees.
- Translate origin back to q

#### General Properties of Affine Mapping
- Affine Mapping (thus also Linear Mapping ) DO NOT preserve distances after transformation, known as Isometry Transformation (e.g. Translation But not Scaling)
- Affine Mapping (thus also Linear Mapping ) DO NOT preserve angles after transformation, consdier the Sheer transformation.
- Affine Mapping preserve straight lines 
- Preserve Ratios
- Preseve Parallelism
