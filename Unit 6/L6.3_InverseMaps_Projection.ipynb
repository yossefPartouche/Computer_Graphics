### The projection Transformation

$\text{Projection = A linear transforamation P from a vector space to itself such that } P^2 ~=~ P$

$\text{This means if you apply P any number of time, you will still get the same result.}$

$$
P(P(\mathbf{v})) = P(\mathbf{v}) \quad \text{for all } \mathbf{v}
$$

#### Example:

$\text{Let \( P \) be the projection onto the x-axis in \( \mathbb{R}^2 \):}$

$$
P = \begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
$$

$\text{Let \( \mathbf{v} = \begin{bmatrix} x \\ y \end{bmatrix} \).  
Then:}$

$$
P \mathbf{v} = \begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
\begin{bmatrix}
x \\
y
\end{bmatrix}
=
\begin{bmatrix}
x \\
0
\end{bmatrix}
$$

$\text{Now apply \( P \) again:}$

$$
P(P \mathbf{v}) = P \left( \begin{bmatrix}
x \\
0
\end{bmatrix} \right)
=
\begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
\begin{bmatrix}
x \\
0
\end{bmatrix}
=
\begin{bmatrix}
x \\
0
\end{bmatrix}
= P \mathbf{v}
$$

✅ $\text{ This confirms that \( P^2 = P \), so \( P \) is a}$ **projection transformation**.


### Properties on Projection Matrix

- $\text{The Identity transformation } I_n \text{  is such a transformation}$

- $\text{If } M_p \text{ is a projection matrix and } M_p \neq I_n \text{ then } M_p \text{ is singular (not invertible).}$

- $\text{ Similarly, the geometric operation for non-Identity projections are also not invertible.}$

### A projection in 2D will reduce dimension and as such can project all the points in 2D space on to:

- A Line
- A point: This is a mapping all the space to the origin, represented by $0_n$ matrix

### Orthogonal Projection

- Any 2D point $(x,y)$ on the coordinate system can be project to $(x,0)$, the corresponding matrix transformation is:

$$
\begin{pmatrix}
1 & 0 \\
0 & 0
\end{pmatrix}
$$

##### Generalized Oblique Projection Matrix (onto a 2D line)

Let $\mathbf{u} = \begin{bmatrix} u_x \\ u_y \end{bmatrix} \in \mathbb{R}^2$ be a non-zero direction vector.  
Then the projection matrix $P$ that projects any vector onto the line spanned by $\mathbf{u}$ is given by:

$$
P = \frac{1}{u_x^2 + u_y^2}
\begin{bmatrix}
u_x^2 & u_x u_y \\
u_x u_y & u_y^2
\end{bmatrix}
$$
