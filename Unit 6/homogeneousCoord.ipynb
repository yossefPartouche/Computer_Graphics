### Homogeneous Coordinates
#### Objectives:
1. To discover a method that will enable us to apply affine mapping via matrix multiplications.

#### Homogeneous Properties of Point
1. To represent a point in n-Dimensions we use n+1 coordinates.
2. If the coordinates of a point are multiplied by a non-zero scalar then the resulting coordinates represent the same point.

Suppose we have a homogeneous coordinate representing a point $p = (x_1,x_2, \ ... \,x_{n+1})$
$\forall \alpha \in \R \ , \alpha \neq 0 (\alpha x_1,\alpha x_2, \ ...\ \alpha x_{n+1})$ all represent the same point.  
It customary to multiply all terms by $\frac{1}{x_{n+1}}$ to produce $(\frac{x_1}{x_{n+1}},\frac{x_2}{x_{n+1}}, \ ... \,1)$

#### Q. 
What does this mean geometrically?
#### A. 
All the point in 3D of the form $(Wx, Wy, W) \ \forall \ W$ lie on a line that passes through the origin when $W=0$ these point will all map to same point in 2D $(x,y)$. 
- This can also bee seen as the point $(x, y, 1)$ where they lie on the plane $z=1$ (this is also a projection in the dimension we desire).

#### Q. 
What is the advantage of using homogeneous coordinates?
#### A. 
We will look at how to represent linear transformations using matrices over homogenous coordinates and then extend to affine Transformations.

Over 2x2 Matrices we had the following form for Linear transformation:
\begin{pmatrix} x_{new} \\ y_{new} \end{pmatrix} =  \begin{pmatrix} a_{11} & a_{12} \\ a_{21} & a_{22} \end{pmatrix}\begin{pmatrix} x \\ y \end{pmatrix} 

Now we must increase the dimension (For it to be Homogeneous) and produce:
\begin{pmatrix} x_{new} \\ y_{new} \\ 1 \end{pmatrix} =  \begin{pmatrix} a_{11} & a_{12} & 0_{13} \\ a_{21} & a_{22} & 0_{23} \\ 0_{31} & 0_{32} & 1_{33} \end{pmatrix}\begin{pmatrix} x \\ y \\ 1\end{pmatrix}
We have solved the case of linear transformation over homogeous coordinate

### Translation.
We place the $x$ and $y$ component of the translation vector in the last column of the matrix: 
\begin{pmatrix} x_{new} \\ y_{new} \\ 1 \end{pmatrix} =  \begin{pmatrix} 1 & 0 & v_{x} \\ 0 & 1 & v_{y} \\ 0 & 0 & 1 \end{pmatrix}\begin{pmatrix} x \\ y \\ 1\end{pmatrix}

We bring both transformations together to produce as Affine Transformation Matrix:
\begin{pmatrix} x_{new} \\ y_{new} \\ 1 \end{pmatrix} =  \begin{pmatrix} a_{11} & a_{12} & v_x \\ a_{21} & a_{22} & v_{y} \\ 0 & 0 & 1 \end{pmatrix}\begin{pmatrix} x \\ y \\ 1\end{pmatrix}

Note that Affine Tranformation matrix is closed under Multiplication, as such we can combine Affine trasnformation matrix just as we did with Linear Transformation matrix.



### Roatating around arbituary point q
In lesson 6.4 we had to do this in 3 steps (Translate, Rotate, Inv Translate).
With Affine matrix we can do as follows:
1. Create Affine Matrix for Translation
 \begin{pmatrix} 1 & 0 & -q_x \\ 0 & 1 & -q_y \\ 0 & 0 & 1 \end{pmatrix}
2. Create Affine Matrix for Roation
\begin{pmatrix} cos(\theta) & -sin(\theta) & 0 \\ sin(\theta) & cos(\theta) & 0 \\ 0 & 0 & 1 \end{pmatrix}
3. Create the inv Translation
\begin{pmatrix} 1 & 0 & q_x \\ 0 & 1 & q_y \\ 0 & 0 & 1 \end{pmatrix}

Multiply in the order $M_3•M_2•M_1$ to produce the combined Affine Transformation around a point q.

#### Reflecting about aribtuary line

Given a point 

$$
\mathbf{p} = \begin{bmatrix} p_x \\ p_y \\ 1 \end{bmatrix},
$$ 

the following sequence of transformations is applied:

1. **Translate to the origin** by vector $ \vec{c} = \begin{bmatrix} \ c_x \\ c_y \end{bmatrix}$  
$$
T_{-c} = 
\begin{bmatrix}
1 & 0 & -c_x \\
0 & 1 & -c_y \\
0 & 0 & 1
\end{bmatrix}
$$

2. Rotate by $ \theta $ 
$$
R_\theta = 
\begin{bmatrix}
\cos(\theta) & -\sin(\theta) & 0 \\
\sin(\theta) & \cos(\theta) & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

3. **Reflect about the y-axis**  
$$
\text{Reflect}_y = 
\begin{bmatrix}
-1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

4. Rotate back by $ - \theta $  
$$
R_{-\theta} = 
\begin{bmatrix}
\cos(-\theta) & -\sin(-\theta) & 0 \\
\sin(-\theta) & \cos(-\theta) & 0 \\
0 & 0 & 1
\end{bmatrix}
=
\begin{bmatrix}
\cos(\theta) & \sin(\theta) & 0 \\
-\sin(\theta) & \cos(\theta) & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

5. **Translate back** by vector $ \vec{c} = \begin{bmatrix} \ c_x \\ c_y \end{bmatrix}$  

$$
T_c = 
\begin{bmatrix}
1 & 0 & c_x \\
0 & 1 & c_y \\
0 & 0 & 1
\end{bmatrix}
$$

---

### Final Transformation:
$$
\mathbf{p'} = T_c \cdot R_{-\theta} \cdot \text{Reflect}_y \cdot R_\theta \cdot T_{-c} \cdot \mathbf{p}
$$

This composition applies the transformations in reverse order (due to matrix multiplication semantics).
