### Homogeneous Coordinates: Key Properties and Transformations
#### Homogeneous Properties of Point
1. To represent a point in $n$-dimensions we use $(n+1)$ coordinates.
2. Homogeneous coordinates are **scale-invariant**: multiplying all components by a nonzero scalar does **not** change the point being represented.


$\text{Let a point in homogeneous coordinates be:}$

$$
\mathbf{p} = (x_1, x_2, \dots, x_{n+1})$$

$$\text{ Then for any } \alpha \in \mathbb{R}, \alpha \ne 0:

\alpha \cdot \mathbf{p} = (\alpha x_1, \alpha x_2, \dots, \alpha x_{n+1})
$$

$\text{represents the same point. It is customary to normalize the point by dividing all components by } x_{n+1} \text{ assuming } x_{n+1} \ne 0\text{, resulting in:}$

$$
\left( \frac{x_1}{x_{n+1}}, \frac{x_2}{x_{n+1}}, \dots, 1 \right)
$$
#### Q. What does this mean geometrically? 

$\text{All homogeneous points of the form:}$
$$
(Wx, Wy, W), \quad \forall W \ne 0
$$

- $\text{Lie along a line in 3D space that passes through the origin and intersect the plane}$  $z = 1$ at the point $(x, y, 1)$.
- $\text{When projected back to 2D (by dropping the last component), all these points map to the same point }$ $(x, y)$
- $\text{This can be visualized as a projection from 3D onto the plane } z = 1, \text{ which is commonly used in computer graphics and geometric modeling.}$

### Extension of Matrix for homogeneous coordinates

$\text{We will look at how to represent linear transformations using matrices over homogenous coordinates and then extend to affine Transformations.}$

$\text{To turn a linear transformation to Homogenous we must had a dimension on each side of the matrix}$

$$
\begin{pmatrix}
x_{\text{new}} \\
y_{\text{new}}
\end{pmatrix}
=
\begin{pmatrix}
a_{11} & a_{12} \\
a_{21} & a_{22}
\end{pmatrix}
\begin{pmatrix}
x \\
y
\end{pmatrix}
\quad
\xRightarrow{\text{homogenize}}
\quad
\begin{pmatrix}
x_{\text{new}} \\
y_{\text{new}} \\
1
\end{pmatrix}
=
\begin{pmatrix}
a_{11} & a_{12} & 0_{13} \\
a_{21} & a_{22} & 0_{23} \\
0_{31} & 0_{32} & 1_{33}
\end{pmatrix}
\begin{pmatrix}
x \\
y \\
1
\end{pmatrix}
$$

### Roatating around arbituary point q
In lesson 6.4 we had to do this in 3 steps (Translate, Rotate, Inv Translate).
With Affine matrix we can do as follows:

$\text{Let } 0 \ne q \in \  \R^2, \  \theta \in \R$
$\text{Suppose we want to rotate a general point around q then we must:}$
1. $\text{Translate q to the origin:} -v = - (q-0) = (-q_x, -q_y)$
2. $\text{Rotate about the origin: } \theta$
3. $\text{Translate back to q: } v = (q-0) = (q_x, q_y)$

$$ p_{new} = R_{\theta} \cdot (p-v) + v $$

1. $\text{Create Affine Matrix for Translation}$

\begin{pmatrix} 1 & 0 & -q_x \\ 0 & 1 & -q_y \\ 0 & 0 & 1 \end{pmatrix}

2. $\text{Create Affine Matrix for Roation}$

\begin{pmatrix} cos(\theta) & -sin(\theta) & 0 \\ sin(\theta) & cos(\theta) & 0 \\ 0 & 0 & 1 \end{pmatrix}

3. $\text{Create the inv Translation}$

\begin{pmatrix} 1 & 0 & q_x \\ 0 & 1 & q_y \\ 0 & 0 & 1 \end{pmatrix}

Multiply in the order $M_3•M_2•M_1$ to produce the combined Affine Transformation around a point q.

#### Reflecting about aribtuary line

Given a point 

$$
\mathbf{p} = \begin{bmatrix} p_x \\ p_y \\ 1 \end{bmatrix},
$$ 

the following sequence of transformations is applied:

1. **Translate to the origin** by vector (The amount + direction required to arrive at the origin) $ \vec{c} = \begin{bmatrix} \ c_x \\ c_y \end{bmatrix}$  
$$
T_{-c} = 
\begin{bmatrix}
1 & 0 & -c_x \\
0 & 1 & -c_y \\
0 & 0 & 1
\end{bmatrix}
$$

2. Rotate by $ \theta $ the angle required for the line to be at an axes
$$
R_\theta = 
\begin{bmatrix}
\cos(\theta) & -\sin(\theta) & 0 \\
\sin(\theta) & \cos(\theta) & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

3. **Reflect about the y-axis or x-axis**  
$$
\text{Reflect}_y = 
\begin{bmatrix}
-1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

4. Rotate back by $ - \theta $  
$$
R_{-\theta} = 
\begin{bmatrix}
\cos(-\theta) & -\sin(-\theta) & 0 \\
\sin(-\theta) & \cos(-\theta) & 0 \\
0 & 0 & 1
\end{bmatrix}
=
\begin{bmatrix}
\cos(\theta) & \sin(\theta) & 0 \\
-\sin(\theta) & \cos(\theta) & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

5. **Translate back** by vector $ \vec{c} = \begin{bmatrix} \ c_x \\ c_y \end{bmatrix}$  

$$
T_c = 
\begin{bmatrix}
1 & 0 & c_x \\
0 & 1 & c_y \\
0 & 0 & 1
\end{bmatrix}
$$

---

### Final Transformation:
$$
\mathbf{p'} = T_c \cdot R_{-\theta} \cdot \text{Reflect}_y \cdot R_\theta \cdot T_{-c} \cdot \mathbf{p}
$$
