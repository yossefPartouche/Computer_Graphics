### Mapping Coordinate Systems 

#### Q. Why do we need to learn about Transformations?

#### A. 2D Example: Digital Printing
- $\text{Letter are described using mathematical curves (boundary curve), which define their respective shape}$ 
- $\text{The curves are define in a}$ **local coordinate system** $\text{ that represent the shape of each letter}$
- $\text{When printed the white space inside the letter is filled in black ink.}$ 
- $\text{We still need to place that shape in the correct position on the page.}$

- $\text{This is an example mapping a letter from a local coordinate system (of the letter) to a global coordinate system (Text).}$
- $\text{The page (global) coordinate system is usually upside down.}$

### Mathematically Defining a Mapping


$\text{we'll define the local coordinate system with Axis u Equivelant to x axis and Axis v Equivelant to y}$
- $\text{For each letter in this system is deined in the bounding box (i.e. in the range) } [0, u_1]$ x $[0, v_1]$. 
- $\text{Suppose we know the global position of this letter } (x_0, y_0)$. 
- $\text{The size of the letter (in local) must be adjusted in the global page.}$  
- $\text{ For now assume we know this range of values, in which we deduced the size to be between } x_0, \ x_1 \text{ and } y_0, \ y_1$:
- $\text{The scale process is:}$ $$ S_x = \frac{x_1-x_0}{u_1-0}$$ $$ S_y = \frac{y_1-y_0}{v_1-0}$$ 

$\text{Note: In the denominator it generalises to the form } u_1 -u_0 \text{ and }  v_1 -v_0 \text{ for and x and y respectively.}$

### Mapping from Local to Global requires:

1. $\text{Translation Correct starting position of the letter in the page } (u_0, v_0) \text{ to } (x_0, y_0)$
2. $\text{Scaling: The width of the letter, } (u_0, u_1) \text{ to } (x_0, x_1)$
3. $\text{Scaling: The hieght of the letter, } (v_0, v_1) \text{ to } (y_0, y_1)$

$\text{Summarised with: }$

$$x = x_0 + uS_x$$ 

$$y = y_0 + vS_y$$ 


$\text{The Transformation can be done via vector operations (When we say vector we also mean matricies): }$
\begin{align*}
\begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + \begin{pmatrix} S_x & 0 \\ 0 & S_y \end{pmatrix} \begin{pmatrix} u \\ v \end{pmatrix}
\end{align*}  

\begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + \begin{pmatrix} e_{11} & e_{12} \\ e_{21} & e_{22} \end{pmatrix}\begin{pmatrix} u \\ v \end{pmatrix} 

\begin{align*}
= \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + \mathbf{A} \begin{pmatrix} u \\ v \end{pmatrix}
\end{align*}

Note: If the the two vector $u$ and $v$ are not orthonormal to another they it will have some distortional affect on the letter (italix font for example).
