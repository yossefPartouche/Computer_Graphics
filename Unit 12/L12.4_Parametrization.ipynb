### Parametrization

$\text{Paramerization := Defining values (u,v) of the vertices of all triangles}$
- $\text{Texture, is an image; it's defined in 2D space (UV-Space), where u and v are the texture coordinates}$
- $\text{To be able to map the texture objects on the 3D objects, we need to map the triangles and faces of the 3D object onto the 2D space of the texture}$
    - $\text{Assuming we know this mapping, then we can define the (u,v) coordinates of every vertex on the mesh and then we can use these (u,v)} \\ \text{values during rasterization to obtain the colors of eery pixel inside every face, by interpolating the (u,v) values of every vertices.}$
- $\text{Mapping a 3D surface object onto a 2D domain isn't easy}$

#### Example

$\text{Consider mapping a Sphere on to a flat 3D domain}$
$\text{First note that a sphere and 2D domain don't have the same topology so not every point of the sphere is mappabble}$


<img src="image_U12/Screenshot 2025-06-28 at 21.20.33.png" height="400" width="400"/>

$\text{Stereographic projection := A one-to-one mapping of all points on the sphere (apart from a single point, the projection point) to points on the plane}$
$\text{When picking a projection point, on the sphere, we then send a ray from the projection point, to any point p on the sphere and map p at the intersection point of the ray and the plane}$

 <img src="image_U12/Screenshot 2025-06-28 at 21.36.11.png" height="400" width="400"/>


$\text{The neat thing about this is that it's practically a bijection mapping, so we can map from the plane back to the sphere}$
$\text{it also preserves angles, But it doesn't preserve distances}$

- $\text{Sometimes having a projective mapping that is an isometry is very important for example the world map}$

#### Parametrization problem in CG

$\text{Parametrization problem := To find a bjective mapping from a surface mesh M (surface of 3D object), to a 2D domain}$
$\text{It's called parametrization since, the moment we find this mapping, we can define a parametric representation of the object} \\ \text{since we can use the inverse of the mapping, from the 2D domain to 3D surface rep}$
