#### Texture Mapping
#### Objective: 
- $\text{Method to break away from vertices and face constraints on object representation to define the color of the object}$

#### Limitation of using Geometry to define Color

1.  $\text{Complicates the object's represenation}$
2.  $\text{It'l only be a crude approximation of the true representation}$

#### Solution
- $\text{Texture Mapping := Breals the link between geometric details and color details of 3D objecs by mapping textures (images) onto the geometry}$
- $\text{Texture Mapping, allows us to add as many color and images to the object, and map 3D images to the object whilst keeping the geometry simple}$
- $\text{Doing this, will enable greater color detail and saves rendering time by using simple 3D representation.}$

#### Examples: 
1. $\text{We can map an entire city using boxes and then have image of the outside of the building representing the color and texture}$
2. $\text{We can model a person, using the basic mesh and then have the dace, the hair, skin color, and clothes to cover the surface}$

### How is the Mapping done:

 $\text{Reminder: } \\
 \text{1. In Gourad Shading, we use the vertices of the triangle to apply interpolation and then determine the color} \\
 \text{2. In Phone Shading, we interpolates the normals to calculate the color of each pixel} $

 - $\text{In Texture mapping we obtain the color in a different way. We fetch the color from a texture, (Like a lookup table) given an index you get a color}$$
 - $\text{Since a texture, is a 2D image, the index must be 2D as well}$
 - $\text{We use a u, v coordinate system to define the index of the color of the texture image.}$
 - $\text{We can then use the points of the vertices of a triangle, we can interpolate these values in the triangle, to get the value of any pixel in rasterisation (Barycentric, Bilinear Interpolation) }$
