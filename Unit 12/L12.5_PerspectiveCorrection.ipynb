## Perspective Correction

#### Objective 
- $\text{Discuss in more detail the calculation of texture coordinate for pixels inside the triangles}$


### Basic Idea (Perspective Divide)

- $\text{We need to interpolate the (u,v) texture coordinates of vertices in each triangle, into the pixels (DURING) the rasterization process}$
- $\text{Suppose we have a mesh of two triangles, mapped to a 2D square domain where each vertex is mapped to one corner of the squeare texture}$

<img src="image_U12/Screenshot 2025-06-29 at 21.20.31.png" height="300" width="400"/>

- $\text{The vertices of one triangle, have the texture coordinates (0,0) (0,1) (1,1).}$

- $\text{The vertixes of the other triangle have the texture coordinates,(0,0), (1,0), (1,1)}$

<img src="image_U12/Screenshot 2025-06-29 at 21.21.50.png" height="200" width="200"/>

- $\text{Note that in this example there's no distortion nor parametrisation, since both domains are identical}$

- $\text{Assume that this mesh is placed somewhere in 3D space,} \\\text{to color the mesh we'll need to interpolate the (u,v) values, from the vertices into the triangle, to then copy the corresponding color from the texture.}$

<img src="image_U12/Screenshot 2025-06-29 at 21.26.38.png" height="200" width="200"/>

- $\text{We'll interpolate each triangle seperately, and then combine them together. However, the issue arises when we interpolate on PROJECTED triangle, since in perspective projection causes foreshortening, so distances and size get smaller as distance increases.}$

<img src="image_U12/Screenshot 2025-06-29 at 21.26.46.png" height="200" width="200"/>
<img src="image_U12/Screenshot 2025-06-29 at 21.28.12.png" height="200" width="200"/>

- $\text{The interpolational values of each vertices should take into consideration the distance (focal point) fromt he camera and tehn adjust values on the screen}$

<img src="image_U12/Screenshot 2025-06-29 at 21.32.48.png" height="300" width="300"/>

$$ u = (1 - \alpha)\frac{u_2}{z_2} + \alpha\frac{u_3}{z_3}$$


$$ u = (1 - \alpha)\frac{v_2}{z_2} + \alpha\frac{u_3}{z_3}$$

$$ \forall 0 \le \alpha \le 1 $$



### Perspective Correction Implications

- $\text{We should apply  Perspecive correction (persepective Divide) when we interpolate vertices on the edges and also when we interpolate pixels inside the triangle (Either Bi-linear or Barycentric coordinates)}$

-$\text{This also implies that we must store that z-value or the focal length f even after the projection to the veiwing plane}$
