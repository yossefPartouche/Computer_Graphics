### Minification & Magnification

#### Objective:

- $\text{Present issues regarding texture mapping, which can create artifacts in the final image such as aliasing}$

- $\text{Texture mapping maps a texture image to a surface in a 3D scene}$
- $\text{AND THEN draws the 3D scene as an image}$
    - $\text{The process can create some issues as to how the picture in the textures are drawn in the final image}$ 
    - $\text{Since texture mapping will map the same peice of texture to the same piece of surface} \\ \text{no matter how it is rendered, meaning it won't take into consideration the distance of the object by comparison of the size of the texture.}$
    - $\text{More specifically this means that in the rasterisation process it's possible for a perticular triangle} \\ \text{it will take a large area of the triangle and rasterise over many pixels,} \\ \text{where on another triangle, which takes a small portion of the texture, will only require few pixels to cover the triangle}$

$\text{For the sake of this discussion we'll call the texture pixels (i.e. the image of the original testure) Texels and the the final image as pixels}$

- $\text{The issue can be summarised as follows, it's possible that a single pixel covers multiple texels and vice versa}$

- $\text{Minification := When one pixel covers MANY texels}$
- $\text{Magninfication := When one texel covers MANY pixels}$

- $\text{Magnification can cause alisaing since we'd need to create gaps between each pixels, but this means that the final image won't be complete (PROBLEM)}$

- $\text{Minification can cause alisaing since, we have many pexel colors that map to one pixel, this means we need to sample texels for a single pixel}$

### Solution : Texture Filtering

- $\text{The color of the textred mapped pixel is determined by filtering the texels}$

- $\text{To solve the Magnification, reconstruction filters are used}$
- $\text{To solve Minification, smoothing or average filters are used}$

$\text{This being said, the process of filtering isn't as simple, since we've created a mapping system in 3D, for each face of the mesh, and indeed they can have different shapes as a set of triangles covering an area, even if we define the neighborhood a rectangular shape the mapping isn't straightforward due to the orientation}$
