## 📌 Lookup Tables in Graphics

### 🎯 **Goal:**  
Understand how **lookup tables** are used in computer graphics, particularly in **texture mapping**, **shading**, and **appearance modeling** (e.g., bump mapping and reflection mapping).

---

## 🔄 Texture Mapping as a Lookup Table

In **texture mapping**, each **vertex** of a triangle has texture coordinates \((u,v)\). During **rasterization**, these are **interpolated** across the triangle to compute a \((u,v)\) coordinate for **each pixel**.

That \((u,v)\) value is used as an **index into a texture image**, retrieving the pixel color:

✅ Thus, the texture image **acts as a 2D lookup table**.

---

## 💡 Shading Calculations in the Graphics Pipeline

The **shading formula** defines how light interacts with the surface:

$$
I_{\text{refl}} = I_E + K_A I_{\text{AMB}} + \sum_j \left( K_D I_{L_j} (\vec{N} \cdot \vec{L}_j) + K_S I_{L_j} (\vec{V} \cdot \vec{\hat{L}}_j)^n \right)
$$

Two common **shading approaches**:
1. **Gouraud Shading** — interpolate **colors** across the triangle.
2. **Phong Shading** — interpolate **normals**, then compute color per pixel.

These methods **do not use texture mapping** by default — they rely purely on lighting.

---

## 🎨 Texture Mapping + Shading = Textured Illumination

To simulate more realistic surfaces, we can **combine** texture mapping with the lighting model:

$$
I = T(u,v) \cdot I_{\text{refl}}
$$

This scales the lighting result by the texture color.

---




## 🔬 Using Lookup Textures for Material Properties

Instead of applying the texture **only** as a final multiplier, we can use **different textures** as lookup tables for material coefficients:

$$
I = T(u,v) \cdot \left(I_E + 
K_A(u,v) I_{\text{AMB}} + 
\sum_j \left( 
K_D(u,v) I_{L_j} (\vec{N} \cdot \vec{L}_j) + 
K_S(u,v) I_{L_j} (\vec{V} \cdot \vec{\hat{L}}_j)^n 
\right)\right)
$$

This allows for **per-pixel variation of material properties** (ambient, diffuse, specular), which can represent:
- Polished vs. rough areas
- Painted vs. metallic textures
- Irregular material surfaces

## 🧱 Bump Mapping (Geometry Appearance from Textures)

**Bump mapping** simulates surface detail by **altering the surface normal** $\vec{N}$ based on a texture:

$$
\vec{N} \rightarrow \vec{N}(u,v)
$$

- Gives the illusion of surface detail (bumps, wrinkles) without modifying the geometry.
- Light interacts with altered normals, enhancing realism.

⚠️ **Limitation**: Geometry remains unchanged — can result in visual artifacts like:
- Incorrect silhouettes
- Shadows that don’t reflect bump details

## 🌐 Reflection Mapping (Environment Mapping)

**Goal**: Simulate mirror-like surfaces using **precomputed reflections**.

### How it works:
1. Render the environment from the **object’s point of view**.
2. Store that image as a **texture map**.
3. When rendering the object:
   - Use the **reflection vector** at each pixel to index into the environment map.
   - Fetch the reflected color from the texture.
