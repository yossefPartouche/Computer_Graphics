## 🌑 Shadow Mapping in Computer Graphics

**Shadow mapping** is a technique used to determine whether a point (pixel) in the scene is **in shadow or illuminated** relative to a given **light source**.

---

## 🎯 Goal:
For each pixel on the screen, determine **if it's in shadow** with respect to a light source.

---

## 🔍 Core Idea:
Use a **depth buffer (z-buffer)** from the light's point of view to store the depth of the closest surface seen by the light.  
Later, during rendering from the camera's point of view, compare each pixel's depth (from the light's perspective) to this shadow map.

---

## 📦 What is a Z-Buffer?

> A **z-buffer** is a 2D array (same size as the screen/image) used in rasterization to determine **visibility** of objects in the scene.

### Key Features:
- Each pixel stores a **z-value** (depth).
- The **z-value** represents the distance of a surface point from the **camera**.
- During rendering:
  - For each pixel, the z-buffer holds the **smallest z-value encountered so far**.
  - If a new object has a smaller z-value at that pixel → it **occludes** previous objects (is visible in front).

🟦 `Z-Buffer := Determines visibility of surfaces by comparing depth values.`

### 🔧 Preprocessing Step (Shadow Map Generation):

1. **Render the scene from the point of view of the light source** \(L\), as if the light had a virtual camera.
2. Store the **z-buffer** from this rendering as a **depth map** (also called **shadow map**).
   - This shadow map represents, for each pixel in the light's image plane, the **distance to the nearest object visible to the light**.

---

### 🎨 Rendering Step (Shading with Shadows):

1. While rendering the final scene from the **camera's** point of view:
   - For each fragment (pixel), obtain its **world-space position** \( p = (x, y, z) \).
   - For each light source $L$:
     - Transform $p$ into the **light’s coordinate system** (i.e., into the light's view space):
       $$
       p' = (x', y', z') = M_L \cdot p
       $$
       where $M_L$ is the affine transform from world space to the light's view space.

2. Use $(x', y')$ to **index into the light’s depth map (shadow map)**.
   - Let $z_{\text{shadow}}$ be the depth stored in the shadow map at $(x', y')$.
   - Compare $z'$ (depth of $p$ from the light) with $z_{\text{shadow}}$.

3. **If**:
   - $z' \leq z_{\text{shadow}} + \epsilon$ → **illuminated** by the light.
   - $z' > z_{\text{shadow}} + \epsilon$ → **in shadow**.

   (The small $\epsilon$ is used to avoid floating-point precision issues.)

## ❗ Aliasing & Artifacts

### 🔍 Problem: Shadow Aliasing

- When transforming coordinates between camera and light spaces, and due to resolution mismatch between the screen and shadow map, you may get **aliasing artifacts**:
  - Jagged shadow edges
  - Incorrect shadows (false positives or negatives)

### 🛠️ Common Fixes:
- **Percentage-Closer Filtering (PCF)**: Smooths the shadow boundary by sampling multiple nearby depths.
- **Higher-resolution shadow maps**: Reduces quantization errors.
- **Cascaded shadow maps**: Use multiple shadow maps with varying detail levels for different depth ranges.
- **Depth bias**: Offset depth values slightly to avoid "shadow acne" (self-shadowing errors).

---

## ✅ Summary

Shadow mapping works in **two passes**:
1. **Shadow Map Generation**: Render from light's perspective → store closest depth values in shadow map.
2. **Scene Rendering**: Render from camera → transform each point into light’s view space → compare depths.

This allows us to efficiently determine which parts of the scene are **lit or shadowed** by each light source.
