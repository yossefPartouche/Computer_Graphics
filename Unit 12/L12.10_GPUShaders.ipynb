## 🎮 GPU Shaders and Image-Based Rendering (IMR)

### 🔧 GPU Shaders

Modern GPUs support programmable **shaders**, which are small programs that control how geometry and pixels are processed. There are two main types:

1. **Vertex Shaders**
   - Operate on each **vertex**.
   - Transform vertex positions through the pipeline (model → world → view → clip).
   - Compute attributes like normals and lighting at the vertex level.
   - **Gouraud shading** is typically implemented here.

2. **Pixel (Fragment) Shaders**
   - Operate on each **pixel (fragment)**.
   - Handle effects like lighting, texturing, and bump mapping.
   - **Phong shading** is typically implemented here (more accurate lighting).

#### Shader Efficiency

Shader performance depends on scene complexity. A key metric is:


- If **R is large** → Too many triangles per pixel → Wasteful rendering.
- If **R is small** → Many pixels per triangle → Efficient rendering.

Hardware shaders are fast, but high triangle counts still affect performance.

---

### 🖼️ Image-Based Rendering (IMR)

**Image-Based Rendering (IMR)** is a technique where 2D images are used to represent part or all of a scene, instead of modeling everything in 3D.

It enables rendering new views **without full 3D geometry**, saving effort and computational cost.

---

### 📌 IMR Techniques

#### 🔳 Billboard Rendering

- A 2D image (e.g. a tree) is mapped onto a flat square that always faces the camera.
- Used in games for sprites, fire, grass, etc.

<img src="image_U12/Screenshot 2025-07-03 at 12.07.52.png" height="400" width="400"/>
<img src="image_U12/Screenshot 2025-07-03 at 12.08.08.png" height="400" width="400"/>

**Pros:**
- Simple and fast
- Lightweight for GPU

**Cons:**
- Appears flat from side angles
- Needs to rotate to face the camera

#### 🧊 Cube Mapping

- The camera is placed inside a cube textured with 6 images (one per face).
- Gives a 360° environment look.
- Used in skyboxes and reflection environments.

<img src="image_U12/Screenshot 2025-07-03 at 12.11.54.png" height="400" width="400"/>
<img src="image_U12/Screenshot 2025-07-03 at 12.49.32.png" height="400" width="400"/>


#### 🌀 Cylindrical Mapping

- A 360° panoramic image is wrapped around a vertical cylinder.
- Camera is placed inside.
- Used in virtual tours and VR previews.
