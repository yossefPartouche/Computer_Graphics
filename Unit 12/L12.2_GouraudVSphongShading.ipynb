### Gouraud vs. Phong Shading 
#### Objective: 
- $\text{Show how a slightly different model of interpolation could solve the temporal aliasing artifact presented in L12.1}$
- $\text{Present the Gouraud and Phong Shading models}$

#### Gouraud Shading - Cause for Temporal Aliasing
$\text{Gouraud Shading := The color of points inside the triangle is determined by interpolating the colors of the vertices of the triangles}$
- $\text{This method provides a quick and simple way to fill the colors inside the triangles} \\ \text{in a smooth manner (This is what we've describing so far in the rendering pipeline.)}$

- $\text{The problem is that it can create artifacts}$
- $\text{Aliasing artifacts in smooth shading, are caused by pre-calculating the colors on the vertices, and using interpolation into the triangle.} \\ \text{in some cases specular highlights that should appear are gone, since the position of the vertices are positioned in way that their normals are outide the range for such a highligh to appear}$
- $\text{So we have two normals that don't have high specular value} \rightarrow \text{The interpolated range between the two vertices won't have a high specular value}$

#### Re-analyze Lighting Stage in the Rendering Pipeline

- $\text{We're trying to understand if there's a way to solve the artifacts that Gouraud shading can produce,} \\ \text{without calculating the color at every fragment (to reduce complexity)}$
- $\text{The lighting stage uses the following formula:}$

$$
\color{orange}{I_{\text{refl}}} = 
\color{teal}{I_E} + 
\color{yellow}{K_{AMB}} \cdot \color{teal}{I_{\text{AMB}}} + 
\sum_j \left(
  \color{green}{K_D} \cdot \color{brown}{I_{L_j}} \cdot (\vec{N} \cdot \vec{L}_j) +
  \color{red}{K_S} \cdot \color{brown}{I_{L_j}} \cdot (\vec{V} \cdot \vec{\hat{L}}_j)^n
\right)
$$

---

### Color Code Breakdown

| **Component**                       | **Meaning**                                            | **Color**     |
|------------------------------------|--------------------------------------------------------|---------------|
| $\color{orange}I_{\text{refl}}$                  | Reflected intensity (final color)                     | 🟧 `orange`   |
| $\color{teal}I_E$                              | Emissive light                                         | 🟦 `teal`     |
| $\color{yellow}{K_{AMB}}, \color{teal}K_D, \color{red}K_S$                | Material properties (ambient, diffuse, specular)       | 🟦 `blue` (AMB), 🟩 `green` (D), 🟥 `red` (S) |
| $\color{teal}{I_{\text{AMB}}}, \color{purple}I_{L_j}$          | Ambient and point light intensities                    | 🟦 `teal`, 🟫  `brown` |
| $\color{brown}\vec{N}, \vec{L}_j, \vec{V}$      | Geometry (normal, light dir, view dir)                 | 🟫  `brown`             |
| $\color{brown}(\vec{N} \cdot \vec{L}_j)$        | Diffuse term (dot product between normal and light)    | 🟫  `brown`             |
| $\color{brown}(\vec{V} \cdot \vec{\hat{L}}_j)^n$| Specular term (view/light angle, raised to shininess)  | 🟫  `brown`             |

---

### Breakdown of Contributors

1. **Global Material Properties**  
   - $K_{AMB}$, $K_D$, $K_S$ control how the surface interacts with different light components.

2. **Global Light Sources**  
   - $I_E$, $I_{\text{AMB}}$ are constant for the whole object.  
   - $I_{L_j}$ vary by light source $j$.

3. **Local Geometry (Per-Vertex)**  
   - Calculated using:
     - Vertex position
     - Vertex normal $\vec{N}$
     - Direction to light $\vec{L}_j$
     - View direction $\vec{V}$
     - Reflected light direction $\vec{\hat{L}}_j$

---

> **Goal**: Improve the visual smoothness (e.g., eliminate temporal aliasing/blinking) of **Gouraud shading**, without computing full lighting per-fragment (as in Phong shading), by reconsidering which parts can remain global or interpolated.

#### Phong Shading

- $\text{Phong Shading := The color of points inside the trinagle is determined by interpolating the normals of the vertices of the triangles, per pixel}$
- $\text{This can solve the aliasing artifacts since when we interpolate two or more normals,} \\ \text{do don't create specular highlights the interpolated normal will point towards the direction that does create specular highlights}$
- $\text{We see that this produces more realistic renders, but it's far more computationally expensive}$
