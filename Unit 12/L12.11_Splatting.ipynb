## 🌐 Point-Based Rendering – Splatting

Point-based rendering is a **geometry-centric** approach to representing and rendering 3D objects without relying on mesh connectivity. Instead of triangles or polygons, objects are described using **discrete sample points** on their surface, each carrying **local geometric and material information**.

---

### 🔍 Definition

**Point-Based Rendering (PBR)** represents objects as a **set of unconnected points** sampled over their surfaces. These samples are referred to as **surfels** (surface elements). Each surfel encodes:
- A 3D **position**
- A **normal vector**
- **Material properties** (e.g., diffuse, specular coefficients)
- **Color**

> This representation discards mesh topology and retains only **local surface properties**.

---

## ✴️ Splatting: The Rendering Process

In splatting, each point is projected into image space (viewing plane) and contributes to pixel values by **covering a small image region** called a **splat**.

### 🛠 Rendering Pipeline

1. **Surface Sampling**:
   - The object’s boundary is sampled densely to produce a set of points.
   - Each point stores:
     - Position $\vec{p} \in \mathbb{R}^3$
     - Normal $\vec{n} \in \mathbb{R}^3$
     - Material parameters (e.g., $K_A, K_D, K_S$)
     - Optional: texture coordinates, reflectance models

2. **Phong Illumination Model**:
   - To apply lighting using the Phong model, each point must have:
     1. **Material properties**
     2. **Position** (to calculate light/view directions)
     3. **Normal vector** (for Lambertian and specular terms)

   - The reflected intensity at each point is computed as:
     $$
     I_{\text{refl}} = I_E + K_A I_{\text{AMB}} + \sum_j \left( K_D I_{L_j} (\vec{N} \cdot \vec{L}_j) + K_S I_{L_j} (\vec{V} \cdot \vec{\hat{L}}_j)^n \right)
     $$

3. **Splat Generation**:
   - Each point is **projected** onto the viewing plane.
   - A **screen-space kernel** (splat) is drawn centered at the projected location.
   - A splat is typically:
     - A **disc**, **ellipse**, or **square**
     - Aligned with the point’s **normal vector**
     - Sized to ensure surface coverage

4. **Coverage Considerations**:
   - **Completeness**: Splats must be large enough to **cover the surface without holes**.
   - **Overdraw Avoidance**: Splats must not be so large as to cause **blurring or overlapping artifacts**.

   > Analogous to ink dots in high-resolution printing – dots must be small enough to be imperceptible, but numerous enough to cover the paper.

5. **Output Sensitivity**:
   - Rendering quality can be adapted by the **number and size of points**.
   - A **higher level of detail (LOD)** implies:
     - More points
     - Smaller splats
     - Finer visual resolution

## 📐 Mathematical Considerations

Let:
- $P = \{p_i\}_{i=1}^N \subset \mathbb{R}^3$ be the set of sampled points
- Each point $p_i$ has a normal $\vec{n}_i$ and attributes $\theta_i$ (material, color, etc.)
- Viewing projection: $\Pi: \mathbb{R}^3 \rightarrow \mathbb{R}^2$
- Let $S_i \subset \mathbb{R}^2$ be the image of the splat for $p_i$

Then the final rendered image $I(x,y)$ is approximated as a **splat convolution**:
$$
I(x,y) = \sum_{i=1}^N I_i \cdot K_i(x,y)
$$
where:
- $I_i$ is the shaded intensity at point $p_i$
- $K_i(x,y)$ is a 2D kernel centered at $\Pi(p_i)$

This kernel can be Gaussian, flat disc, or elliptical, depending on the renderer.

## ✅ Summary

| Concept                | Description |
|------------------------|-------------|
| **Point-Based Rendering** | Object surfaces represented as a set of unconnected, attribute-rich points. |
| **Splatting**             | Rendering method where each point projects a small kernel (splat) onto the image plane. |
| **Inputs for Illumination** | Position, normal, and material properties are needed for shading calculations. |
| **Coverage Control**       | Balance between splat size and point density ensures visual fidelity without gaps or blur. |
| **Adaptability**           | Rendering quality can scale with resolution by increasing point density and decreasing splat size. |

---

## 🧠 Advantages of Point-Based Splatting

- No need for mesh connectivity
- Efficient level-of-detail control
- Naturally handles complex geometry (e.g., foliage, scanned surfaces)
- Parallelizable rendering

## ⚠️ Limitations

- Visibility ordering and blending of splats may be non-trivial (especially in semi-transparent surfaces)
- Gaps or overlaps may occur if point distribution or splat size is poorly chosen
- Not yet as hardware-accelerated as mesh-based methods in standard GPUs
