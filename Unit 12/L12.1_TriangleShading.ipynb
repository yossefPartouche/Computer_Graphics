## Triangle Shading

#### Objective: 


#### $\text{How is the color of a point on a 3D object determined in the rendering pipeline}$
- $\text{We have two ways to define colors on objects}$
1. $\text{To directly define it for triangles or vertices}$
2. $\text{To use material properties and the interaction with light}$

- $\text{Note that color isn't calculated or defined for every point on the object, but rather (calculated or defined) on the vertices or the triangles themselves of the object}$
- $\text{The color is calculated in the second step called lighting, where it involves the calculation of color of the vertices using the formula learnt in unit 5.}$
- $\text{These calculations depend in the position of the veretx and its normal and therefore, must be calculated in 3D before the triangle is projected in 2D}$
$$
I_{\text{refl}} = I_E + K_{AMB} I_{\text{AMB}} + \sum_j \left( K_D I_{L_j} (\vec{N} \cdot \vec{L}_j) + K_S I_{L_j} (\vec{V} \cdot \vec{\hat{L}}_j)^n \right)
$$

| **Symbol**         | **Definition**                          | **Type / Space**                  | **Meaning**                                  |
|--------------------|------------------------------------------|-----------------------------------|----------------------------------------------|
| $I_{\text{refl}}$  | Reflected light intensity                | RGB color                         | Total light reflected from the surface       |
| $I_E$              | Emissive color                           | RGB color                         | Light emitted by the object itself           |
| $K_{\text{AMB}}$   | Ambient reflection coefficient           | Scalar or RGB                     | How the surface responds to ambient light    |
| $I_{\text{AMB}}$   | Ambient light intensity                  | RGB color                         | Ambient light in the scene                   |
| $K_D$              | Diffuse reflection coefficient           | Scalar or RGB                     | Material's base color under diffuse light    |
| $I_{L_j}$          | Light intensity of light $j$             | RGB color                         | Intensity/color of light source $j$          |
| $\vec{N}$          | Normal vector at surface point           | Unit vector (view/world space)    | Perpendicular to the surface                 |
| $\vec{L}_j$        | Direction to light $j$                   | Unit vector                       | From surface point to light source           |
| $K_S$              | Specular reflection coefficient          | Scalar or RGB                     | Material's shininess response                |
| $\vec{V}$          | View direction                           | Unit vector                       | From surface to camera                       |
| $\vec{\hat{R}}_j$  | Reflection direction of light $j$        | Unit vector                       | Reflected light direction (mirror reflection)|
| $n$                | Shininess exponent                       | Scalar                            | Controls sharpness of specular highlight     |

- All color quantities (e.g. $I_E$, $K_D$, $I_{L_j}$) are typically **RGB vectors**.
- All direction vectors (e.g. $\vec{N}$, $\vec{L}_j$, $\vec{V}$, $\vec{\hat{R}}_j$) are **normalized** (unit length).
- The exponent $n$ makes highlights sharper as it increases — used in **Phong reflection model**.
- The summation $\sum_j$ covers **multiple light sources**.

- $\text{So the color of an object only occurs in the rasterisation stage even though we calculated the color properties of the vertices in the lighting stage}$
- $\text{Since we saw that the color of a triangle can be determined using: }$
1. $\text{Barycentric Method}$
2. $\text{Linear interpolation}$

### Example: Smooth Shading on a Polytetrahedral Sphere


• A sphere constructed from flat triangles can be modeled as a *polytetrahedron*.  


• When using $\text{smooth shading}$, this polyhedral object appears visually smooth, effectively resembling a sphere, even though it’s composed of flat faces.


---

#### Temporal Aliasing Effect


• The issue arises when the object is **animated or rotated** over time.  


• In smooth shading, **color (including highlights)** is interpolated across **vertices** of the mesh.  


• **Specular highlights**, in particular, are highly **direction-dependent**—they appear only where the angle between the light source, the surface normal, and the viewer aligns.  


• Between two vertices where one has a highlight and the other doesn’t, the **interpolated region** may lack any highlight entirely.  


• As the object rotates, these regions **“blink” on and off**, creating a visual artifact—this is a form of $\text{temporal aliasing.}$  


• This artifact occurs because the triangle mesh is not dense enough to capture continuous highlight motion, resulting in an **unrealistic shimmering** effect.
