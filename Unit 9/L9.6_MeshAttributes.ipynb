Since a mesh is made from 3 entities (point, Line, Face) we can define the attributes for each one. Furthermore, using the index face seperable representation, we can just insert more properties to each entry. This means we can add color to each face or even to each vertex, in the vertex or face table respectively by adding the RGB colors.

Note that to add properties to edges will require to create an edge table (O(n)) and then add those desired properties.

So if we wanted to compute the normal of each face (Triangle Face in this case) we can compute it in constant time and add it to the Face List making it a Face & Normal List.


#### How do we acheive smoothness of object (smooth shading)?
#### We require that each vertex (Point) has a normal instead of each Triangle face

#### How do we estimate the normal at a point (vertex)?
- Since we're able to deduce the neighoborhood of faces surrounding the vertex, we can calculate this normals. 
- Then Sum all those normal and divide by number of normals calulated.
- This will create an average of the normals for that point smoothing out the flatness of an object.

<img src="image_U9/9.6/Screenshot 2025-06-02 at 14.43.29.png" height="400" width="400"/>

We conclude that we need the mesh representation to support the V --> F incidence, since "for each vertex we want to obtain the normals surrounding this point". This isn't stored explicitly as such to find triangles incident to a vertex and to apply this to all vertixes will fortunately only cost us O(n) number of faces (triangles). This is because we can build a table to store the V->F List, for all the vertices in O(n), thus containing m entries for m vertices, and each entry will store an array of entries containing the triangle around the vertex.

We itereate through the list of F ->V table and for each Triangle whose vertices are $v_i1, v_i2 \ and \ v_i3$, we insert $t_i$ to the V-> F list.

### Why bother with this?

Note that one query without the table wouldn't have cost us O(n), and since populating the whole table will take O(n) the following expected query will only cost O(1). Furthermore, with F->V and V -> F we obtain V-V in constant time as well (for free). 

### Example

Suppose we wanted to obtain all the vertices neighboring some vertex. 
- Go to the V -> F table and retriece the faces
- Then retrieve the vertices of these face
- We obtain all neighbouring vertices.

### Example

Suppose we want neighboring faces of a perticular Triangle
- Go to to the F -> V table obtain neighboring vertcies. 
- The for each unique vertex, go to V -> F table 
- We obtain in O(1) all neghbouring faces of a perticular face.

#### Edge Case

if have two face orthonormal to another, this means it produces a sharp edge on the object itself, and this may be the way we'd want to represent the object as such for that perticular edge connecting two faces that are orthonormal, must store both normals. (in essence go according to the flat shading effect). This is an extension to the mesh data structure, since it will contain two normal entries and not one.
