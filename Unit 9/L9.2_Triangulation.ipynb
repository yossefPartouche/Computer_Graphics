## Triangulation

#### Objective
 We'll Define the combination of multiple triangles to create shapes in 2D

$\text{Given a set of n point in 2D we can create a triange by just connecting 3 points.}$
#### Q. How do we combine a set of points to create a convex region of triangle meshes?
 
$\text{Tirangulation} = \text{is a set of n points in 2D which is connected set of triangles such that:}$
1. $\text{The vertices of the triangles come only from the n-points}$
2. $\text{Every point in the n-point set corresponds to at least one triangle}$
3. $\text{The interiors of any two triangles do not intersect}$
4. $\text{If two triangles are not disjoint then they share a vertex or a whole edge}$
5. $\text{The union of all triangles is the convex hull of P}$

#### How can we locate the position of a point inside a triangulation?

$\text{Formally := Given a triangulation T of n points P and a point } p \notin P \text{ where } p \in T, \text{ we want to find the triangle } t_i \text{ that contais the point}$

1. $\text{Go over all triangles (where we know each vertices)}$
2. $\text{Apply the barycentric Method, detemine if P is in the current Triangle}$
3. $\text{Complexity is linear in the number of triangles. O(T)}$

#### ⚡ Optimization: “Walking” Point Location Using Barycentric Coordinates

$\text{Instead of checking all triangles, we start from an arbitrary triangle and use the sign of the barycentric coordinates to guide us toward the correct triangle — this is called a “walk” in the triangulation.}$

$\text{For a triangle }  \triangle v_1v_2v_3, \text{ the barycentric coordinates of a point p are:}$
$$p = \alpha_1 v_1 + \alpha_2 v_2 + \alpha_3 v_3 \quad \text{where} \quad \alpha_1 + \alpha_2 + \alpha_3 = 1$$
$\text{If:}$
$\alpha_1, \alpha_2, \alpha_3 > 0 \Rightarrow p \in \text{interior of triangle}$
$\text{At least one } \alpha_i < 0 \Rightarrow p \text{ lies outside the triangle, across the edge opposite to vertex }$ v_i.

##### Algorithm: 
1. $\text{Start at any triangle t in the triangulation}$
2. $\text{Compute the barycentric coordinates } (\alpha_1, \alpha_2, \alpha_3) \text{ for point } p$
3. $\text{If all } \alpha_i \text{are positive - Done}$
4. $\text{If one is negatvie i.e. } \alpha_2 < 0$
    - $\text{That means that p lies outside the triangle, across the edge opposite to vertex } v_i \text{ in our case } v_2$
    - $\text{That edge is shared with a neighboring triangle (By definition of Triangulation) - move to that triangle}$
5. $\text{Repeat this walk until you find the triangle containing p}$

#### Complexity:
$\text{Worst-case complexity:} \mathcal{O}(T) $

$\text{But in practice (especially with Delaunay triangulations), walking takes:}$

$\mathcal{O}(\sqrt{T}) \text{steps on average.}$

$\text{Even better with preprocessing or spatial hierarchy (like triangulation trees).}$

##### $\text{What data structure and data, do we need to iterate through the triangulation?}$

$\text{Reccomended: Adjacecy List that stores the neighboring triangles, or Null if it's on the boundary of the triangulation.}$

$\text{For example:}$

$T_i \rightarrow [T_{n1},\ T_{n2},\ \text{null}] \quad \text{// third edge is a boundary}$

### Type of relations

- 3 Adjancecy Relation 
    1. $\text{Vertex - Vertex}$ - Which vertices are directly connected (via an edge)
    2. $\text{Edge - Edge}$ - Adjacent edges sharing a vertex
    3. $\text{Triangle - Triangle}$ - Neighboring triangles sharing an edge (**most useful** for traversal)
- 6 Incidence Relations
    1. $\text{Vertex - Edge}$ - Which edge connect to a vertex 
    2. $\text{Vertex - Triangle}$ - Which triangles connect to a vertex
    3. $\text{Edge - Vertex}$ - Two vertices that define an edge
    4. $\text{Triangle - Vertex}$ - Three vertices of a triangle
    5. $\text{Edge  - Triangle}$ - Which triangle(s) an edge belongs to (at most 2 by Defintion of Triangulation)
    6. $\text{Triangle - Edge}$ - The three edges making up a triangle

- All these types of relations that we choose to store will greatly effect the efficiency of operations and algorithms.
