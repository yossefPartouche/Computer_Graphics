## Triangulation

#### Objective
1. We'll Define the comination of multiple triangles to create shapes in 2D

- Given a set of n point in 2D we can create a triange by just connecting 3 points. 
#### Q. How do we combine a set of points to create a convex region of triangle meshes?

#### A. 
$\text{Tirangulation} = \text{is a set of n points in 2D which is connected set of triangles such that:}$
1. The vertices of the triangles come only from the n-points
2. Every point in the n-point set is a vertex of at least one triangle
3. The interiors of any two triangles do not intersect
4. If Two triangles are not disjoint then they share a vertex or a whole edge
5. The union of all triangles is the convex hull of P

#### How can we locate the position of apoint inside a triangulation?

Given a triangulation $T$ of n points P and a point $p \notin P$  where  $p \in T$, we want to find the triangle $t_i$ that contais the point

- Go over all triangles (where we know each vertices) and apply the barycentric Method.
- Complexity is linear in the number of triangles.

#### Can we do better?

- If we know that a point isn't inside a triangle we said that there will be at least one negative barycentric coordinate. 
- We check the coordinate that returned (Most) negative value, and we know the side it's located **RELATIVE** to our desired location. 

- In the example below since $\alpha_2$ is negative, we know that the point p lies on the otherside in which that barycentric coordinate is 0.
- We then move to neighboring triangles of this edge and making sure we're getting closer and continueing this process.

### Type of realtions

- 3 Adjancecy Relation 
    1. Vertex - Vertex
    2. Edge - Edge
    3. Triangle - Triangle 
- 6 Incidence Relations
    1. Vertex - Exdge 
    2. Vertex - Triangle 
    3. Edge - Vertex
    4. Triangle - Vertex
    5. Edge  - Triangle
    6. Triangle - Edge

- All these types of relations that we choose to store will greatly effect the efficiency of operations and algorithms.
