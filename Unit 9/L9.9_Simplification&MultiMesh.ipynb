### $\text{Mesh Simplification := Aims to reduce the number of elements \{ Vertices, edges, Triangles \} }$

This is done for the following reasons:
1. May not fit in memory (Large Meshes)
2. Will take a long time to render (Large Meshes)
3. Will have many triangles per image pixel, each pixel can only have a single color, so we're still limited by the number of pixels.

As such we must fill the hole in the mesh with triangles such that the resulting mesh would still be a valid triangulation, so the mimum number of triangles we can use to fill the hole is 4.

#### Q. Doesn't this counter the benefit of mesh simplification?
- Removing a vertex will always remove 2 triangles from the mesh, so this will reduce the total count of tirangles and edges as well. Usually removing a triangle mesh affects the topology and the geometry of the mesh as well, hence mesh simplification is a type $\text{Remeshing}$.

#### Q. How do we choose the triangulation that fills the hole?
- The hole depends on the resulting geometry, we would like to change the geomerty of the mesh as litle as possible, so me need to minimize the difference in shape between the troangle surfaces, before removing the surface and after, as such we must define a measure of distance between surfaces.

- Just as an example, we could measure the change in volume before and after the removal of the mesh, and then we just calulate the minimizing set of meshes.

#### Mesh Simplification Algorithm 


```pseudo
Input: Mesh M = (V, E, F)
Output: Simplified mesh M'

1. Compute an error metric for each edge (e.g., Quadric Error Metric or edge length)
2. Insert all edges into a priority queue Q sorted by error cost

3. while number of vertices > target or desired error not reached:
    a. Pop the edge e = (v1, v2) with lowest cost from Q
    b. Check if collapsing e is valid (e.g., no topological issues or inverted normals)
    c. Compute the optimal collapse position v_new (e.g., midpoint or error-minimizing point)
    d. Collapse edge e:
        i. Replace v1 and v2 with v_new
        ii. Update affected faces and edges
    e. Recompute error metrics for edges affected by the collapse
    f. Update Q accordingly

4. Return the simplified mesh M'

### Notes:
- You can replace the error metric with **Quadric Error Metrics (QEM)** for more advanced simplification.
- You may include an image in the cell to visually explain edge collapse:
  ```

- Note, the more iterations we apply the more simple the object representation thus simpler to draw and also the larger the geometric error.
- In certain cases we can afford to simplify the object to a great deal, such as when the image will be far, so difficult to distinguish the details.

<img src="image_U9/9.9/Screenshot 2025-06-03 at 20.07.10.png" height="350" width="250"/>
<img src="image_U9/9.9/Screenshot 2025-06-03 at 20.27.01.png" height="350" width="250"/>
<img src="image_U9/9.9/Screenshot 2025-06-03 at 20.07.25.png" height="350" width="250"/>
<img src="image_U9/9.9/Screenshot 2025-06-03 at 20.27.28.png" height="350" width="250"/>
<img src="image_U9/9.9/Screenshot 2025-06-03 at 20.27.40.png" height="350" width="250"/>
<img src="image_U9/9.9/Screenshot 2025-06-03 at 20.27.59.png" height="350" width="1265"/>

### Ratio Triangle : Pixel

$\text{R = The ratio between the number of triangles and the number of pixels}$
- If we have a small $\text{R Value}$ it means that more pixels than Triangles which is good utilisation for rendering.
our goal is $\text{Output sensitive Algorithms = that depends not only on the size of the input(Number of Triangles) but also on the size of the output (Number of pixels)}$

- To enable this sensitivity on demand, is to store differen versions of the 3D mesh representation, each with different levels of simplification also known as $\text{Multi-Resollution/Level Of details}$

<img src="image_U9/9.9/Screenshot 2025-06-03 at 21.04.48.png" height="325" width="900"/>

- As mentioned above, when we need high level of detail, we can use the more complicated/detailed mesh representation, and when we need lower level resolution or the need so much detail we use simpler representation.


<img src="image_U9/9.9/Screenshot 2025-06-03 at 21.05.50.png" height="325" width="800"/>
<img src="image_U9/9.9/Screenshot 2025-06-03 at 21.06.54.png" height="325" width="800"/>


- In addition, when clipping the reprsentation, in the areas in which the object won't be visible we can simplify the mesh, and increase the complexity of resolution in the visible areas.

<img src="image_U9/9.9/Screenshot 2025-06-03 at 21.08.47.png" height="325" width="325"/>


- Another way mesh sensitivity can help, is light effect, consider an area of an object in which a specific area has high level of specular effects, we could increase the mesh density in that area, to increase the specular value (since it's dependant on the number of normal vectors) and indeed reduce the mesh density in other areas.

<img src="image_U9/9.9/Screenshot 2025-06-03 at 21.10.23.png" height="325" width="325"/>
