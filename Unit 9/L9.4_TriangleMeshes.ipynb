#### Objectives:
1. Present Possbile representations of surfaces
2. Define the triangle mesh representation

$\text{Boundary Triangle Mesh = Is a triangulation embedded in 3D that is Orientable, 2-Manifold and usually also water tight}$

#### Triangle Mesh Properties:
1. The mesh is orientable. 
2. The mesh boundas finite volume (Water tight).
3. The mesh contains no self-intersections.
4. Two non-dosjoint triangles can only share an edge or a vertex.
5. Only two triangles can share an edge.
6. The triangles neighborhood around a vertex can be mapped to the Euclidean Plane.
