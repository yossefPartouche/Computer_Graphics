#### $\text{Definitions}$

$\text{Affine Combination of n points} = \sum_{i=1}^{n} \lambda_i p_i \ \text{where} \ \sum_{i=1}^{n} \lambda_i = 1$

$\text{Convex Combination of n points} = \sum_{i=1}^{n} \lambda_i p_i \ \text{where} \ \sum_{i=1}^{n} \lambda_i = 1 \ \text{and} \forall \ i \ \lambda_i \ge 0$

$\text{Convex Set S} = \text{A subset of an affine space that is closed under convex combination}$

$\text{Convex Region R}= \text{In Eucledian space is a region where, for ever pair of points within the redion,}$
$\text{every point on the straight line-segment that joins that pair of point s also within the region}$ 

$= \forall \ p_0, p_1 \in R, \ and \  t \in [0,1] \ if \ p = (1-t)p_0 +tp_1 \rightarrow p \in R$

### $\text{Triangle Simplex}$

By convex combination we have that 

$$\forall \ p_1, p_2,p_3 = \text{Triangle, we have} \ p \in \text{Triangle} \iff \ p = \alpha_1 p_1 + \alpha_2 p_2 + \alpha_3 p_3$$

$\text{Barycentric Coordinates} = \text{the coefficients of the convex combination of the Triangle} \ \alpha_i \ i \in \{1, 2, 3\}$


#### $\text{Calculating Barycentric Coordinates }$
- Add line segments from the point p to each vertex.
- Look at the ratio of area of triangles created and area of whole triangle

- $\alpha_1 = \frac{A_1}{A_0}$
- $\alpha_2 = \frac{A_2}{A_0}$
- $\alpha_3 = \frac{A_3}{A_0}$
- $\alpha_1 + \alpha_2 + \alpha_3 = 1 $

To Calculate Area of a triangle: 

$$P_1 = (x_1,y_1) / P_2 = (x_2,y_2) / P_3 = (x_3,y_3) \ then \
\text{Area} = \frac{1}{2} 
\left|
\begin{vmatrix}
x_1 & y_1 & 1 \\
x_2 & y_2 & 1 \\
x_3 & y_3 & 1
\end{vmatrix}
\right|
$$


##### Example:
Point lies on the edge of the triangle then: 

$$ p = 0p_1 + \alpha_2p_2 + \alpha_3p_3 = (1-\alpha_3)p_2 +  \alpha_3p_3$$

We can conclude the $$ p_1 = (1,0,0), p_2 = (0,1,0) \ and \ p_3 = (0,0,1)$$

### Affine Combination Implications

- If we drop the convex constraint of non negative coefficients, we'll get any point in the affine space relative to the triangle vertices.

- We use this to look at iso-contours (3 different type of lines), created by integer coordinates, we'll create mesh of triangles covering the whole space.
- In each direction the lines coordinates correspond to all the points where one of the coordinates equals some integer value.
- Points on the vertices of each triangle tile will have integer coordinates and inside will have non-integer coordinate, and any points outside the relative triangle will have at least one negative coordinate.
