#### Objective

- We look at 3D space, and define properties of a boundary representation of 3D object or shape

1. $\text{We will first define a method to represent 3D objects' shapes}$
2. $\text{Then add other properties to the representation of 3D objects}$

- $\text{Note in Computer Grphics in order to render an image, the only thing that's important is the outside shape.}$
- $\text{Which is why we represent objects as empty shells and thus concentrate defining only the outside boundary surface of the object.}$
- $\text{If an object is 3D it's boundary is a closed 2D Surface} \\ \text{(you can imagine it as if you have wrapping paper to gift wrapping a box shape with no box or gift inside)}$
- $\text{The boundary will define what is outside and what is inside the object.}$
- $\text{As such the boundary representation of an object is a 2D surface embedded in 3d space}$
- $\text{It's important to note that some embedded surfaces are open as such won't be a boundary for an object.}$

#### 2D surfaces embedded in 3D - Properties 
- $\text{Specifically surfaces that represent 3D Objects}$
- $\text{Orientability = A surface is orientable if it's possible to make a consistent choice of surface normal vector at every point}$
    - This means once we choose a normal vector from one surface it's possible to find a normal vector for all other points (outside normal and inside normal).
    - The mobious Strip is an example where we start with the normal on the outside and continue creating normals along the surfaces and arrive at a normal on the inside this a $\text{Non-Orientable Surface}$

- $\text{Maniflod Surface = a topological space that local resembles Euclidean Sapce near each point,} \\ \text{each point of an n-manifold has a neighbourhood that is homeomorphic to the Euclidean Space of dimension n}$

    - $\text{A 2D surface would be a 2-Manifold if it locally resembles the 2D Euclidean space.}$

    - $\text{Homeomorphism = A bijection between topological spaces that performs some continuous} \\  \text{stretching/bending with creating any holes in the surface}$
    - $\text{In Computer Graphics, we usually suggest that the object is a 2-Manifold,} \\ \text{this means the surface in 3D cannot include self intersections, since it will break the bijectivity.}$

- In Triangulation in 3D, the criteria that no two objects intersecting get's rid of the $\text{Non-Manifold}$ property, but they can still share an edge or a vertex.
    - Note that in 3D more than 2 triangles can share an edge so a $\text{Non-Manifold Edge}$. 
    - Many Triangles can also share a vertex so a $\text{Non-Manifold Vertex}$
