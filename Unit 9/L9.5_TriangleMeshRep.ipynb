##### Practical Problem 
Note that it's possible that a point connected to multiple edges can look as if that point is in a constant position in space but actually have different coordinates! As such if we wanted to changes a coordinate of a triangle we'd need to find the connected edges to adjust the mesh topology of the neighborhood, but since a single vertex can have minor differences between each connected (other) vertex, we wouldn't be able to rely on the V-V relation, just wouldn't be efficient. 

##### Soup Mesh Conclusion

This means that the list of vertices are useful for just containing a list of triangles, but we can't assume any relation between the triangles in the list due to the inconsistencies.
which is what we call a $\text{Triangle Soup}$.

#### Explicit Represention of Vertices

This is known as the $\text{Indexed Face set Triangles Mesh Representation}$


1. Add a list of all vertices that are included in all triangles  in a mesh.
2. Each vertex represent by three coordinates, so with m vertices we create a vertcies list of m points
3. We will use this list to define the triangles, via indexing meaning we create a list of faces (still containing triplets of coordinates).
4. each element in the triplet will refer to an index in the vertices list (To avoid the above pitfall) 

So now if we wanted to change the position of a vertex using this system, we'd find the faces that map to the desired vertex in the vertices list. Complexity $O(1)$

#### Indexed Face Set Mesh Representation

- Edges in Indexed face set are NOT stored explicitly, instead we would create a third list of faces and each pair of edge will necessarily be an edge (This is derived by the Convex property). 
- So we do this over the whole list of trinagles and then iterate over this list once more to remove the duplicated edges as each edge will share at more 2 triangles 
- We could create a mapping to reduce this complexity to $O(n)$ (Triangles).

We conclude that using this representation we can store the following relations:
- $ F \rightarrow V$
    - Take O(1)
- $ F \rightarrow F$
    - To do this we start with a triangle and find it's vertices (using F to V) 
    - We then go over he list of triangles to check if each triangles contains 2 of the 3 vertices
    - O(n)
- $ V \rightarrow V$
    - To do this we start with the vertex and iterate throught the list of faces
    - We must find the triangles that have this vertex and extract the other two vertices
    - O(n)
- $ V \rightarrow F$
    - We must go through the list of faces to check if a triangles has the vertex.
    - O(n)
