### The projection Transformation

This is a linear transforamation P from a vector space to itself such that $P^2 ~=~ P$
So if you apply P any number of time, you will still get the same result.

The identity transformation is such a transformation

If $M_p$ is a projection matrix and $M_p \neq I_n$ then $M_p$ is singular (not invertible). Similarly, the geormetric operation of projection is also not invertible.

A projection in 2D must reduce the dimension and as such can project all the points in 2D space on to: 
- A Line
- A point: This is a mapping all the space to the origin, represented by $0_n$ matrix

### Projection on the Axis

- Any 2D point $(x,y)$ on the coordinate system can be project to $(x,0)$, the corresponding matrix transformation is $$
\begin{pmatrix}
1 & 0 \\
0 & 0
\end{pmatrix}$$  
- Any 2D point $(x,y)$ on the coordinate system can be project to $(0,y)$, the corresponding matrix transormation is  
$$\begin{pmatrix}
0 & 0 \\
0 & 1
\end{pmatrix}
$$
