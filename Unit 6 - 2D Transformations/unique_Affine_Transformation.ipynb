#### Uniquely Defining Affine Transformation
##### Objective:
1. To answer the following Question: How many mappings of 2D points are needed to uniquely define a 2D Affine Transformation

Currently we know that given an Affine Transformation $T$, the for every point $p$ we can find the traget point $p'$ by applying $T(p) = p'$  
Suppose now we're given a set of points $\{p_1, p_2, \ ... p_k\}$ and their resulting mappings $\{p'_1, p'_2, \ ... p'_k\}$  
1. Can we find the mapping $T$ that produces these results?
2. How many points are needed to uniquely define a 2D affine transformation?
3. Can we define an affine transformation for any mapping of any number of k points?

To answer the second question we must ask:
- How many free parameters deinf a 2D Affine transformation (How many can be changed/are variables?) 
 \begin{pmatrix} a_{11} & a_{12} & v_x \\ a_{21} & a_{22} & v_{y} \\ 0 & 0 & 1 \end{pmatrix}

A. **Answer: 6** $\{a_{11}, a_{12}, a_{21}, a_{22}, \vec{v_x}, \vec{v_y}\}$
These unknowns denote the degrees of freedom.

#### Details: 
- Every mampping of a point add two constraints to our problem
    1. We know that $x_1$ is mapped to $x'_1$
    2. We know that $y_1$ is mapped to $y'_1$
- We take the general form of the Affine tranformation and multiply it with our known homogeous form of $p$ and equate to our known homogeous form of $p'$  
$$p' =  \begin{pmatrix} a_{11} & a_{12} & v_x \\ a_{21} & a_{22} & v_{y} \\ 0 & 0 & 1 \end{pmatrix}\begin{pmatrix} x \\ y \\ 1\end{pmatrix}$$
- We then obtain two equations (the two constraints)
$$ a_{11}x_1 + a_{12}y_1 + v_x = x'_{1} $$ 
$$ a_{21}x_1 + a_{22}y_1 + v_y = y'_1 $$

- Since we have 6 unknowns and each point defines two constriants, we need the mapping of exactly 3 points to uniquely define an affine transformation in 2D, to uniquely define an affine transformation. Since this will provide 6 linear equations to solve for the unknowns.
