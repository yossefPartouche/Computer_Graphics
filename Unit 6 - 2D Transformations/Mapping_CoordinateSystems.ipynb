### Mapping Coordinate Systems 

#### Q. Why do we need to learn about Transformations?

#### A. Consider the following 2D Example
In the World of digital print: Letter are described using mathematical curves, which define their respective shape, and the curves are define in a local coordinate system that represent the shape of each letter, and when printed the white space inside the letter is filled in black ink. 
We still need to place that shape in the correct position on the page. 
this is an example of mapping a letter from a local coordinate system (of the letter) to a global coordinate system (Text). The page (global) coordinate system is usually upside down.

How do we define the mapping mathematically?  

we'll define the local coordinate system with $Axis \ u $ (Equivelant to $x$) and $Axis \ v$ (Equivelant to $y$) 
- For each letter in this system is deined in the bounding box (i.e. in the range) $[0, u_1]*[0, v_1]$. 
- Suppose we know the global position of this letter $(x_0, y_0)$. 
- The size of the letter (In local) must be adjusted in the global page.  
Assume we know this range of values, in which we deduced the size to be between $x_0, \ x_1$ and $y_0, \ y_1$:
The scale process is $$ S_x = \frac{x_1-x_0}{u_1-0}$$ $$ S_y = \frac{y_1-y_0}{v_1-0}$$ 

Note: In the denominator it generalises to the form $u_1 -u_0$ and $v_1 -v_0$ for and $x$ and $y$ respectively.


So The general Process is: 
1. Translation Correct starting position of the letter in the page, $(u_0, v_0)$ to $(x_0, y_0)$
2. Scaling: The width of the letter, $(u_0, u_1)$ to $(x_0, x_1)$
3. Scaling: The hieght of the letter,  $(v_0, v_1)$ to $(y_0, y_1)$  
And is summarised with: $$x = x_0 + uS_x$$ $$y = y_0 + vS_y$$

The Transformation can be done via vector operations (When we say vector we also mean matricies):
\begin{align*}
\begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + \begin{pmatrix} S_x & 0 \\ 0 & S_y \end{pmatrix} \begin{pmatrix} u \\ v \end{pmatrix}
\end{align*}  

\begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + \begin{pmatrix} e_{11} & e_{12} \\ e_{21} & e_{22} \end{pmatrix}\begin{pmatrix} u \\ v \end{pmatrix} 

\begin{align*}
= \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} + \mathbf{A} \begin{pmatrix} u \\ v \end{pmatrix}
\end{align*}

Note: If the the two vector $u$ and $v$ are not orthonormal to another they it will have some distortional affect on the letter (italix font for example).
