### Viewport Transformations

##### Objective:
- We will discuss the final transformation which will convert the coordinate sysmtem of the object in the veiwing window into the device coordinate systems in the view port

- $\text{ Viewport Transformation := Converts the coordinates of 2D shapes inside the window to a devide coordinate system that are actually pixel coordinates of the device which will display the scene}$

- $\text{The viewport := A window that can be place at any poistion within the device screen, where it's defined by the top left coordinate (local/viewport Origin) and the bottom right pixel (max size of the viewport)}$
- $\text{ Our GOAL: To map a 2D scene created by the graphical rendering pipline after clipping all objects into the veiwing window to the viewport on the device.}$
- $\text{The viewing window is also defined by two point as well } (w_{minx}, w_{miny}) \text{ on the bottom left and } (w_{maxx}, w_{maxy}) \text{on the top right }$
- $\text{Viewport Transformation := Map the viewing window to the veiwport an Affne tranformation.}$

### Viewport Transformation Steps

#### Translation
- $\text{Translate the veiwing window to the origin by using the following translation matrix}$
$$
\begin{bmatrix}
1 & 0 & -W_{\min x} \\
0 & 1 & -W_{\min y} \\
0 & 0 & 1
\end{bmatrix}
$$
#### Scaling 

- $\text{Scaling maps the range of value in x and y of the viewing window to the range of values in the veiwport, in the horizontal direction the values i the vewing window are between } (W_{max_x}, W_{min_x}) \rightarrow \text{ viewport } (V_{max_x}, V_{min_x})$
- $\displaystyle \text{The scaling factor in the x direction is } s_x = \frac{V_{max_x} - V_{min_x}}{W_{max_x}- W_{min_x}}$
- $\displaystyle \text{The scaling factor in the y direction is } s_y = -\frac{V_{max_y} - V_{min_y}}{W_{max_y}- W_{min_y}}$
- $\text{Note that we add a minus sign in the y-scale as the coordinates in the viewport is flipped}$

- $\text{producing the folowing matrix}$
$$
\begin{bmatrix}
\displaystyle
\frac{V_{max_x} - V_{min_x}}{W_{max_x}- W_{min_x}} & 0 & 0 \\
0 & \displaystyle-\frac{V_{max_y} - V_{min_y}}{W_{max_y}- W_{min_y}} & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

#### Translation

$\text{We want to map the viewport to its position in the devicee in the screen:}$
$\text{In the x-direction this would simply mean a translation by } V_{min_x} \text{ but in the y direction the coordinates are flipped so we produce the following matrix}$
$$
\begin{bmatrix}
1 & 0 & V_{\min x} \\
0 & 1 & V_{\max y} \\
0 & 0 & 1
\end{bmatrix}
$$

### Together:
$$
\begin{bmatrix}
1 & 0 & V_{\min x} \\
0 & 1 & V_{\max y} \\
0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
\displaystyle
\frac{V_{max_x} - V_{min_x}}{W_{max_x}- W_{min_x}} & 0 & 0 \\
0 & \displaystyle-\frac{V_{max_y} - V_{min_y}}{W_{max_y}- W_{min_y}} & 0 \\
0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
1 & 0 & -W_{\min x} \\
0 & 1 & -W_{\min y} \\
0 & 0 & 1
\end{bmatrix}
$$



### Aspect Ratio
- $\text{The aspect ratio of the viewing window is:}$
$$ \displaystyle \frac{W_{max_y} - W_{min_y}}{W_{max_x}-W_{min_x}}$$
- $\text{The aspect ratio of the veiwport is:}$
$$ \displaystyle \frac{V_{max_y} - V_{min_y}}{V_{max_x}-V_{min_x}}$$

- $\text{If the aspect ratio of the viewing window is different from the aspect ration of the viewport window the viewing transformation will also create a squeeze or a stretch of the image}$
