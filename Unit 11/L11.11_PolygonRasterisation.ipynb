## Polygon Rasterization

#### Objective: 
- $\text{Extent the notion of rasterisation beyond triangles and to general polygons}$


$\text{The simplest way to rasterize a polyon is via "Divide and Conquer" method}$
$\text{We know how to raster traingles and we know that any polygon can be divided into subshapes of triangles}$
$\text{So if we divde the polygon into triangles and then rasterise each triangle, so long as we stay consistent in the process especially on the edges, then we should be able to rasterise the whole polygon}$

#### Flood Fill -  Simple Polygons (No-self intersections + No Holes)
$\text{Idea: you pick a color and point inside the polygon and then the polygon is flooded with this color} \\$
$\text{Idea: You start with a point inside the polygon and you look at its neighbors:} \\ $
    - $\text{If they are note boundary pixels and not colored already, you color them and repeat the process on the neighbors}$

$\text{If we're not careful about the polygon it's possible to fill the while veiwing window with that color}$

```
FloodFill (polygon P, pixel, (x,y), color c){
    if (OnBoundary(x, y, P), || Color(x,y) = c){
        return ; //Don't color
    } else {
        SetPixel(x,y,c); 
        FloodFill(P, x+1, y, c);
        FloodFill(P, x-1, y, c);
        FloodFill(P, x, y+1, c);
        FloodFill(P, x, y-1, c);
    }
}
```

### $\text{Main Problems with Flood Fill and Solution}$
- $\text{The recursive call doesn't control repetion of a pixel being checked, which leads to a slow and inefficient running time.}$
- $\text{In Convex polygons, we described a method using the implicit equation of lines defining the boundary of the polygon} \\ \text{and also a normal vector to the edge for each edge of the polygon. We then defined and consistently check if a point lies on the positive side or negative side of the edge (point classification in implicit equations).}$
- $\text{For General polygon, given a perticular point p}  \in R^2, \text{ we can choose a direction (any you want) and define a ray (unbounded ray) then count the number of edges crossings of the edges of the polygon.}$
$\text{If we find an odd number of crossings we know that p is inide the polygon (1 cross =  in, 2 cross = out, 3 cross = 1, 4 cross = out ...).} \\ \text{In fact for more robus results we'll have multiple rays in different directions (odd number) and then choose the mojority of the conclusion to classify the point,} \\ \text{this is so that we don't have any edge case of hitting a vertex (= 2 edges) or parallel and on an edge}$
