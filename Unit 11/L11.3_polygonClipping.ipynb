### Polygon Clipping Algorithm
$\text{Suppose we have a polygon that is partially inside the region}$
$\text{A polygon boundary is composed of connected segments, we may think to apply the algortihms learnt in Lesson 11.2 however the following example shows otherwise}$


<p align="center">
    <img src="image_U11/Screenshot 2025-06-20 at 1.10.18.png" height="400" width="900"/>
<p/>

$\text{The reason for this effect is because the polygon may not only be defined by the intersection points of the segmens}$
$\text{Since we know that the window lies on one side of every edge, as such any clipped part of a given shape must lie on this side as well}$
$\text{We'll clip the polygon againts each edge seperately one by one, doing this will guarantee that each shape that is clipped is a subshape of the original shape, and that they always lie on the correct side of the edge}$
$\text{As soon as we've iterated through each side of the viewing window and clipped the remaining shaped must be the clipped shape inside the region}$

$\text{Our problem has now beeen reduced to clipping a polygon againts an edge}$

1. $\text{Let } P_i \in \R^2 \text{ ordered vertices where each consecutive pair of vertcies defines an edge i.e. } \left((P_1, P_2), (P_2, P_3), (P_3, P_4), (P_4, P_5), (P_5, P_1)\right)$
2. $\text{We loop around the the defined order of edges, and determine which of the vertices are inside the region where the window lies and which are outside}$
3. $\text{To determine if a vertex lies on a perticular side of the edge line (inside or outside), we make sure that the normals of the edge point towards outside and the use an implicit representation of a line using the equation}$
$\text{If we assign the vertex coordinates to the implicit line equation, and get a negative value, the vertex is inside the region otherwise the vertex is outside}$

$ \text{Let v = current vertex and u = previous vertex where we define the vector} \vec{UV} = v - u $
$\text{if v is outside the clipping region and u was also outside the clipping region then there's nothing to add to the polygon.}$
$\text{If v is inside the clipping region and u was also inside the clipping region then we simply add v to the new polygon, and there's no need to add u since it was added in the previous step.}$
$\text{If v is outside the region and u was inside the region it means that the end between them crossed the boudnary from inside to outside. As such we find the intersection point }  w \in \R^2 \text{and add it to the new polygon.}$
$\text{if v is inside the region and u was outside the region it means that the end between them crossed the boudnary from OUTSIDE to INSIDE. We must find the intersection point w, in this case we add both w and v to the new polygon.}$

### Sutherland and Hodement Clipping PsuedoCode 

<pre>
Initialise: 

L_{in} = Input Polygon Vertices 

L_{out} = \emptySet

For each window edge e {

    u = getLastVertex(L_in)

    Loop over v \in L_{in} {

        if v is inside relative to e then:
            if u is outside relative to e 
                then L_out = L_{out} \UNION intersect(e, <u, v>)
            L_{out} = L_{out} \UNION {v}
        Else
            if u is inside relative to e
                then L_{out} = L_{out} \UNION intersect(e, <u,v>)
        u = v
    }
    L_{in} = L_{out}
    L_{out} = \emptySet
    return L_{in}
}
<pre>

- $\text{The same algorithm will work over any convex polygon}$
- $\text{Clipping any polygon againts a convex polygon can be done by clipping it againts all of the convex polygon's edges}$
- $\text{Note that the resttriction were only placed on the clipping polygon and NOT the Clipped polygon, which means we can clip any polygon (convex or not) againts a convex polygon window using the algorithm}$
- $\text{Int eh graphical pipeline, we'll always be clipping againts a rectangle window, and only clipping triangles thus we'll always have a single polygon after clipping process}$
