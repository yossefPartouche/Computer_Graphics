### Properties of Polygon Clipping
$\text{Since we know that the window lies on one side of every edge, as such any clipped part of a given shape must lie on this side as well.}$

$\text{We'll clip the polygon againts each edge (of the viewing window) seperately one by one.} \\ \text{Doing this will guarantee that each shape that is clipped is a subshape of the original shape nd thus  always lie on the correct side of the edge.}$

$\text{As soon as we've iterated through each side of the viewing window and clipped the remaining shaped must be the clipped shape inside the region}$

### Polygon Clipping

$\text{We reduce the problem of polygon clipping to clipping a polygon against one edge of the clipping window. This process is repeated for all edges of the window.}$


#### Step-by-Step Algorithm:

1.	$\text{Input Polygon: Let the polygon be defined by a sequence of ordered vertices:}$
$$
P_1, P_2, P_3, \dots, P_n \in \mathbb{R}^2
$$
- $\text{Assume: Each consecutive pair defines an edge of the polygon:}$
$$
(P_1, P_2), (P_2, P_3), \dots, (P_{n-1}, P_n), (P_n, P_1)
$$

2. $\text{Select a Clipping Edge}$

- $\text{We clip the polygon against one edge of the window at a time (e.g., left, right, top, bottom).}$

3.	$\text{Determine Inside vs. Outside}$

- $\text{For each vertex, determine whether it lies inside or outside the clipping region defined by the current edge.}$
- $\text{Represent the edge implicitly using the line equation:}$
$$
a_ix + b_iy + c_i = 0
$$

- $\text{Let } \vec{N}_i = (a_i, b_i) \text{ be the outward-facing normal of the edge.}$
- $\text{For a point } P = (x, y)$:
    - $\text{ If } a_ix + b_iy + c_i < 0, \text{ then P lies inside the edge.}$
	- $\text{Otherwise, P lies outside the edge.}$

4. $\text{Process Each Polygon Edge}$

- $\text{For each edge of the polygon } (P_k, P_{k+1}), \text{ perform the following:}$
	- If both $P_k$ and $P_{k+1}$ are inside, add $P_{k+1}$ to the output list.
	- If $P_k$ is inside and $P_{k+1}$ is outside, make the intersection point = of the polygon edge with the clipping edge as the new vertex of the polygon.
	- If $P_k$ is outside and $P_{k+1}$ is inside, add the intersection point and $P_{k+1}$.
	- If both are outside, add nothing.

5.	$\text{Repeat for All Edges of the Clipping Window}$

- The output of one edge becomes the input polygon for the next edge. After all four edges, the result is the final clipped polygon.

### Vertex Handling in Polygon Clipping (e.g., Sutherland–Hodgman)

$\text{Let v = current vertex and u = previous vertex where we define the vector} \vec{UV} = v - u $

$\text{We examine each edge (u, v) of the polygon against the clipping boundary:}$

|Cases | Action | Reason| Visual|
|------|--------|-------|-------|
|Both u and v are inside| Add v to the output polygon| Entire edge lies inside. |<img src="image_U11/Screenshot 2025-06-20 at 1.28.07.png" height="200" width="200"/>|
|Both u and v are outside | Add nothing |  Entire edge lies outside. |  <img src="image_U11/Screenshot 2025-06-20 at 1.26.50.png" height="200" width="200"/>|
|u is inside, v is outside | Add the intersection point w to the output polygon | Edge exits the clipping region. |  <img src="image_U11/Screenshot 2025-06-20 at 1.30.03.png" height="200" width="200"/>|
|u is outside, v is inside | Add both **intersection point** w and **vertex** v.| Edge enters the clipping region. |<img src="image_U11/Screenshot 2025-06-20 at 1.30.45.png" height="200" width="200"/> |

- $\text{In the algorithm we make a descision on the points since we iterate between consecutive points which are edges of the polygon}$
- $\text{In the algorithm case where the edge partially in the plane, we consider the direction of the the edge}$


### Sutherland and Hodgement Clipping PsuedoCode 

```cpp
Initialise: 

L_{in} = Input Polygon Vertices 

L_{out} = \emptySet

For each window edge e {

    u = getLastVertex(L_in)

    Loop over v \in L_{in} {

        if v is inside relative to e then:
            if u is outside relative to e 
                then L_out = L_{out} \UNION intersect(e, <u, v>)
            L_{out} = L_{out} \UNION {v}
        Else
            if u is inside relative to e
                then L_{out} = L_{out} \UNION intersect(e, <u,v>)
        u = v
    }
    L_{in} = L_{out}
    L_{out} = \emptySet
    return L_{in}
}
```
