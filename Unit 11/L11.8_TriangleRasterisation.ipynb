- $\text{If all thee barycentric coordinates of the point are non-negative, the point is inside the triangle, otherwise it's outside}$

- $\text{We could also use the convexity properties of a triangle to detetmine if a point lies inside the circle}$
- $\text{We'll use an implicit representation of the lines to represent the edges of a triangle, and store the normal of each edge (all pointing outside or all inside)}$
- $\text{You evaluate the three implicit functions (of lines) at the given point}$

$$ f_1(x,y) = a_1x + b_1y + c_1 $$
$$ f_2(x,y) = a_2x + b_2y + c_2 $$

- $\text{And the we check the following:}$
$$ f_1(x,y) < 0, \ f_2(x,y) < 0 \ and \ f_1(x,y)<0 \rightarrow (x,y) \text{ in the triangle}$$

<pre>
Given Triangle T,
Find the bounding-box B of T
    for each pixel p = (x,y) inside B {
        if (PoinInside(T, p))
        SetPixel(x, y);
    }

#### Optimisation #2

- $\text{We can optimise the algorithm, by finding only the pixels inside the triangle}$
- $\text{Since we know how to rasterise the edges of the triangle, we can use this boundary to rasterise the whole triangle}$
- $\text{Notice that a rasterised triagnle is constructed from a list of rows of pixels each row spans all pixels between the left side and the right side}$
$\text{Scanline Algorithm:}$

<p align="center">
    <img src="image_U11/Screenshot 2025-06-20 at 18.23.44.png" height="400" width="800"/>
<p/>

1. $\text{We start at the top of the triangle}$
2. $\text{Move down to the left and right along the edges where in each row, we find the pixels on the left and on the right edges of the triangle}$

<pre>
Draw top vertex (xt, yt) in Triangle T
Initialise left and right edges EdgeL and EdgeR 
y = yt
for each scanline y do {
    xL = Rasterise the next pixel of EdgeL in y
    xR = Rasterise thenext pixel of EdgeR in y
    if both EdgeL and EdgeR end => stop
    if EdgeL ends, then switch EdgeL to next edge.
    if EdgeR ends, then switch EdgeR to next edge.
    for (int x = xL; x <= XR; x++){
        SetPixel (x,y);
    y++
    }
}
<pre>
