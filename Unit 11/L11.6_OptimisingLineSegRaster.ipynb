### Optimizing Line Segment Rasterization

#### Objective:
- $\text{Discuss the complexity of line segment rasterisation algorithm}$
- $\text{Discuss how to improve the algorithm}$

#### Case 1: When $ 0 < m < 1 $
- $\text{In this case we iterate in the x-direction and caluculate the y-value of the pixels}$
#### Original Implementation

```cpp
float m = (y1 - y0) / (x1 - x0);

for (x = x0; x <= x1; x++) {
    y = m * x + b;
    SetPixel(x, Round(y));
}
```


##### Cost
1. $\text{Multiplication}$
2. $\text{Addition}$
3. $\text{Rounding}$

- $\text{Applying these operation over every pixel will sum up to a high complexity}$

- $\text{ We'll use the following representation of grid intersection to represent the pixels instead the centre of the big squares}$
- $\text{In the image below, notice that the line segment is continous, so there will exist many point between the pixels before rasterising}$

- $\text{We still move a constant step in the x direction by adding +1 to the previos x value}$
- $\text{Since we're dealing with a linear function we know for every 1 unit moved in the x-direction we move m units in the y-direction}$
- $\text{This means we only need to add m to the previous y value at each step of the iteration instead of computing from scratch}$

#### Optimization #1

```cpp
float m = (y1 - y0) / (x1 - x0);
float y = y0

for (x = x0; x <= x1; x++) {
    SetPixel(x, Round(y))
    // instead of   
    // y = m * x + b;
    // SetPixel(x, Round(y));
    y += m;
}
```



#### Optimisation #2

- $\text{As seen in the first chapter, this is another optimization example known as forward differecing}$
- $\text{Note: We still have some optimisation problem with the rounding operation which is still expensive}$

$\text{We try to solve this by noticing that the pixel positions are represented by the intersections of the grid lines meaning:}$

$\text{For some } x_i \text{ in iteration i during the drawing we mark the } y_i \in \R \text{ and the rounded value of the pixel drawn is } Round(y_i)$

$\text{The accumelated error would be } e = Error_{i} = y_i - Round(y_i) \text{ and }  |Error_{acc}| < 0.5$

$\text{This helps us because we know that } y_{i+1} \text{ changes by m (as shown above) so } Error_{i+1} = y_{i+1} - Round(y_{i+1}) = e + m $

$\text{Note we must check if the error exceeds 0.5 if it doesn't then we accumelate on the same change of y as before otherwise we apply 1 - y}$

#### Optimisation #3

- $\text{Notice that we haven't computed any rounding operations, used an integer starting point and from there incremented either the x value by 1 or the y value by 1}$
- $\text{Having said that still have a number of addition and subtraction operation occuring in the for loop}$
- $\text{ We can get rid of all floating points arithmatics by removing the float dx, we do this by multiplying all values in the algorithm by dx this means:}$
1. $ m = \frac{dy}{dx} \rightarrow m = \frac{dy}{1}$
2. $ err = m \rightarrow err = err * dx$
- $\text{We now fix the terms}$
1. $ err = err + m \rightarrow err = err + dy$
2. $ err > 0.5 \rightarrow err > \frac{dx}{2} $
3. $ err = err -1 \rightarrow err = err - dx$

|concept | Floating-Point | Integer Version|
|--------|----------------|----------------|
|Slope | $\frac{dy}{dx}$ | $dy$ (scale by dx)|
|Error increment| ```err += m``` | ```err += dy```|
|Threshold check | ```err > 0.5 ``` | ```err > dx/2```|
|Error reset | ```err -= 1 ``` | ```err -=dx ``` |

``` cpp
int dy = (y1 - y0);
int dx = (x1 - x0);
int err = 0;
int over  = dx/2
y = y0

for (x = x0; x <= x1; x++) {
    SetPixel(x, y);
    err =+ dy;
    if (err > over) then {
        y++
        err = err - dx;
    }
}
```
