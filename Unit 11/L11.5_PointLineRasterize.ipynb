## Point and Line Rasterisation
#### Objective: 
- $\text{Learn about the conversio of 2D clipped points and lines into pixels}$


- $\text{To rasertize any 2D shape we need to answer two questions:}$
1. $\text{Which pixels to we need to draw}$
2. $\text{What colors we need to assign to the pixels}$

- $\text{Assume we have a procedure called }$ `SetPixel()` $\text{which recieces 3 parameters: 1. The column x 2. The row y 3. The color value}$
- $\text{This proceedure will assign the given color to the pixel (x,y). This will be the basis proceedure in all rasterisation algorithms}$
- $\text{ASSUMPTION: the starting/origin coordinate of the pixel starts at the bottom left (though as we saw it's at the top Left)}$
- $\text{ASSUMPTION: We will also ignore the assignment of the color value for now}$

### Point Rasterisation

- $\text{A point in a device coordinate system will simply have two pixel coordinate (x,y)}$
- $\text{To rasterise a point we only need to "turn on" the pixel at coloumn x and row y }$
- $\text{Note: The difference between pixel point and mathematical point is that a pixel occupy some space and have dimensions whereas a mathematical point doesn't}$

### Line Rasterisation 
- $\text{A line segment is defined by two end points: } (x_0, y_0) \ \ (x_1, y_1)$
- $\text{ASSUMPTION: the line isn't verticle thus the gradient is defined: } m = \frac{y_1-y_0}{x_1-x_0}$
- $\text{if the line was verticle then } x_0 = x_1 \text{ and we draw each pixel between } y_0 \ - \ y_1$
- $\text{Similarly if the line was horizontal then } y_0 = y_1 \text{ and we draw each pixel between } x_0 \ - \ x_1$
1. $\text{We start by "turning on" pixel at  }(x_0, y_0)  \text{As it's a point on the line}$
2. $\text{We then move } x_0 + 1 \text{ and using the equation of the line (defined over the initialising points) to determine } y_{x_0+1}$
3. $\text{We then move } X_0 + 2 \text{ and using the equation of the line to determine } y_{x_0+2}$
4. $\text{We continue this process until we've completed the line segment}$
- $\text{NOTE: that the equation of the line is defined over } \R \text{so after each y computation we round to integer value}$

#### Careful

- $\text{if the gradient of the line is too steep for integer changes of x values will skip values of y thus not create a pixelised line segment}$
- $\text{So we check the gradient value and choose a threshold like angle is > 45 i.e. m > 1 then we iterate over y}$

##### Final Line segment algorithm 

` 
if (x_0 > x_1 and -1 <= m <= 1>) or (y_0 > y_1 and (m >1 or m < -1>)) then Swap ((x_0, y_0), (x_1, y_1))
