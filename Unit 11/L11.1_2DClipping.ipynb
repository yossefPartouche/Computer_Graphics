## 2D Clipping

- $\text{Right now imagine you have a plane (infinite plane) which contains a square window that contain 2D images inside the window and parts are outside the window and some objects are a bit of both}$
- $\text{We will present clipping algorithms for clipping:}$
1. $\text{points}$
2. $\text{Line segments}$
3. $\text{Polygons}$

- $\text{We now define a rectangle window, assume that the edges are perpendicular and aligned with the axis then we define the edges as follows}$
$$ x = x_{min} \ x = x_{max} \text{ For verticle edges}$$
$$ y = y_{min} \ y = y_{max} \text{For horizontal edges}$$
- $\text{We can there for define the window by its two corner points} (x_{min}, y_{min}) \ (x_{max}, y_{max})$$

### Clipping a point to the Window ✂️

- $\text{To determine whether a point (x,y) lies within the window we just compute } x_{min} \le x \le x_{max} \ and \ y_{min} \le y \le y_{max} $

- $\text{We can further divide the plane into 9 regions by extending the edges of the window to infinity, if we mark each region using one bit we will have a 4 bit code for each region:}$
- $\text{The first left bit marks the left region, the 2nd marks the right rigion the 3rd marks the bottom region and 4th marks the top region}$
- $\text{ the image below describes how to bit configuration describes in which quatre the point is located}$
- $\text{The above described marks the outcode of the a 2D point}$
- $\text{If a point is inside the window the bits of its outcode would all be zeros, otherwise there will exists a least a non-zero bit}$

### Clipping a Line Segments (Cohen-Sutherland) to the Window ✂️

- $\text{A line segment is linear combination of two endpoints}$
- $\text{If both endpoints lie within the window then the line segment lies within the window}$
- $\text{We will now learn about the Cohen-Sutherland line segment clipping}$

1. $\text{The first simple case we deal with is when the line segment lies within the window}$

`Given a line segment from p1 to p2`

`let c1 and c2 be:` 

`c1 = outCode(p1)`

`c2 = outCode(p2)`

`if (c1 || c2 == 0) then the segment is inside the window`

2. $\text{We then check to see if the whole segment lies outside the window by checking the results of the outCode as per the above image}£
$\text{Indeed if both end point lie in the same region the the line segment won't be insde the window, this means that their outCode must share at least a 1-bit}$

`while (not done) {` 

    `if (c1 || c2 == 0) then the segment is inside the window`

    `else if (c1 and c2 != 0) the segement is outside`
    `else (clip an outside part of segment and update the codes)`
`}`

3. $\text{Assuming we have a line segment that crosses the window we need to find the boundary points on the line segment that intersect the window edges and clip the rest}$

### Example 


$ \text{Suppose we have a line segment defined by two points } P_0 = (x_0, y_0) \text{ and } P_1 = (x_1, y_1). $

$ \text{Equation of the line segment:} $

$$
(x(t), y(t)) = (1 - t)(x_0, y_0) + t(x_1, y_1), \quad t \in [0, 1]
$$

---

- $ \text{To find } t, \text{ we proceed when either } x \text{ or } y \text{ is known (e.g., from the boundary of the viewing window).} $

1. $ \text{Check the region in which both points lie using outcodes e.g., LSB order: (Top, Bottom, Right, Left).} $

2. $ \text{Suppose } P_0 \text{ lies in region with outcode } C_0 = 0101 \text{ (Bottom Left) and } P_1 \text{ lies in region with outcode } C_1 = 1000 \text{ (Top Center)} $

$$
\Rightarrow \text{The segment crosses the viewing window and must be clipped.}
$$

3. $ \text{To clip from } P_0, \text{ we check where the segment intersects the left boundary (i.e., use } x_{\min}). $

$ \text{Substitute } x_{\min} \text{ into the parametric equation and solve for } t: $

$$
x(t) = x_0 + t(x_1 - x_0) = x_{\min}
$$

$$
\Rightarrow t = \frac{x_{\min} - x_0}{x_1 - x_0}
$$

$$
\text{Then compute } y(t) = y_0 + t(y_1 - y_0)
$$

$$
\text{Let the new intersection point be } P_{\text{new}} = (x_{\text{new}}, y_{\text{new}}).
$$

4. $ \text{Clip the segment: the new segment is from } P_{\text{new}} \text{ to } P_1. $

5. $ \text{Repeat the process until the line lies fully within the central clipping region.} $
