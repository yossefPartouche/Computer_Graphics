## ✂️ Line Clipping – Example using Parametric Equation

Suppose we are clipping a line segment using the **Cohen-Sutherland** algorithm and the **parametric line equation**.

### 🔷 Given

We have a line segment defined by two endpoints:  
- $P_0 = (x_0, y_0)$ 
- $P_1 = (x_1, y_1)$ 


### 🧮 Parametric Equation of the Line Segment

$(x(t), y(t)) = (1-t)(x_0, y_0) + t(x_1, y_1)$


### 🚪 Step-by-Step Clipping Procedure

1. **Compute the outcodes** for $P_0$ and $P_1$ using the region codes:  
   Format: **[Top, Bottom, Right, Left]**

2. **Suppose**:  
   - $P_0$ has outcode $C_0 = 0101 \rightarrow$ Bottom Left  
   - $P_1$ has outcode $C_1 = 1000 \rightarrow$ Top Center  
   ✅ This means the segment crosses the clipping window and must be clipped.

3. **Find intersection with the left boundary** (`x = x_min`):
   - Solve for `t`:
     ```
     x(t) = x₀ + t(x₁ - x₀) = x_min
     ⇒ t = (x_min - x₀) / (x₁ - x₀)
     ```
   - Then compute:
     ```
     y(t) = y₀ + t(y₁ - y₀)
     ```
   - Gives new point:
     ```
     P_new = (x_min, y(t))
     ```

4. **Replace P₀ with P_new**, update its outcode, and repeat for the next boundary if needed.

5. When both endpoints have outcode `0000`, the segment lies fully inside the window and can be drawn.

## 🧪 Example Problem

$\text{Problem: Clip the line segment from:}$

$P_0 = (5, 0)$

$P_1 = (20, 60)$

$\text{ against a rectangular clipping window:}$

$$
x_{\min} = 10 \ 
x_{\max} = 50
$$
$$

y_{\min} = 10 \ 
y_{\max} = 40
$$

#### 🧮 Step 1: Compute Outcodes

$ P_0 = (5, 0) \\ Outcode(P_0): \\ x = 5 < x_{min} = 10 → LEFT → 0001 \\ y = 0 < y_{min} = 10 → BOTTOM → 0100 → C_0 = 0101$

$\\  P_1 = (20, 60)\\ Outcode(P_1): \\ y = 60 > y_{max} = 40 → TOP → 1000 → C_1 = 1000 \\$

#### Step 2: Check Trivial Accept/Reject

$$ C_0 \mid C_1 = 0101 \mid 1000 = 1101 \neq 0 \rightarrow \text{not trivially inside} $$  


$$ C_0 \& C_1 = 0101 \& 1000 = 0000 \rightarrow \text{not trivially outside}  $$ 



#### ✂️ Step 3: Clip P_1 to Top Edge

$ \text{We intersect with the top edge } y = y_{\max} = 40:$ 

$ \\ y(t) = y_0 + t(y_1 - y_0) = 0 + t(60 - 0) = 60t $

$40 = 60t \Rightarrow t = \frac{2}{3}$

$\text{Now compute x(t):}$

$x(t) = x_0 + t(x_1 - x_0) = 5 + \frac{2}{3}(20 - 5) = 5 + \frac{2}{3} \cdot 15 = 5 + 10 = 15$

$\text{New point: } P_1’ = (15, 40)$

⸻

#### ✂️ Step 4: Clip P_0 to Bottom/Left Edge

$\text{We now clip against the left edge } x = 10:$

$x(t) = x_0 + t(x_1 - x_0) = 5 + t(15 - 5) = 5 + 10t$

$10 = 5 + 10t \Rightarrow t = \frac{1}{2}$

$\text{Then:}$

$y(t) = y_0 + t(y_1 - y_0) = 0 + \frac{1}{2}(40 - 0) = 20$

$\text{New point: } P_0’ = (10, 20)$
