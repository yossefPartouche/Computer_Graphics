### More Line segment Clipping 

#### Objective: 
- $\text{Learn a more efficient Line segmement clipping algorithm, which also allows extension beyong clipping to a rectangle window}$

- $\text{The Cohen Sutherland Algorithm is an iterative algorithm that uses repetitive clipping of the line segments}$
- $\text{The idea of the new clipping algorithm is to:}$
1. $\text{find all intersection points of the line with the window edges.}$
2. $\text{Classify them}$
3. $\text{Determine directly the part of the line segment inside the window, if there is any.}$

- $\text{We will now define the line segment from } P_0 \text{ to } P_1 \text{ parametrically by:}$
$$ p(t) = P_0 + t\vec{V} = P_0 + t(P_1-P_0) \mid t \in [0,1]$$

### Base Case Check - Parallel segment

- $\text{We can check that the line segment is parallel to an edge by:}$

$$ \vec{N_{i}} \cdot \vec{V} = 0$$ 

$\text{If the line segmement is indeed parallel We use the bit-outCode to determine if it crosses the veiwing window otherwise reject}$

---

### General Case

$\text{Assuming that the line isn't parallel to any of the edges, we'll represent a perticular edge (veiwing window) implicitely:}$
1. $\vec{N_i} = (a_i, b_i)$
2. $q_i = P_{min} \ or \ P_{max}$

$$ a_ix + b_iy + c = 0 = \vec{N_i} \cdot (\vec{P} - q_i) = 0$$

$\text{We now have:} \\ \text{1. An implicit representation of the edge lines} \\ \text{2. The parametric representation of the line segment}$

$\text{1. Extend the line segment to a ray:}$ $$ p(t) = P_0 + t\vec{V} = P_0 + t(P_1-P_0) \mid t \in [0, ∞)$$ 

$\text{2. Look at the intersection between the ray and the edge line}$

- $\text{substitute p(t) for P}$
$$\vec{N_i} \cdot ((P_0 + t\vec{V}) - q_i) = 0$$

$$ \vec{N_i} \cdot (P_0 - q_i + t\vec{V}) = 0$$

- $\text{Distribute the dot produce}$

$$\vec{N_i} \cdot (P_0 - q_i) + t(\vec{N_i} \cdot \vec{V}) = 0$$

- $\text{solve for t:}$

$$ t_i = \frac{\vec{N_i} \cdot (q_i - P_0)}{\vec{N_i} \cdot \vec{V}} = \frac{\vec{N_i} \cdot (P_0 - q_i)}{-\vec{N_i} \cdot \vec{V}}$$

#### Interpretation of the Result 

$\text{This formula gives you the value of t at which the line hits edge i}$

| $\vec{N_i} \cdot \vec{V} < 0$| $\text{Line is Entering through Edge i}$ | $\text{The angle between the two vectors is } \gt 90$ | <img src="image_U11/Screenshot 2025-06-20 at 0.08.34.png" height="300" width="300"/>|
| $\vec{N_i} \cdot \vec{V} = 0$ | $\text{Line is Parallel to Edge i}$ | $\text{The angle between the two vectors is } = 90$ | |

$\text{3. You collect all entering and exiting } t_i \text{ values across all the Edge i and find:}$

$$ t_E = \max(\text{all entering } t_i)$$
$$t_L = \min(\text{all leaving } t_i)$$

$\text{4. Consider the cases:}$

- $\text{If } t_E > t_L \rightarrow \text{the line is OUTSIDE}$
- $\text{IF } t_E \le t_L \rightarrow \text{then the visiable portion of the line is INSIDE}$
$$ p(t), \ t \in [t_E, t_L]$$
