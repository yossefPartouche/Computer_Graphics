### More Line segment Clipping 

#### Objective: 
- $\text{Learn a more efficient Line segmement clipping algorithm, which also allows extension beyong clipping to a rectangle window}$

- $\text{The Cohen Sutherland Algorithm is an itertive algorithm that uses repetitive clipping of the line segments}$
- $\text{The idea of the new clipping algorithm is to:}$
1. $\text{find all intersection points of the line with the window edges.}$
2. $\text{Classify them}$
3. $\text{Determine directly the part of the line segment inside the window, if there is any.}$

### Set up and Algorithm
- $\text{We extend the properties of the edges of the window, by defining 4 Normal vectors points outside from the window}$

- $\text{We will now define the line segment from } P_0 to P_1 \text{ parametrically by:}$
$$ p(t) = p_0 + t\vec{V} = p_0 + t(p_1 p_0)$$

- $\text{We can check that the line segment is parallel to an edge by computing the dot product with a respective normal and determining if it's 0}$

$$ \vec{N_{i}} \cdot \vec{V} = 0$$ 
$\text{ if the line segmement is indeed parallel then we check the simple case of checking the portion of the line being inside the central region}$
- $\text{Assuming that the line isn't parallel to any of the edges, we'll represent a perticular edge line in the implicit form we need:}$
1. $\vec{N_i}$
2. $q_i = P_{min} \ or \ P_{max}$
- $\text{Now that we have an implicit representation of the edge line and the parametric representation of the line segment, we look at the intersection between the line segement (extended to a line ray) and the edge line (also extended to line ray) }p(t)$
$\text{We create a vector with the relative corner point} p(t) - q_i \text{ This vector must be perpendicular to one of the Normals since } q_i \text{ and p(t) lie on the edge}$ 
- $\text{We define an equation that can be used to find t at the intersection point:}$
$$ \vec{N_i} \cdot (p(t) - q_i) = 0 \rightarrow t_i = \frac{\vec{N_i}\cdot(p_0-q_i)}{-\vec{N_i}\vec{V_i}}$$

- $\text{We will now distinguish between two types of intersection points, points where the line goes into the window region and points where the line goes out of the window region, assume the direction of the line is } P_0 \rightarrow P_1$
- $\text{To determine mathematically whether an intersection point is going into the region or out, by looking at the sing of the dot produvt between V and and } N_i $
- $\text{So if V•N ie negative the angle between both vectors is larger than 90 detrees and the segment goes into the window and if V•N is positive then the angle between between the two vector is less than 90 degress and the intersection is an out point}$

- $\text{We have now identified all the intersection points, and using the paramteric equation we plug each point and obtain t and then sort each t increasing order from whch we determin the clipping portion that will be inside the windpw}$
- $\text{We will now bound the t values to have a line segment that is stricly inside the region, this is done by taking the LARGEST (t-Begin) t-value going in and the SMALLEST t-value (t-End) going out}$
