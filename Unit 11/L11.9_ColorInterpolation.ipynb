### Color Interpolation with Line segments

##### The problem
- $\text{consider the case, where we must rasterise a line segment given two end points } p_0, p_1 \text{ and the rasterised line pixel segment has each pixel with a different color}$
- $\text{How do we assign the correct color for each pixel? Even if we know which pixels to draw to rasterise the line segment, how would we know what color to give each pixel?}$
##### The solution
- $\text{Idea: Use linear interpolation of two colors along the segment}$
- $\text{Use the distance from each of the vertices as a weighing factor to combine their colors.}$
- $\text{Every point p along the segment between} p_0 \text{ to } p_1 \text{ can be represented as a linear combination of } p_0 \text{ and } p_1$
$$ p = (1-\alpha)p_0 + \alpha p_1$$
- $\text{Where } \alpha \text{tells us how close we are to } p_1 \text{ and } 1 - \alpha \text{ tells us how close we are to } p_0$
- $\text{Let } c_0, c_1 \text{be the colors of } p_0 \ and \ p_1 \text{ respectively then the color of a point p on the line }  p = (1-\alpha)p_0 + \alpha p_1 \text{ between them would be: }$
$$ c = (1 - \alpha)c_0 + \alpha c_1$$

##### Meaning behind interpolating colors

- $\text{To interpolate colors all we need to do is interpolate the coordinates of the color space representation one by one}$
##### RGB 
$c_0 = (R_0, G_0, B_0), \text{ is the color of } p_0 \text{ and } c_1 = (R_1, G_1, B_1), \text{ is the color of } p_1 \text{ then the color of a point } p = (1-\alpha)p_0 + \alpha p_1 \text{would be:}$
$$ c = (1 - \alpha)(R_0, G_0, B_0) + \alpha(R_1, G_1, B_1) = ((1 - \alpha)(R_0) +\alpha R_1, (1 - \alpha)(G_0) +\alpha G_1, (1 - \alpha)(B_0) +\alpha B_1)$$



#### Triangle Color interpolation

- $\text{REMINDER: The second step in the rendering pipeline is called "Lighting" and it involves the calculation of the color of the verticies of the triangles in the scene that are sent to the rendering pipline.}$
$$ I = I_E + K_AI_A + \sum_j{S_j(K_D I_{L_j}(N \cdot L_j)+ K_sI_{L_j}(V \cdot L_j)^n)}$$
- $\text{This means that this must've been computed in the stage where the object is in 3D, this means that each vertex of the triangle has a color attribute}$
- $\text{ This means for every triangle we have a 3 tuple of colors, which we need to define the colour of the pixels inside the triangle}$
- $\text{To determine the color of a pixel in the triangle, we will use a weight that depends on the distance of the point inside the triangle from each of the vertices, the smaller the distance the larger the weight}$
1. $ \displaystyle d_1 = \frac{1}{distance(p, p_1)}$
2. $ \displaystyle d_2 = \frac{1}{distance(p, p_2)}$
3. $ \displaystyle d_3 = \frac{1}{distance(p, p_3)}$

- $ \displaystyle w_1 = \frac{d_1}{d_1+d_2+d_3}$
- $ \displaystyle w_2 = \frac{d_2}{d_1+d_2+d_3}$
- $ \displaystyle w_3 = \frac{d_3}{d_1+d_2+d_3}$

##### Edge Cases
- $\text{On the edge between the two vertices, a reasonable restriction would be that the interpolation of colors, should be similar to a simple linear interpolation between the colors of the edges vertices}$
- $\text{This is important especially if we want to merge colors of triangles in a triangle mes, we want the colors of the edges to be the same no matter the side of the triangle mesh we're rasterising}$
- $\text{If we consider the example below we notice that the color of a point on the edge between } p_3 and p_1 \text{will be affected by the color of } p_2$
- $\text{to go even further if we have two connected triangle meshes (as presented below) we could have the color of the point betwen } p_3 \text{ and } p_1 \text{ be affected by } p_4 \text{ as well}$

1. $\text{Combine the colors of the three vertices inside the triangle}$
2. $\text{While being a pixel on the edges of the triangle, becomes a color from the linearly interpolated color of the two connecting vertices}$
3. $\text{Simple to Compute}$

#### $\text{Solution: Barycentric Coordinates of Triangle}$

$ if \ p = \sum_{i=1}^3 {\alpha_i p_i} \ where \ \sum_i \alpha_i = 1 \ \forall \alpha_i \ge 0 $

- $\text{We know that on the edges the barycentric coordinate reduces to a linear interpolation between the edge vertices}$
- $\text{The color at the point p will be } c = \alpha_1 c_1 + \alpha_2 c_2 + \alpha_3 c_3$
- $\text{Since in the resterisation algorithm for a triangle we already defined the barycentric method to find the pixel in a triangle, we can incoorperate another functionality without adding complexity}$
