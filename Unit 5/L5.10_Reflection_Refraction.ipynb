#### Reflection
By Definition: The angle of incident, the angle of reflection and all lie on the same plane, and the angle of reflection equals the angle of incident.
This means that given a ray from a light source hitting a reflective object we can calculate the direction of the relfection ray. 

Denote the follow:

- $\vec{L}$ The normalised direction of the incoming (Light) ray. 
- $\vec{R}$ The normalised direction of the outgoing reflected (light) ray.
- $\vec{N}$ The normalised normal vector. 
- $\vec{R} = -\left (\vec{L} \cdot \vec{N} \right ) \vec{N} + \vec{L} - \left (\vec{L} \cdot \vec{N} \right ) \vec{N}$

The ratio of the sines of the angle of icidence $\theta_1$ and angle of refraction $\theta_2$ is equivalent to the inverse of the ratio of the indices of refraction of the two media $$\frac{n_1}{n_2} = \frac{sin(\theta_1)}{sin(\theta_2)}$$

The indicies of the refraction is depenedant on the matrial type. 

Denote the follow:

- $\vec{L}$ The normalised direction of the incoming (Light) ray. 
- $\vec{T}$ The normalised direction of the outgoing reflected (light) ray.
- $\vec{N}$ The normalised normal vector. 
- $\vec{P}$ THe normallised projection of  $\vec{L}$ on the surface.
$$L = - cos(\theta_1) \cdot \vec{N} + sin(\theta_1) \cdot \vec{P}$$
$$T = -cos(\theta_2) \cdot \vec{N} + sin(\theta_2) \cdot \vec{P}$$
- We produce negative signs since L is point towards the surface and T is pointing away from the direction of N.
