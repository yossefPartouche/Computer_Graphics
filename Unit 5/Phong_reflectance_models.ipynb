#### Modelling Material properties
##### Objective: How to models materialistic properties which effects how we visualize objects in Computer Graphics.

#### Phone-Reflectance Model
Idea: The amount of light that is reflected from a surface on of an object is determined by the following formula: 
$$ Light_{out} = Light_{in} - Light_{absorbed} + Light_{emitted}$$

For now we will discuss objects that aren't light sources: 
This means that it can only reflect light which were emitted from the environment. 
The Reflectance Model characterises the reflections as follows: 
1. Ambient Relection 
2. Diffuse Reflection
3. Specular Reflection 

If you add three types of relections then you can obtain the reflection of light from a non-emitting source.

##### Ambiant 

$$ I_A = K_AI_{AMB}$$

##### Diffuse

Let $\vec{N}$ be the normal of to the object. and let $\theta$ be the angle formed from the reflection of the object and the light source. The larger the angle is the more area is spread across the surface of the object. We can use the $cos(\theta)$ to describe this behaviour. 
$$ I = I_L cos(\theta) = I_L(\vec{N} \cdot \vec{L})$$

The above is a specific case, where the diffusion of the matrial is uniform, due to the matrialistic property of an object (like matt surface).  
The more general form is:  
$$ I_D= K_DI_Lcos(\theta) = K_DI_L(\vec{N} \cdot \vec{L})$$

##### Specular 

We define the specular reflectance property of the surface. 
- $\vec{N}$, The normal vector of the matrial/surface
- $\vec{L}$, vector of light source, (in this L is viewed as the light in the direction to the light source)
- $\hat{L}$ : The Largest specular refletance direction, a mirroring accross the normal N of the vector L pointing to the light source.

If our eyes is exactly no angle difference from $\hat{L}$ then we will see the strongest specular reflection on a shiny surface, the larger the angle change the less the specular reflection. we use $\alpha$ as the additional angle change from $\hat{L}$ and factor the specular reflectance by $cos(\alpha) ^n$ where $n = Shininess factor$ 

The shininess factor means, the large the n, smaller and strong the light intesity of the white reflection. 

$$cos(\alpha) ^n = (\vec{v} \cdot \hat{L})^n$$  
here again the vectors are normalised.

As such we model the specular property as follows: 

$$I_S = K_SI_Lcos(\alpha)^n = K_SI_L(\vec{v} \cdot \hat{L})^n$$

#### Conclusion
When we sum all three property equations we obtain a surface colour with materialistic properties expressed.
