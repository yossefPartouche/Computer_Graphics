#### Anti-Aliasing in Ray Casiting

#### Objective: Discussion of the raster systems with discrete pixels, The Aliasing Problem

##### Setting  
Assume we know how to calculate the color of the closest intersection point of a ray going through a pixel.  

Ray casting will use this colour to define the colour of the pixel, through which the ray was cast. We can view ray casting as taking samples of the scene by shooting rays and finding the intersection point. 

##### Problem
Like any sampling process, ray casting is susceptible to aliasing artifacts

##### Case Study
suppose two objects of different colors are in a line and in one moment we shot a ray and obtained a colour of the first object and stored that in the pixel plane but then moved the object just a bit so that now the ray will only hit the second object (if another ray was shot).

- We can use the smoothing process on images to compensate for this effect, however, this will degrade the image. 
- Through each pixel we can sample many points intersection (for example 4/5 per pixel) and obtain a better understanding of the colour complexity of the scene. We will then find the average of those samples and render the result as the color of this pixel. 
    - Note when we shoot rays at a pixel we shoot at different angles inside the pixel, so either we are given the dimension of a pixel or solve to obtain it and then define different offsets to the centre (topLeft, topRight, BottomLeft, BottomRight, Centre).

- We can obtain more samples (since we're the one creating the scene) either 4 rays, 5 rays or 9 rays per pixel to gain a better understanding on the color of the scene at that pixel.
- once we obtain multiple colors within the pixel, we use either:
1. Simple Averging
2. Weighted Averaging
