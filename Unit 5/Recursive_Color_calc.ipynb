#### Recursive Color Calculation

##### Objectives: Learn the recusrive color calculations used in Ray Tracing




#### Pseudo ALgorithm

1. Shoot a ray from the centre of the camera
2. Hit the closes object A in the scene
3. Calulate color and light intensity from intersection
4. Duplicate ray and perform reflection and refraction ray
5. shoot reay from point of intersection on object A
6. repeat process from 2 until we're at the end of the scope of the scene.

#### Q. How do we calculate the colour of the refraction/Reflection intersections

Add 2 more terms to the EXTENDED pong-reflectance model:

$$
I_{\text{refl}} = I_E + K_A I_{\text{AMB}} + \sum_j s_j\left( K_D I_{L_j} (\vec{N} \cdot \vec{L}_j) + K_S I_{L_j} (\vec{V} \cdot \vec{\hat{R}}_j)^n \right) + K_RI_R + K_TR_T
$$

Where 
$K_RI_R$ - reflective term  
- $I_R$ - Color intensity of reflected ray 
- $K_R$ - How much the color reflects.
$K_TR_T$ - refractive term. (Higher values of $K_R$ $\rightarrow$ more reflective)
- $I_T$ - Color intensity of reflected ray 
- $K_T$ - How much the color refracts. (Higher values $K_T$ $\rightarrow$ more transparent)


### How this affects the GetColor Method:

```
RGB GetColor(Scene scene, Ray ray, Point hit){
    //Ambiant and Emission Calculations
    RGB color = calcEmissionColor(scene) + calcAmbientColor(scene);
    // Diffuse & Specular Calculation
    for (int j = 0; j < getNumLights(scene;) j++){
        Light light = getLight(j, scene);
        float sj = calcShadowFactor(scene, hit);
        color = color + sj*calcDiffuseColor(scene, hit, ray, ligth) + sj*calcSpecularColor(scene,hit,ray,light);
    }
    //Reflective and Refractive components 

>    Ray r_ray = ConstructReflectiveRay(ray, scene, hit); <
>    Point r_hit = FindIntersection(r_ray, scene);        <
>    color += K_R * GetColor(scene, r_ray, r_hit);        <
>    Ray t_ray = ConstructRefractiveRay(ray, scene, hit); <
>    Point t_hit = FindIntersection(r_ray, scene);        <
>    color += K_T * GetColor(scene, t_ray, t_hit);        <

    return color;
}
```

#### Correcting the Algorithm to Make it Recursive


Set level := 1 before the first call i.e. before calling from the camera source

```
RGB GetColor(Scene scene, Ray ray, Point hit, int level){
                                                ^^^^^^^
    //Ambiant and Emission Calculations
    RGB color = calcEmissionColor(scene) + calcAmbientColor(scene);
    // Diffuse & Specular Calculation
    for (int j = 0; j < getNumLights(scene;) j++){
        Light light = getLight(j, scene);
        float sj = calcShadowFactor(scene, hit);
        color = color + sj*calcDiffuseColor(scene, hit, ray, ligth) + sj*calcSpecularColor(scene,hit,ray,light);
    }
    level++;
    if (level > MAX_LEVEL);
        return color;
    //Reflective and Refractive components 

    Ray r_ray = ConstructReflectiveRay(ray, scene, hit); 
    Point r_hit = FindIntersection(r_ray, scene);        
    color += K_R * GetColor(scene, r_ray, r_hit, level);        
    Ray t_ray = ConstructRefractiveRay(ray, scene, hit); 
    Point t_hit = FindIntersection(r_ray, scene);        
    color += K_T * GetColor(scene, t_ray, t_hit, level);        

    return color;
}
```
