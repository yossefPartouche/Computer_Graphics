#### Ray Object Intersection

##### Objective:
Find the closest intersection with scene objects.

##### Reminder: 
We have seen in Unit 3, how to find the intersection between a ray and a triangle $^*$ and a ray and a plane. 

If we assume that our object is represented by a mesh of triangles then:
- Iterate through each triangle and check if there's an intersection with one of them and the ray.

 
##### $^*$ Clarification with the Triangles: 
An object in Computer graphics is made up of triangles (as it's by definition the simplest plane) when the camera shoots a ray from its centre, it will hit the front face of the object at a perticular triangle.  
This is why we are boiling down the problem to an intersection between a ray and a triangle.

### Example of intersections

Suppose are given the implicit representation of a sphere around point $Q$ and radius $r$.   
$$S(Q_x, Q_y, Q_z, r) = (x - Q_x)^2 + (y - Q_y)^2 + (z-Q_z)^2 -r^2 = 0$$

##### Goal: To find the intersection with a ray $R(t) = p_0 + t \cdot \vec{v}$ and the sphere
- Plug-in the paramteric ray equation $R(t)$ in to the implicit S(Q_x, Q_y, Q_z, r)$
- Produce the form: $$(R(t_x) - Q_x)^2 + (R(t_y) - Q_y)^2 + (R(t_z)-Q_z)^2 -r^2 = 0$$
    where $R(t_i)$ are the respective dimensional component of $R(t)$
- We obtain an equation in where $t$ is our variable (the rest is known/ hard numbers) as a quadratic equation. 

- We can apply the same method just using different geometric implicit representation of object (cube, Cylinder etc).
- It could be that the ray will intersect with many objects, in the physical a single light will only intersect/bounce/reflect from the first object. 
- We need to find the closest intersection, even when there are a seq. of objects that the ray can intersect. 

##### Q: How can we find closest intersection?
##### A: We will find the intersection for each object, and only retain the minimal $t_i$ parameter.  
We only need the smallest $t_i$ since the coordinate of the object grows in distance as its further away from the centre and since the (parametric) equation is additive it necessarily means that $t_i$ must also grow, so to make life easier finding the solution with smallest $t_i$ and the substituting back to the parametric equation will provide us with the coordinates of the closest intersection.   

$Complexity = O(NT)$   
$N =$ # Objects in Scene   
$T =$ # Triangles

```
Intersection FindIntersection (Ray ray, Scene scene)
{
    min_t = -nf
    min_primitive = NULL
    For each primitice in scene {
        t = Intersect(ray, primitive)
        if (t < min_t) {
            min_primitive = primitive
            min_t = t
        }
    }
    return Intersetion(min_t, min_primitive)
}
```
