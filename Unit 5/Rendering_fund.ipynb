#### The Fundementals of Rendering

$Rendering$ $:$= Takes a an input $3D$ virtual scene in some representation and converts it to a $2D$ image.

Objective: To detrmine the fundemental problems the every rendering needs to solve.

### Converting a 3D scene to a 2D Image

- Every Rendering algorithm completes the following general tasks:

1. Modelling - Creating and defining the 3D virtual object in the scene and defining the lights.
2. Viewing - Defines a model for a virtual camera, and determines its parameters e.g. where is its positions, what are its aspect ratios and opening angle etc
3. Visibility - Determines what is visible in the image, this includes what is visible from the viewpoint of the camera and removing what isn't from the scene.
4. Shading - Defines the colour of each visible 3D element in the scene and 2D pixels in the image. 


#### Ray Tracing Algorithm Intuition

In the physical world a light source will emit its wavelengths of energies and the serrounding object will reflect relative and respective wavelengths thus presenting their coloured appearence, when they hit our eyes.

Suppose we tried to simulate this virtually:

- We create a light source to produce illumination. 
- We then need to track the light source to see what they hit reflect and illuminate. 
- We track these light emissions until they reach the camera.

Problem:
We'd need a lot of light rays, since the probability of a light emmision hits and reflects objects to then reach the camera reciever/detector are very slim. Note that doing the opposite process is also not so possible for the same reason. 

Approximation solution: 
The ray tracing algorithm, will indeed start from the camera, but instead of tracing all the way back to the light source, it will stop at a certain point (halting condition) and determines the colour of the point in 3D space using some approximation.

##### Ray Casting 
This is very basic approximation of ray tracing as it will stop after one relection from the camera to determine the colour. 

- Modelling step - Assume the scene is set up 

1. Start at the camera and shoot "backwards" rays.
2. Stops tracing a ray after the first intersection with an object. 
3. Calculates the colour at the intersection point. 

##### Ray Casting Details

- It will first model the camera and position it in the 3D scene. (Viewing)
- An Image is a rectangle on a plane of the scene at some distance in front of the camera and behind the 3D objects.
- A ray is casted from the center of the camera through the centre of the pixel towards the scene.(Visibility)
- The plane (pixel plane) will retain the intersection point of the closest object in the scene and the ray. (Shading)
