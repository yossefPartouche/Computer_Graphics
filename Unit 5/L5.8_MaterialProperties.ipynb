### Material Properties

##### Objective:

- $\text{Discuss how colors are created using the phong Reflectance model}$


- $\text{The phong reflectance model in Computer Graphic is split into the 3 base Colors}$

$$\color{red}I_{refl} = I_E + K_AI_{AMB} + K_DI_L(\vec{N} \cdot \vec{L}) + K_SI_L(\vec{V} \cdot \vec{\hat{L}})^n $$
$$\color{green}I_{refl} = I_E + K_AI_{AMB} + K_DI_L(\vec{N} \cdot \vec{L}) + K_SI_L(\vec{V} \cdot \vec{\hat{L}})^n $$
$$\color{DodgerBlue}I_{refl} = I_E + K_AI_{AMB} + K_DI_L(\vec{N} \cdot \vec{L}) + K_SI_L(\vec{V} \cdot \vec{\hat{L}})^n $$

Where the parameters defining the reflectance are:

1. $\color{red}K_{AR},\color{green}K_{AG},\color{DodgerBlue}K_{AB}$ - Ambient Light coefficient
2. $\color{red}K_{DR},\color{green}K_{DG},\color{DodgerBlue}K_{DB}$ - Diffuse Coefficient
3. $\color{red}K_{SR}, \color{green}K_{SG}, \color{DodgerBlue}K_{SB}$ - Specular coefficient
4. $n$ - Shininess of surface

The Paramters of types of Lights:

1. $\color{red}I_{AR}, \color{green}I_{AG}, \color{DodgerBlue}I_{AB}$ - The Enviroment ambient Light
2. $I_L$ - The intensity of light from a light source

### Ambiant Light Color intersaction 

$\text{Suppose we have the following}$

$\color{red}(I_{AR}, \color{green}I_{AG}, \color{DodgerBlue}I_{AB}) \color{white} = (1,0,0)$

$\text{and }$

$\color{red}(K_{AR},\color{green}K_{AG},\color{DodgerBlue}K_{AB}) \color{white}= (0, 0.3, 0.1)$

$\text{Then there would be no ambiant light:}$

$(1 \cdot 0, 0 \cdot 0.3, 0 \cdot 0.1) = (0, 0, 0)$

---

### Diffuse Light Color intersaction 

$\text{Suppose we have the following}$

$\color{red}(I_{LR}, \color{green}I_{LG}, \color{DodgerBlue}I_{LB}) \color{white} = (1,1,1)$

$\text{and }$

$\color{red}(K_{DR},\color{green}K_{DG},\color{DodgerBlue}K_{DB}) \color{white}= (0.5, 0, 0.7)$

$\text{Then there would be no ambiant light:}$

$(1 \cdot 0.5, 1 \cdot 0, 1 \cdot 0.7) = \color{purple}(0.5, 0, 0.7)$

---

### Specular Light Color intersaction 

$\text{Suppose we have the following}$

$\color{red}(I_{LR}, \color{green}I_{LG}, \color{DodgerBlue}I_{LB}) \color{white} = (1,1,1)$

$\text{and }$

$\color{red}(K_{DR},\color{green}K_{DG},\color{DodgerBlue}K_{DB}) \color{white}= (0.1, 0.1, 0.1)$

$\text{Then there would be no ambiant light:}$

$(1 \cdot 0.1, 1 \cdot 0.1, 1 \cdot 0.1) = \color{white}(0.1, 0.1, 0.1)$
