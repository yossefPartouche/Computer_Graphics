### Accelerated Search Structures for General Algorithms

## 🌐 Moving Object Consideration — Expanded Notes

**Goal**: Efficiently organize space for ray tracing or collision detection, especially when objects **move**.

---

### 🔴 Uniform Grid (Baseline)

- Divide space into **equal-sized** square (2D) or cube (3D) **cells**.
- Object positions are inserted into corresponding cells.

#### 🔄 Update
- When an object moves, its **cell index must be updated**.

#### ⚖️ Trade-off: Cell Size
- **Small cells** → More accurate but expensive to store/search.
- **Large cells** → Less memory/search time but can overload cells.

**✅ Rule of Thumb**: Use the **average object size** as cell size.

- **Edge case**: Very small and very large objects break balance.
- May cause **overpopulation in some cells** or **wasted space** in others.

## 🌳 Quadtree / Octree Solution (Adaptive Grid)

Instead of fixed cells, **adaptively subdivide** space.

- **2D** → QuadTree: Each node splits into **4** regions (squares).
- **3D** → Octree: Each node splits into **8** regions (cubes).

### 🧱 Construction Rules
1. Start with a root node (bounding volume of the whole scene).
2. For each node:
   - Count how many objects are in the current cell.
   - If the number **exceeds a threshold `Limit`**, subdivide:
     - QuadTree → split into 4 sectors
     - Octree → split into 8 sectors
3. Repeat recursively until each leaf node contains **≤ Limit** objects.

## 📊 Examples: Objects vs. Tree Structure

### ✅ Example 1: Small Scene with 4 Objects

- Scene fits in 1 bounding cube (root).
- **Limit = 2 objects per node**.
- Root contains 4 objects → Subdivide.
- Each of 8 children gets **0–1 objects**.
- **No further subdivision needed**.

**🟩 Result:**
- Octree has **1 root + 8 leaves = 9 nodes total**.
- **Height of tree = 1** (root + 1 level of children).

---

### ✅ Example 2: 40 Objects Scattered Unevenly

- **Limit = 5 objects per node**.
- Some regions dense, some sparse.
- Root contains all 40 → subdivide into 8.
- Some children still have >5 objects → subdivide **those** only.

Let’s say:
- 3 child nodes had >5 objects, each subdivides → adds 3 × 8 = 24 more nodes.

**🟩 Result:**
- Total nodes: 1 (root) + 8 (first level) + 24 (second level)
- **Height varies per branch** depending on local density.
- **Adaptive tree structure**.

---

### ✅ Example 3: One Large and Many Tiny Objects

- 1 big object spans across many regions.
- 100 small objects scattered.
- **Limit = 4**.

What happens:
- Small objects are grouped neatly in different leaves.
- Large object **intersects multiple nodes**, and may force deeper subdivision even in sparse areas.

**🟩 Result:**
- Tree becomes deeper in areas touched by the large object.
- May result in **deep path for a single object**, even if most regions are light.

---

## ✅ Summary: Why Use QuadTree/Octree?

| Feature                   | Uniform Grid       | QuadTree/Octree       |
|---------------------------|--------------------|------------------------|
| Cell Size                 | Fixed              | Adaptive               |
| Memory Efficiency         | Poor for varying density | High               |
| Update Cost               | Constant           | Varies (depends on depth) |
| Handles Uneven Density    | ❌ No               | ✅ Yes                 |
| Construction Complexity   | Simple             | Recursive (per object count) |
