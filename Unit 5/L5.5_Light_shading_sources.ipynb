#### Shading and Light Sources
##### Objective: 

- $\text{How to compute the colour of pixel at an intersection between ray and object point.}$

3 Factors determinign color at a 3D point

1. $\text{Properties of the Light source}$
2. $\text{Material Properties of the object}$
3. $\text{The geometry of the object }$
    - Direction of the normal at each point on the surface of the object, relative to the light direction and the viewer

#### Light Interaction and Object Geometry
- If the ray hits the object in the same direction to the normal of the object face, the color of the face will be brightest
- If the ray hits the object in the perpendicular direction to the normal of the object face, the color of the face will be black.

To determine the color of a 3D point we need to consider the 3 factors defining the light interaction with the object.

#### Modelling Light Source
- We will determine the amount of light reaching the point from the light, which is emanating from a given light-source.
- We know that we have different types of lights, which have different properties summurised as follows: $$Ilumination = (x,y,z, \theta, \phi, \lambda)$$ 
Which describe the intesity of energy, leaving a light source, arriving at location $(x,y,z)$ from direction $(\theta, \phi)$ with a wavelength $\lambda$.

This will be too complicated and cumbersome, instead we will approximate:
- Instead of all wavelength $\rightarrow$ we will use color bases of 3 wavelengths (R, G, B).
- Instead of using complex light source $\rightarrow$ we will use simple mathematical models that approximate complex light sources (directional, point, spot).


---

$ \color{DodgerBlue}I_L \text{ = Light when it hits the point}$

$ \color{DodgerBlue} I_0 \text{= The light when emitted from source }$

#### Directional Light:

1. Ilumination of light source will be a constant: 


$$
I_L  =  I_0 \in \mathbb{R} \Rightarrow 
\begin{cases}
\color{red}I_{LR} = I_{0R} \\
\color{Lightgreen}I_{LG} = I_{0G} \\
\color{DodgerBlue}I_{LB} = I_{0B}
\end{cases}
$$

2. Direction of light 
$$\vec{v} = (v_x, v_y, v_z)$$  


---

#### Point Light: 
1. 

Light intensity $I_0$

2. 

Position $$p = (p_x, p_y, p_z)$$ and the vector for a point $P_i$ is $\vec{PP_i}= P_i - P$
    - $\text{The direction of the light will depend on the position, of the illuminating point}$

3.

$\color{DodgerBlue}\text{Attenuation Factor := The amount the light diminishes over a distance}$

$ f_{att} = K_q(distance)^2 + K_{I}(distance) + K_c$
- $ K_q = \text{Quadratic Coefficent}$
- $ K_l = \text{Linear Coefficient}$
- $ K_c = \text{Constant Constant}$

4. Illumination at object point is: $$I_L = I_0 \cdot \frac{1}{f_{att}}$$

#### Spot Light

1. 

Position $p = (p_x, p_y, p_z)$
 
producing central direction $\vec{v_d}$ and angel $\gamma$ relative to $\vec{v_d}$.

2. 
- $\text{The intensity of the light at the viewers position is dependant on the angle difference}$

$\gamma$ defines the angle between the viewers point and the direction of the light.

As the angle increases $I_L$ decreases (meaning the further the angle from the light source that harder the light will eminate) 

$I_L = 0 \rightarrow reaching \gamma = 90^\circ$.

We will acheive this property by $$cos(\gamma) = \vec{v} \cdot \vec{v_d}$$ (The vectors are normalised)
- $\text{The intensity of the light is also dependant on the distance:}$

$ f_{att} = K_q(distance)^2 + K_{I}(distance) + K_c$
- $ K_q = \text{Quadratic Coefficent}$
- $ K_l = \text{Linear Coefficient}$
- $ K_c = \text{Constant Constant}$

3. 
$$ I_L = \frac{I_{0}(\vec{v} \cdot \vec{v_d})}{f_{att}}$$
