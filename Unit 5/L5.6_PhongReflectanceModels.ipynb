#### Modelling Material properties
##### Objective: How to models materialistic properties which effects how we visualize objects in Computer Graphics.

#### Phone-Reflectance Model
Idea: The amount of light that is reflected from a surface on of an object is determined by the following formula: 
<p align="center">
<img src="images_U5/Screenshot 2025-06-21 at 17.39.57.png" height="325" width="400"/>
<p/>

$$ Light_{out} = Light_{in} - Light_{absorbed} + Light_{emitted}$$

For now we will discuss objects that aren't light sources: 

This means that it can only reflect light which were emitted from the environment. 

The Reflectance Model characterises the reflections as follows: 
1. Ambient Relection 
2. Diffuse Reflection
3. Specular Reflection 


If you add three types of relections then you can obtain the reflection of light from a non-emitting source.

---


### Ambiant 

$$
\begin{array}{|c|c|c|c|}
\hline
\textbf{Symbol} & \textbf{Definition} & \textbf{Space} & \textbf{Meaning} \\
\hline
I_A & \text{Ambient light intensity} & \mathbb{R}^+ & \text{Background light present everywhere} \\
K_A & \text{Ambient reflection coefficient} & [0,1] & \text{Material's responsiveness to ambient light} \\
I & \text{Total ambient intensity seen} & \mathbb{R}^+ & \text{Light seen due to ambient reflection} \\
\hline
\end{array}
$$

$$
I = K_A I_A
$$

$\text{📘 Explanation:} \\
\quad \bullet\ \text{Ambient light is a uniform light that is scattered in the environment and doesn’t come from a specific direction.} \\
\quad \bullet\ \text{It ensures that no object is completely dark, even if not directly lit.} \\
\quad \bullet\ \text{The ambient reflection coefficient } K_A \text{ is a material property that controls how much of this ambient light is reflected.} \\
\quad \bullet\ \text{This isn't dependant on the object Geometry}$

### Diffuse (Lambartian Characteristics)

$$
\begin{array}{|c|c|c|c|}
\hline
\textbf{Symbol} & \textbf{Definition} & \textbf{Space} & \textbf{Meaning} \\
\hline
\vec{N} & \text{Surface normal vector} & \mathbb{R}^3,\ |\vec{N}| = 1 & \text{Perpendicular to the surface} \\
\vec{L} & \text{Light direction vector} & \mathbb{R}^3,\ |\vec{L}| = 1 & \text{Points toward the light source} \\
\theta & \text{Angle between } \vec{N} \text{ and } \vec{L} & [0, \pi/2] & \text{Illumination angle} \\
K_D & \text{Diffuse reflection coefficient} & [0,1] & \text{Material’s diffuse strength} \\
I_L & \text{Light source intensity} & \mathbb{R}^{+} & \text{Brightness of the light source} \\
I_D & \text{Diffuse intensity} & \mathbb{R}^{+} & \text{Light reflected diffusely} \\
\hline
\end{array}
$$

#### 📘 Explanation

$\text{Let the light hit a rough (e.g. matte) surface. The amount of reflected light depends on the angle} \theta \text{between the}$ **surface normal** $\vec{N}$ and the **light direction** $\vec{L}$.  
$\text{Lambartian object spread the light that hit evenly in every direction so it's not dependant on matrialistic property and appears to the observer the same regardless of the angle}$
- $\text{As } \theta \text{ increases, light spreads over a wider area, reducing perceived brightness.}$ 
- $\text{We model this with } \cos(\theta)$.

$\text{Base formula}$:
$$
I = I_L \cos(\theta) = I_L (\vec{N} \cdot \vec{L})
$$

$\text{With material property}$:

$$
I_D = K_D I_L \cos(\theta) = K_D I_L (\vec{N} \cdot \vec{L})
$$

$\text{The dot product works because all vectors are normalized, and:}$
$$
\vec{N} \cdot \vec{L} = \cos(\theta)
$$

### Specular
$$
\begin{array}{|c|c|c|c|}
\hline
\textbf{Symbol} & \textbf{Definition} & \textbf{Space} & \textbf{Meaning} \\
\hline
\vec{N} & \text{Surface normal vector} & \mathbb{R}^3,\ |\vec{N}| = 1 & \text{Perpendicular to the surface} \\
\vec{L} & \text{Light direction vector} & \mathbb{R}^3,\ |\vec{L}| = 1 & \text{Points toward the light source} \\
\hat{L} & \text{Mirror reflection of } \vec{L} \text{ about } \vec{N} & \mathbb{R}^3,\ |\hat{L}| = 1 & \text{Direction of perfect reflection} \\
\vec{v} & \text{View (eye) direction vector} & \mathbb{R}^3,\ |\vec{v}| = 1 & \text{Points toward the viewer or camera} \\
\alpha & \text{Angle between } \vec{v} \text{ and } \hat{L} & [0, \pi] & \text{Controls specular falloff} \\
n & \text{Shininess exponent} & \mathbb{R}^{+} & \text{Controls highlight size and sharpness} \\
K_S & \text{Specular reflection coefficient} & [0,1] & \text{Material’s specular strength} \\
I_L & \text{Light source intensity} & \mathbb{R}^{+} & \text{Brightness of the light source} \\
I_S & \text{Specular intensity} & \mathbb{R}^{+} & \text{Amount of specular light seen by viewer} \\
\hline
\end{array}
$$
<p align="center">
<img src="images_U5/Screenshot 2025-06-21 at 17.42.21.png" height="325" width="400"/>
<p/>

$\text{When a light ray hits a surface, it reflects off in a mirror direction. If the viewer is exactly aligned with that direction, we see a bright highlight. This is called specular reflection.}$
$\text{We compute the strength of the reflection using:}$
$$
\cos(\alpha) = \vec{v} \cdot \hat{L}
$$

$\text{Raising to a power sharpens the response:}$

$$
\cos(\alpha)^n = (\vec{v} \cdot \hat{L})^n
$$

$\text{This gives us the Phong specular term:}$

$$
I_S = K_S \cdot I_L \cdot (\vec{v} \cdot \hat{L})^n
$$



#### Conclusion
$\text{When we sum all three property equations we obtain a surface colour with materialistic properties expressed.}$
