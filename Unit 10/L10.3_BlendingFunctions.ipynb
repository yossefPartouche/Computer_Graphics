### Blending Functions
##### Objectives: 
- Deal with functional Graphs
- Move from 2D to 3D using parametric curves

#### Goal: Define a continous shape using discrete elements such as points

$\text{We have a set of n points } (x_i, y_i) \text{ in 2D or } (x_i, y_i, z_i) \text{ in 3D and want to define the same of a continuous curve based on these points}$

- for a given two points $p_0, p_1 \in \R^n$ can be defined in any dimension using the following paramteric equation: $$ S(t) = (1-t)p_0 + tP_1 \ \forall \ 0 \le t \le 1$$
- We can think of any point between $p_0, p_1$ as a blend between the points $p_0, p_1$  using two blending functions being: 
1. $$f_0(t) = (1-t)$$
2. $$f_1(t) = t$$
- where $p_0, p_1$ are the blending factor and $f_0(t) + f_1(t) = 1$

### Conclusions
1. When t = 0, the segmenet passess through $p_0$, because $f_0(t) = 1$ and  $f_1(t) = 0$
2. When t = 1, the segmenet passess through $p_0$, because $f_0(t) = 0$ and  $f_1(t) = 1$
3. As blending functions are linear and add up to 1 for any t , any point on the segment is a linear combination of the two endpoints.

### Using a set of different blending functions
$$ s(t) = (1-t)^2p_0 + t^2p_1 \ \forall 0 \le t \le 1$$
1. $$f_0(t) = (1-t)^2$$
2. $$f_1(t) = t^2$$
- where $p_0, p_1$ are the blending factor and $f_0(t) + f_1(t) = 1$

### Conclusions
1. When t = 0, the segmenet passess through $p_0$, because $f_0(t) = 1$ and  $f_1(t) = 0$
2. When t = 1, the segmenet passess through $p_0$, because $f_0(t) = 0$ and  $f_1(t) = 1$
3. So the bleind curve will still pass through $p_0$ and $p_1$

#### Conclusion 
- We're not limited by the parametric equation form that can be used to create a blending function
