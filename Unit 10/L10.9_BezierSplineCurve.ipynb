#### Problems with a single Bezier curve
- $\text{We showed that a Bezier curve can be defined for any number of control points blending their geometry changing all the control points will change the appearence of the whole curve}$
- $\text{This brings back the problem of non-local control (where a single point can change the whole curve) and indeed a single Bezier curve doesn't interpolate all points}$
- $\text{Thus it's difficult to make a curve pass a single points}$

#### Advantages of Piecewise curves (splines)
- $\text{Desing complex shapes easier}$
- $\text{Can force the curve to pass through certain points}$
- $\text{Allow local control over the shape}$

#### Defining Peicewise Bezier

- $\text{We can decide not to use all the control points to define one curve, instead parts of the points to define part of the Bezier curve}$
- $\text{For example: We can decide to divide the points into group of four points, and define a different Bezier curve for each group}$
- $\text{Since each Bezier curve will pass through the beginning and end points then these curve will form a connected curve}$
$\text{where the end of one curve is the beginnning of another = } C_0 \text{ Continuity}$
- $\text{Assures that we have a piecewise continuous curve: }C_0-\text{Continuity}$

##### Q. how do we make this curve smooth at C_0-Conituity points?

- $\text{We require higher degree continouity , this mean to demand that the first derivatives are also continous (Definition of Smooth)}$
- $\text{The same requirement means that the tangent vectors at begininning of the next curve and the end of the curve before are the same}$
- $ C_1-Continuity = \text{continouity between two piecewise curve assures that they are continous and their their first derivative (tangents) will also be the continous at the connection}$
- $ C_n-Continuity = \text{Continuity between two piecewise curves assures that all their dertivatives up to N are continouos at the connecting points}$

#### Restriction of the $C_i$ Continuity on the shape of curve
- $\text{We know that the tangent vector at the beginning 3 times the vector between the second and the first points and the tangent at the end of the curve formally as we saw:}$

  $3(P_1 - P_0) \ \text{and }3(P_3 -P_2)$
- $\text{This means that the vector between the last and one before last in one curve must be equal to the vector between the second and the first point}$
$\text{in the connecting curve (assuming we want smoothness between throughout the curve)}$
- $\text{This also imposes restrictions on the editing of these points, if we move connecting points, this will affect, two curves, which connect at that point}$
- $\text{When we move a control point in a piecewise cubic Bezier spline, that is } C_1 \text{Continous, we affect a t most curves, which means we have local control}$


### Example : Quadratic Bezier Spline $C_1$ continous

- $\text{ In this case, every three points define a quadratic Bezier Curve}$

- $\text{Due to } C_1 \text{ Continuity there are constraints on the poisitons of the control points both tangents,}$
$\text{at the beginning and the end of the curve are lined to the second control point of the curve}$

- $\text{Therefore, moving the second control point in a specific Bezeir curve, affects both the curve before and the curve adter thisw Bezier}$
$\text{ specifically the vector between the first and second points, must be the same as the vector between the second and third points of the previous connecting curve and the vector}$ $\text{betweent the second and third points must be the same as the vector between the first ansd second points of the next connecting curves.}$

- $\text{This means that the last point of one curve and first point of the next curve must always be in the middle of two second control points:}$
$\text{ one from the curve before and the other from the curve after, this attribute is sometimes used to drop the connecting points altogether, we just use the off curve control points}$
