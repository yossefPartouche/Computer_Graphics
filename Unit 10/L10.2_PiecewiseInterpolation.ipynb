### Piecewise Spline Interpolation

#### Objective
- To represent piecewise representaion schemes which solves the sensitivity problem of the lagrange polynomial curves.

- $\text{The Interpolation Problem := We first check, if given a set of points can we define a curve that passes through these points.}$
- $\text{ Define n+1} \in \N s.t. \ (x_i, y_i) \text{where } x_i \ne x_j \ \forall i \ne j \  \forall x_i \text{ we have } f(x_i) = y_i $

#### Intuitive solution:
- $\text{To connect the dots}$ 
- $\text{Define a local function } f_i(x) \text{ for any sub-domain of x between }  x_i \text{ and } x_{i+1} \text{ and then the combined function becomes the continous of each peice.}$

$$
f(x) =
\begin{cases}
f_1(x) & \text{if } x_0 \le x \le x_1 \\
f_2(x) & \text{if } x_1 \le x \le x_2 \\
... \\
f_n(x) & \text{if } x_{n-1}\le x \le x_n
\end{cases}
$$

$\text{Notice that in this case every point that is changed will affect at most two neighboring segments,} \\  \text{we conclude that this will be more stable than the lagrange polnomial.}$


$\text{The only problem with this solution isn't it forms cusps and is not smooth as desired.}$

#### <u> Spline Piecewise Cubic functions with $4n$ Unknowns </u>

</div>



- $\text{Each local cubic function can be defined as}$
$$f_i(x) = a_ix^3 + b_ix^2 + c_ix + d_i \ \forall \ i = 1, ..., n$$

$$
f(x) =
\begin{cases}
f_1(x) & \text{if } x_0 \le x \le x_1 \\
f_2(x) & \text{if } x_1 \le x \le x_2 \\
... \\
f_n(x) & \text{if } x_{n-1}\le x \le x_n
\end{cases}
$$

- $\text{So we need to find our coeficients for each sub-function}$
- $\text{There will be 4n unknown variables needed to define such n functions } a_i, b_i, c_i, d_i$

<div align="center">

#### <u> In Order to create smoothness - Our $4n$  constraints: </u>

|Mathematical Expression | Description in Words| number of Constraints | 
|------------------------|---------------------|-----------------|
|$\displaystyle\lim_{x \to x_i^+} f_{i+1}(x) = y_{i+1} \quad \forall \ i = 0, \dots, n-1$| The right-side limit of segment $f_{i+1}$ at $x_i$ equals $y_{i+1}$. <br> Enforces function value continuity at each junction.| $$n$$
|$\displaystyle\lim_{x \to x_i^-} f_i(x) = y_i \quad \forall \ i = 1, \dots, n$|The left-hand limit of the function segment $f_i$ as $x \to x_i$ equals $y_i$.| $$n$$ |
|$\displaystyle\lim_{x \to x_i^-} f_i’(x) = \lim_{x \to x_i^+} f_{i+1}’(x) \quad \forall \ i = 1, \dots, n-1$|The left and right limits of the first derivative match at $x_i$ ⇒ $C^1$ continuity.| $$n-1$$ |
|$\displaystyle\lim_{x \to x_i^-} f_i’’(x) = \lim_{x \to x_i^+} f_{i+1}’’(x) \quad \forall \ i = 1, \dots, n-1$|The left and right limits of the second derivative match at $x_i$ ⇒ $C^2$ continuity.| $$n-1$$|
|$\displaystyle\lim_{x \to x_0^+} f_0’’(x) = 0$|The second derivative of the first segment at the left boundary is zero (natural spline start).|$$1$$|
|$\displaystyle\lim_{x \to x_n^-} f_n’’(x) = 0$|The second derivative of the last segment at the right boundary is zero (natural spline end).|$$1$$|

## 🧮 Example: Cubic Spline Interpolation

### Given points:

$$
(x_0, y_0) = (0, 1), \quad (x_1, y_1) = (1, 3), \quad (x_2, y_2) = (2, 2)
$$

We construct two cubic polynomials:

$$
f_1(x) = a_1x^3 + b_1x^2 + c_1x + d_1 \quad \text{on } [0, 1]
$$

$$
f_2(x) = a_2x^3 + b_2x^2 + c_2x + d_2 \quad \text{on } [1, 2]
$$

---

### 🧩 Unknowns:

$$
a_1, b_1, c_1, d_1, \quad a_2, b_2, c_2, d_2 \quad \text{(8 total)}
$$

---

### 🔒 Constraints:

#### Function value continuity (4 constraints):

$$
f_1(0) = y_0 = 1
$$

$$
f_1(1) = y_1 = 3
$$

$$
f_2(1) = y_1 = 3
$$

$$
f_2(2) = y_2 = 2
$$

#### First derivative continuity at \(x = 1\):

$$
f_1'(1) = f_2'(1)
$$

#### Second derivative continuity at \(x = 1\):

$$
f_1''(1) = f_2''(1)
$$

#### Natural spline boundary conditions:

$$
f_1''(0) = 0
$$

$$
f_2''(2) = 0
$$

---

### 📌 Derivatives:

$$
f_i'(x) = 3a_ix^2 + 2b_ix + c_i
$$

$$
f_i''(x) = 6a_ix + 2b_i
$$

---

### 🎯 Total: 8 equations for 8 unknowns → solvable system

---

[1, 2] & f_2(x) = a_2x^3 + b_2x^2 + c_2x + d_2 \\
\hline
\end{array}
