## Cubic Hermite Interpolation
##### Objective
- To use the tangent vector to govern the shape of curves
- Define Cubic Hermite Interpolation

#### Reminder
$\text{For a given two points } p_0 \text{ and } p_1 \text{we define a curve segement as:}$

$\text{We can define many blending functions of the form } s(t) = f_0(t)p_0 \ + \ f_1(t)p_1 \ \forall \ 0 \ \le t \le 1$

|Definition||
|-------|---|
|Cubic Hermite Blending Functions| Given a parameter $t \in [0,1]$, the four cubic Hermite basis functions (also called blending functions) are: $$\begin{aligned} h_1(t) &= 2t^3 - 3t^2 + 1 \\ h_2(t) &= -2t^3 + 3t^2 \\ h_3(t) &= t^3 - 2t^2 + t \\h_4(t) &= t^3 - t^2 \end{aligned}$$ |
|Cubic Hermite Curve Formula | Given two points $P_0$ and $P_1$, with tangents $\vec{v_0}$ and $\vec{v_1}$, the cubic Hermite curve is defined as: <br> $$H(t) = h_1(t) P_0 + h_2(t) P_1 + h_3(t) \vec{v_0} + h_4(t) \vec{v_1}$$|

$\text{If we use an appropriate blending function we can define an interpolating curve, so that the beginning and the end of the curve we be governed by the position of } p_0 \text{ and } p_1$

#### Q. What type of curve do we need?

$\text{Linear segments cannot have enough degrees of freedom and Quadratic curve are planar and cannot twist in space}$

$\text{We interpolate on two points } p_0 \text{ and } p_1 \text{ using the Lagrange polynomial, we know for certain that the directions of the tangent to the curve at } p_0 \text{and } p_1 \text{ are parallel to some intersecting plane}$

$\text{We search a cubic polynomal curve: }$ 

We define the curve as:
$$\mathbf{s}(t) = \big(x(t), y(t), z(t)\big)$$
where each of x(t), y(t), z(t) is a cubic polynomial. Assuming:
$$\mathbf{s}(t) = \mathbf{a}t^3 + \mathbf{b}t^2 + \mathbf{c}t + \mathbf{d} \quad \text{with } \mathbf{a}, \mathbf{b}, \mathbf{c}, \mathbf{d} \in \mathbb{R}^3$$


$\text{These are our 12 unknowns or degrees of freedom} $
$$\mathbf{a} = (a_x, a_y, a_z)$$
$$\mathbf{b} = (b_x, b_y, b_z)$$
$$\mathbf{c} = (c_x, c_y, c_z)$$
$$\mathbf{d} = (d_x, d_y, d_z)$$

$\text{Then:}$

$$x(t) = a_x t^3 + b_x t^2 + c_x t + d_x$$

$$y(t) = a_y t^3 + b_y t^2 + c_y t + d_y$$

$$z(t) = a_z t^3 + b_z t^2 + c_z t + d_z$$

#### What we know

$\text{For } t = 0:$

$$
\mathbf{s}(t) = \mathbf{a}t^3 + \mathbf{b}t^2 + \mathbf{c}t + \mathbf{d} = \mathbf{p}_0 \quad \text{with } \mathbf{a}, \mathbf{b}, \mathbf{c}, \mathbf{d} \in \mathbb{R}^3
$$

$$
x(t) = a_x t^3 + b_x t^2 + c_x t + d_x = p_{0x} \quad \Rightarrow \quad d_x = p_{0x}
$$

$$
y(t) = a_y t^3 + b_y t^2 + c_y t + d_y = p_{0y} \quad \Rightarrow \quad d_y = p_{0y}
$$

$$
z(t) = a_z t^3 + b_z t^2 + c_z t + d_z = p_{0z} \quad \Rightarrow \quad d_z = p_{0z}
$$

$\text{For } t = 1:$

$$
\mathbf{s}(t) = \mathbf{a}t^3 + \mathbf{b}t^2 + \mathbf{c}t + \mathbf{d} = \mathbf{p}_1 \quad \text{with } \mathbf{a}, \mathbf{b}, \mathbf{c}, \mathbf{d} \in \mathbb{R}^3
$$

$$
x(t) = a_x t^3 + b_x t^2 + c_x t + d_x = p_{0x} \quad \Rightarrow \quad a_x + b_x + c_x + d_x = p_{1x}
$$

$$
y(t) = a_y t^3 + b_y t^2 + c_y t + d_y = p_{0y} \quad \Rightarrow \quad a_y + b_y + c_y + d_y = p_{1y}
$$

$$
z(t) = a_z t^3 + b_z t^2 + c_z t + d_z = p_{0z} \quad \Rightarrow \quad a_z + b_z + c_z + d_z = p_{1z}
$$

#### Conclusion replacing t = 0 and t = 1 provides us with 6 constraints (unknowns)


### Cubic Polynomial Curves

$\text{ The tangent of } s(t) at^3 + bt^2 + ct + d \text{ at t is the vector: }$

$$\vec{V_{tangent}}(t) = \frac{d}{dt}s(t) = \left(3a_xt^2 + 2b_xt + c_x, 3a_yt^2 + 2b_yt + c_y, 3a_zt^2 + 2b_zt + c_z\right)$$

$\text{Similarly to the above For } t = 0$
$$\vec{V_{tangent}}(t) \frac{d}{dt}s(t) = \vec{v_0}$$

$$3a_xt^2 + 2b_xt + c_x = v_{0x} \Rightarrow  c_x = v_{0x}$$

$$ 3a_yt^2 + 2b_yt + c_y = v_{0y} \Rightarrow c_y = v_{0y}$$

$$ 3a_zt^2 + 2b_zt + c_z = v_{0z} \Rightarrow c_z = v_{0z}$$


$\text{Similarly to the above For } t = 1$
$$\vec{V_{tangent}}(t) \frac{d}{dt}s(t) = \vec{v_1}$$

$$3a_xt^2 + 2b_xt + c_x = v_{0x} \Rightarrow  3a_x + 2b_x + c_x = v_{1x}$$

$$ 3a_yt^2 + 2b_yt + c_y = v_{0y} \Rightarrow  3a_y + 2b_y + c_y = v_{1y}$$

$$ 3a_zt^2 + 2b_zt + c_z = v_{0z} \Rightarrow  3a_z + 2b_z + c_z = v_{1z}$$



#### Conclusion: replacing t = 0 and t = 1 provides us with more constraints 6 constraints (unknowns)

### Big conclusion: 12 equations with 12 unknowns $\implies$ solvable
- $\text{by construction, we obtain a cubic curve The Cubic hermit Curve that interpolates } $P_0$ and $P_1$

### Cubic Hermite Blending Functions and Properties

1. $\text{We find that the sum of the positions of blending functions} H_0 \text{ and } H_1 \text{is always 1}$
2. $\text{When t = 0 } H_0 \text{ the blending function  of } p_0 \text{ is 1 while others are zero}$
3. $\text{when t = 1 } H_1 \text{ the blending function  of } p_1 \text{ is 1 while others are zero thus insuring the curve interpolates between both points}$

$\text{Suppose we want to interpolate on n points using the cubic hermite interpolation between each pair of points, we must make sure to use the same vector in every middle point for each of the curves}$
