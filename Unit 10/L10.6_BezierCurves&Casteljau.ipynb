### 2D Example of Bezier Curve

- $\text{If we start with two points the Bezier curve between the two points is simple the line segment between them}$
- $\text{Suppose we now add a thirs point, the Bezier Curve will look like a quadratic polynomial, whilst it DOES NOT pass through the midpoint}$
- $\text{Adding more and more points produces a polynomial curve, where the curve only passes the first and last point,} \\ \text{however, the shape of the curve is indeed governed by the set of points}$

- $\text{The resulting curve is known as an:} \\ \text{APPROXIMATING CURVE := A curve that doesn't pass thorugh the points (doesn't interpolate them) but approximates their shape}$
- $\text{The input points are called the control points, as they control the shape of the curve}$
- $\text{A Bezier curve in 2D is formalised as: s(t) = } (x(t), y(t)) \  \forall \ 0 \le t \le 1$
- $\text{To create this paramteric equation we must define the control points, and this is done using de-Casteljau Algorithm}$

### de-Calsteljau Algorithm 

$\text{Suppose that t = 1/3 (although this holds for any t)}$
1. $\text{We begin by connecting each pair of connsecutive points to form line segments := Control Polyline/gon}$
2. $\text{For each line segment in the Control Polyline, we find a point that's 1/3 from the beginning of the segment and 2/3 away from the end of the segment}$
3. $\text{We Apply step 1 over the set of the n-new points (we've reduced the number of points by 1)}$
4. $\text{This process continues until we reach a single point}$

#### Basis of Algorithm

$\text{The basis of the algorithm is re-applying a simple linear interpolation over again: } P_{new} = (1-t)P_{start} + tP_{end}$

#### Generalisation defining Bezier point for any t

<img src="image_U10/Screenshot 2025-06-15 at 20.17.35.png" height="325" width="325"/>


$\text{Assume we have n = 4, this means we have 5 control points}$
$\text{We will denote a new control point as follows } b_i^k \text{ where k := Defines the level of recursion and i := Defines the order of points in the polygon}$

<img src="image_U10/Screenshot 2025-06-15 at 20.20.27.png" height="325" width="325"/>


$\text{In the first level of recursion, we define new points in each segment}$
$$b_0^1 = tb_1^0 + (1-t)b_0^0$$
$$b_1^1 = tb_2^0 + (1-t)b_1^0$$
$$b_2^1 = tb_3^0 + (1-t)b_2^0$$
$$b_4^1 = tb_4^0 + (1-t)b_3^0$$

<img src="image_U10/Screenshot 2025-06-15 at 20.22.54.png" height="325" width="325"/>

$\text{In the second level of recusion, we define new points in each segment again}$
$$b_0^2 = tb_1^1 + (1-t)b_0^1$$
$$b_1^2 = tb_1^2 + (1-t)b_1^1$$
$$b_2^2 = tb_1^3 + (1-t)b_2^1$$

<img src="image_U10/Screenshot 2025-06-15 at 20.27.55.png" height="325" width="325"/>

$\text{In the third level of recusion, we define new points in each segment again}$
$$b_0^3 = tb_1^2 + (1-t)b_0^2$$
$$b_1^3 = tb_2^2 + (1-t)b_1^2$$

<img src="image_U10/Screenshot 2025-06-15 at 20.29.43.png" height="325" width="325"/>


$\text{In the final level of recusion, we define new points in each segment again}$
$$b_0^4 = tb_1^3 + (1-t)b_0^3 = s(t) $$

<img src="image_U10/Screenshot 2025-06-15 at 20.30.50.png" height="325" width="325"/>

$\text{A key advantage of this algorithm is that we only needed 2 points in order to produce the approximation and this means we could apply it in any dimension}$

#### Drawing the Bezier Curve

$\text{We use enumeration to ffind points on the curve and connect them to draw the curve, since the Bezier curve is a parametric curve, we apply the same process}$

##### Example t = (0, 0.25, 0.5, 0.75, 1)
- This will provide a very rough approximation, to resolve this, we just create larger discrete values of t and this will provide a better curve approximation 

$\text{Suppose we wish to compute a bezier curve for n points, the number of linear interpolations we need to compute } b_i^2 \text{ is: }  \frac{i(i+1)}{2} \ \forall \ 0 \le i \le n-1$
