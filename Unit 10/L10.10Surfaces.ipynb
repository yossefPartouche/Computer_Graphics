## Surfaces

##### Objective:
- $\text{ How to move from curve to surfaces (breifly)}$

#### Reminder: 
- $\text{ The main motivation of this unit was to show how boundaries of objects are represented more accurately and smoothly}$
- $\text{ We focused mostly on curve representations and showed several possibilities how a curve can be defined in 2D or 3D using just points or vectors}$
- $\text{ So far we've seent the following:}$
$$\text{In 2D: } s(t) = (x(t), y(t)) \forall t \in [0,1]$$

$$ \text{In 3D } s(t) = (x(t), y(y), z(t)) \forall t \in [0,1]$$

$\text{The general idea was to use points or vectors to define the shape of the curve and some bleinding functions,}$
$\text{which are scalar functions of the parameter t, with special characteristics (constraints)}^*$

$^* \text{Review Lesson 10.3 for more on the characteristics}$

#### How does this change over surfaces?

$$\text{In 2D: } s(u,v) = (x(u,v), y(u, v)) \forall u,\ v \in [0,1]$$

$$\text{In 3D: } s(u,v) = (x(u,v), y(u, v), z(u,v)) \forall u,\ v \in [0,1]$$

- $\text{ The surfaces are still blending functions, blending the geometry of the control points or/also the control vectors}$

- $\text{ This is an extension of the Hermite Curve mapping the 2D parametric domain to 3D geometric space by constructing a cubic polynomial patch defined by the following geometric constraints:}$
1. $\text{Defined using 4 corner points}$
2. $\text{it will have the following direction vectors}$
    - $\text{ 4 tangent vectors in the u direction at each point which should be equal to the partial derivative of the surface w.r.t. u at each point}$
    - $\text{4 tangent vectors in the v direction at each point which should be equal to the partial derivative of the surface w.r.t. v at each point}$
    - $\text{ 4 tangent twist vectors that should be equal to the partial derivatices w.r.t. to both u and v at each point}$
    - $\text{Using the above you can define a single cubic polynomial patch that fulfills these constraints}$

- $\text{The Bezier surface is an extension of the Bezier curve which is define by a grid of n+1 by m+1 control points for the two major directions u and v}$
- $\text{ The surface itself is still defined using the Bernstein Bezier polynomial blending functions that blend the control points in both u and v directions}$
- $\text{ The nice properties of the Bezier curve carry over to a Bezier surface, where a Bezier surface of n+1 x m+1 control points}$
$\text{ in an n-degree polynomial surface in one direction and an m-degree polynomial surface in the other direction}$
- $\text{ A Bezier surface blends the positions of its control points using Bernstein-Bezier polynomials }$
- $\text{ A Bezier surface is an approximating surface interpolatiung only its four corner points}$
- $\text{ The surface lies within the convex hull of its control points}$
- $\text{ If two Bezier surfaces interpolating 2 set of n points; S and T where we know that there's no intersection between the convex hulls of S and T. We can conclude that both surfaces don't intersect}$

-$\text{These properties allow simple control over the shape over the shape of the surface}$
