## Bezier as polynomial Curve

#### Objective
- Understand that combining all levels of recursion on the de-Casteljau Algorithm defines an algebraic formula for the Bezier Curve as a polynomial curve.

- $\text{During the Casteljau algorithm we defined } b_i^k$
- $\text{Read as "point i in recursion level k" where the points in level k were a linear combination of two point in level k-1}$
#### Example 
- $ s(t) = b_0^4 = tb_1^3 + (1-t)b_0^3$

- $\text{Where } b_1^3 \text{ and } b_0^3 \text{ are also linear combination of}$

- $s(t) = b_1^3 = tb_2^2 + (1-t)b_1^2$
- $s(t) = b_0^3 = tb_1^2 + (1-t)b_0^2$

- $\text{Which produces: }

- $s(t) = b_0^4 = t(tb_2^2 + (1-t)b_1^2) + (1-t)(tb_1^2 + (1-t)b_0^2)$

- $\text{Where }  b_2^2 \text{ and } b_1^2 \text{ and } b_0^2 \text{ are also linear combination of}$

- $s(t) = b_2^2 = tb_1^3+ (1-t)b_1^2$
- $s(t) = b_1^2 = tb_1^2+ (1-t)b_1^1$
- $s(t) = b_0^2 = tb_1^1+ (1-t)b_1^0$

- $\text{Which produces: }$

- $s(t) = b_0^4 = t(t(tb_1^3+ (1-t)(tb_1^2+ (1-t)b_1^1)) + (1-t)(tb_1^2+ (1-t)b_1^1)) + (1-t)(t(tb_1^2+ (1-t)b_1^1)+ (1-t)(tb_1^1+ (1-t)b_1^0))$

- $\text{In the end we simplify and produce the following polynomial formula}$

- $ s(t) = t^4P_4 + 4t^3(1-t)P_3 + 6t^2(1-t)^2P_2 + 4t(1-t)^3P_1 + (1 -t)^4P_0$

- $\text{The resulting function is also a blending function, where the geometry are the points} P_0 - p_4 $

### Generalising the Bezier Curve 
#### 4 Control Points
- $s(t) = t^3P_3 + 3t^2(1-t)P_2 + 3t(1-t)P_1 + (1-t)^3P_0$
#### 3 Control Points
- $s(t) = t^2P_2 + 2t(1-t)P_1 + (1-t)^2P_0$
- $\text{A bezier curve of n+1 control-points is an n-degree polynomial curve}$
- $\text{A Bezier curve blends the positions of its control points using Bernstein-Bezier polynomials:}$

$$B_i^n(t) = (nCi)t^i(1-t)^{n-i}$$

- $\text{They are the terms of the euqation:}$
$$ (t+ (1-t))^n$$

#### Bernstein Bezier Polynomials for n=2

$$B_0^2(t) = (2C0)t^0(1-t)^{2-0} = (1-t)^2$$
$$B_1^2(t) = (2C1)t^1(1-t)^{2-1} = 2t(1-t)$$
$$B_2^2(t) = (2C2)t^2(1-t)^{2-2} = t^2$$

#### Bernstein Bezier Polynomials for n=3

$$B_0^3(t) = (3C0)t^0(1-t)^{3-0} = (1-t)^3$$
$$B_1^3(t) = (3C1)t^1(1-t)^{3-1} = 3t(1-t)^2$$
$$B_2^3(t) = (3C2)t^2(1-t)^{3-2} = 3t^2(1-t)$$
$$B_3^3(t) = (3C3)t^3(1-t)^{3-3} = t^3$$

### Bernstein Bezier polynomials properties
- $\text{The n+1 polynomails of degree n sum up to 1 for any t value, since they are the terms of the equation:} (t+(1-t))^n$
- $\text{For t=0 only one of them i.e. } (1-t)^n \text{ is one and the others are zero}$
- $\text{For t=1 only one of them i.e. } (t)^n \text{ is one and the others are zero}$

### The Algebraic form of the Bezier Curve

$$s(t) = \sum_{i=0}^n B^n_i(t)P_i$$

### The Algebraic form of the Bezier Curve over Higher dimensions

#### 2D
$$s(t) = \sum_{i=0}^n B^n_i(t)P_i = \begin{cases} 
\sum_{i=0}^n B^n_i(t)P_{ix} \\ \sum_{i=0}^n B^n_i(t)P_{iy} \end{cases}$$

#### 3D

$$s(t) = \sum_{i=0}^n B^n_i(t)P_i = \begin{cases} 
\sum_{i=0}^n B^n_i(t)P_{ix} \\ \sum_{i=0}^n B^n_i(t)P_{iy} \\ \sum_{i=0}^n B^n_i(t)P_{iz} \end{cases}$$

##### Cubic Bezier Curve 

$s(t) = t^3P_3 + 3t^2(1-t)P_2 + 3t(1-t)P_1 + (1-t)^3P_0$
