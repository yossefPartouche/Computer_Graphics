#### Finding the tangent Vector in 3D (Parametric Curve)

$\text{Given a parametric curve s(t) = (x(t), y(t), z(t)) being a scalar function of the parameter t, defining one coordinate of the point of the curve.}$

$$\text{Tangent Vector :=}\vec{V_{tangent}}(t) = \frac{d}{dt}s(t) = \left(\frac{d}{dt}x(t),\frac{d}{dt}y(t),\frac{d}{dt}z(t)\right)$$

$\text{The above can be expanded and reduced to any dimesion}$

### Example
We blend two points in 3D: $p_0 = (1,1,1)$ and $p_1 = (1,2,2)$ using the following blending functions $\forall 0 \le t \le 1$ $$f_0(t) = (1-t)^2$$ and $$f_1(t) = t^2$$ 
The Tangent vector to the curve for $\forall 0 \le t \le 1$
1. $\text{Define the blended curve: }$

$$ s(t) = f_0(t)p_0 + f_1(t)p_1 = s(t)$$
$$ = \left[
f_0(t) \cdot x_0 + f_1(t) \cdot x_1,\,
f_0(t) \cdot y_0 + f_1(t) \cdot y_1,\,
f_0(t) \cdot z_0 + f_1(t) \cdot z_1
\right]$$

$$ s(t) = \left[
(1 - t)^2 \cdot 1 + t^2 \cdot 1,\,
(1 - t)^2 \cdot 1 + t^2 \cdot 2,\,
(1 - t)^2 \cdot 1 + t^2 \cdot 2
\right] $$

2. $\text{Simplify Each component:}$
$$x(t) = (1 - t)^2 + t^2 = 1 - 2t + t^2 + t^2 = 1 - 2t + 2t^2$$

$$y(t) = (1 - t)^2 + 2t^2 = 1 - 2t + t^2 + 2t^2 = 1 - 2t + 3t^2$$

$$z(t) = (1 - t)^2 + 2t^2 = 1 - 2t + t^2 + 2t^2 = 1 - 2t + 3t^2$$


3. $\text{Compute the Tangent Vector}$
	$$\frac{dx}{dt} = -2 + 4t$$
	$$\frac{dy}{dt} = \frac{dz}{dt} = -2 + 6t$$
    $$\frac{dz}{dt} = \frac{dz}{dt} = -2 + 6t$$

4. $\text{Final Tangent Vector Expression:}$
$$ \vec{v}_{\text{tangent}}(t) = \left(-2 + 4t,\; -2 + 6t,\; -2 + 6t\right)$$

$\text{We already saw this in action in 2D to find the tangent vector curve and indeed we could use it to find the normal vector, the issue arises in 3D as there are infinite normal vectors to choose from given a tangent vector of a point on the curve}$
$\text{These normal vectors are all the vectors that are parallel to a plane which the normal is the the tangent vector}$

#### Fernet Frame Of 3D parametric Curve

1. $\text{The normal vector will be defined as the derivative of the trangent vector with respect to the arclength parametrication}$
2. $\text{We can then apply the cross product between this normal and the tangent vector we will produce another vector known as the bi-Normal vector}$
3. $\text{Together they form a normal orthonormal coordinate system called the "Fernet Frame" which uniquely deines the shape of a continous and smooth 3D curve at any point on the curve}$

#### Tangents and normals of surfaces

1. $\text{Given a paramteric surface s(u,v) = (x(u,v), y(u,v), z(u,v)) there are two tangent vector at a point s(u,v)}$
Let  
$$
\mathbf{s}(u, v) = \big(x(u, v),\; y(u, v),\; z(u, v)\big)
$$

Then the two **tangent vectors** at a point $(u, v)$ on the surface are:

#### Tangent Vector in the $u$-direction:
$$
\mathbf{v}_u = \frac{\partial \mathbf{s}}{\partial u} = \left( \frac{\partial x}{\partial u},\; \frac{\partial y}{\partial u},\; \frac{\partial z}{\partial u} \right)
$$

#### Tangent Vector in the $v$-direction:
$$
\mathbf{v}_v = \frac{\partial \mathbf{s}}{\partial v} = \left( \frac{\partial x}{\partial v},\; \frac{\partial y}{\partial v},\; \frac{\partial z}{\partial v} \right)
$$

#### These Tangents Give You:
- A **tangent plane** to the surface at point $\mathbf{s}(u, v)$  
- A **surface normal** via:
$$
\mathbf{n}(u,v) = \mathbf{v}_u \times \mathbf{v}_v
$$

#### Example

Given a surface in parametric representation:

$$
\mathbf{p}(u,v) = \big(3\cos(u),\; 3\sin(u),\; v\big)
$$

To compute the **tangent vector in the $u$-direction**, we differentiate $\mathbf{p}(u, v)$ with respect to $u$:


$$
\frac{\partial \mathbf{p}}{\partial u} = \left( \frac{d}{du}(3\cos(u)),\; \frac{d}{du}(3\sin(u)),\; \frac{d}{du}(v) \right) = \left( -3\sin(u),\; 3\cos(u),\; 0 \right)
$$
