### Curves and Interpolation


### Pros vs Cons on Triangle Meshes for Object representaion
| Aspect         | Advantages                                                                 | Disadvantages                                                                 |
|----------------|----------------------------------------------------------------------------|--------------------------------------------------------------------------------|
| **Simplicity** | Triangles are the simplest polygon, always planar and convex.              | Complex shapes require a large number of triangles, increasing data size.     |
| **Universality** | Widely supported by all rendering hardware and graphics pipelines.         | Not always intuitive or efficient for modeling smooth or curved surfaces.     |
| **Performance** | GPU-accelerated rendering and rasterization is highly optimized for them. | Can suffer from aliasing or visual artifacts without sufficient resolution.   |
| **Flexibility** | Any surface, no matter how complex, can be approximated using triangles.  | Precision and quality depend on mesh resolution; low-res meshes look blocky.  |
| **Interoperability** | Standard format for exporting/importing 3D models across tools and engines. | May not encode semantic or structural information (e.g., surface type).       |
| **Mathematics** | Easy to compute normals, barycentric coords, intersections, etc.         | Meshes can have holes, non-manifold edges, or other geometric issues.         |

##### Q. What if we actually need represent a smooth surface and not display it, increasing mesh density will only provide a closer approximation?
##### A. Our solution lies in being able to represent continuity and smoothness of any surfaces using discrete entities.

#### Representation of a Curve using Finite points

- $\text{The Interpolation Problem := We first check, if given a set of points can we define a curve that passes through these points.}$
- $\text{ Define n} \in \N s.t. \ (x_i, y_i) \text{where } x_i \ne n_j \ \forall i \ne j \  \forall x_i \text{ we have } f(x_i) = y_i $

$\text{for two points we can just define a linear function}$ 
$$f(x) =\frac{y_2 -y_1}{x_2-x_1}x + \frac{(x_2-x_1)y_1- (y_2 -y_1)x_1}{(x_2-x_1)}$$


$\text{for n points we construct an n-1 degree polynomial that passes thorugh any n points with different x values}$
1. $\text{We define n polynomial terms in the following way:} \\ \text{(j won't subtract from itself at any point to avoid 0 and division by s)}$ 
    $$I_j(x) = \frac{\prod_{i=1}^{n} x-x_i}{\prod_{i=1}^{n}x_j -x_i}$$
2. $\text{We build the Lagrange Polynomial by adding the } I_j \text{ polynomial terms multiplied by } y_j \text{ factors}$ 
    $$P(x) = \sum_{j=1}^{n}y_jI_j(x)$$
3. $\text{We have produced a Continious and smooth function}$

### Sensitivity on Lagrangian Polynomial

- $\text{If we change the points the curve will change, which is good as it's adaptive however the sensitivty means it can change drastically for small changes.}$
- $\text{This is because the function is of degree n-1, and for high degree functions are likely to have many extreme points,} \\ \text{thus jaggy and unstable, also we lack local control (one point can affect the whole curve).}$

### Example
Given are three points (0,0), (1,3), (2,8) which interpolate via the Lagrange polynomial. Recall we calculate a linear combination of three polynomials of degree 2 :  L1(x), L2(x), L3(x). 

$(x_0, y_0) = (0, 0)$

$(x_1, y_1) = (1, 3)$

$(x_2, y_2) = (2, 8)$

### Step 1: Compute the basis polynomials
$L_0(x) = \frac{(x - 1)(x - 2)}{(0 - 1)(0 - 2)} = \frac{(x - 1)(x - 2)}{2}$

$L_1(x) = \frac{(x - 0)(x - 2)}{(1 - 0)(1 - 2)} = \frac{x(x - 2)}{-1} = -x(x - 2)$

$L_2(x) = \frac{(x - 0)(x - 1)}{(2 - 0)(2 - 1)} = \frac{x(x - 1)}{2}$


### Step 2: Combine them into the interpolation polynomial

$P(x) = 0 \cdot L_0(x) + 3 \cdot L_1(x) + 8 \cdot L_2(x)$

### Substitute the expressions:

$P(x) = 3 \cdot \left(-x(x - 2)\right) + 8 \cdot \left(\frac{x(x - 1)}{2}\right)$

#### Simplify:

$P(x) = -3x(x - 2) + 4x(x - 1)$

$\text{Now expand:}$

$-3x(x - 2) = -3x^2 + 6x$

$4x(x - 1) = 4x^2 - 4x$

#### Add together:

$ P(x) = (-3x^2 + 6x) + (4x^2 - 4x) = x^2 + 2x$
