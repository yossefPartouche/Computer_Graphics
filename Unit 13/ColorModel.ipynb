### General Method for Solving RGB Light on CMY Ink Color Questions

Core Concepts
- RGB: additive color model (light).
- CMY: subtractive color model (ink/paint).
- Perceived color = Light reflected by the inked surface.

---

#### Step-by-Step Method

|Step| Method | Example |
|-----|-------|---------|
| $$1$$ | $\text{Identify the light source color (RGB)}$ | $\color{green}\text{Green lamp → RGB = (0, 1, 0)}$ | 
| $$2$$ | $\text{Convert the ink color from CMY to RGB}  \\ \text{RGB} = 1 - \text{CMY}$| $\color{cyan}\text{CYAN: (CMY = 1, 0, 0) → RGB = (0, 1, 1) }$|
| $$3$$ | $\text{Multiply component-wise the light RGB by the ink-reflected RGB}$ | Light = (0, 1, 0) <br> Ink = (0, 1, 1) <br> Reflected = (0×0, 1×1, 0×1) = (0, 1, 0) | 
| $$4$$ | $\text{Interpret the resulting RGB as the perceived color}$ |$\color{Green}\text{Green}$|

---

#### Edge Cases to Consider
|Case | Point | Example |
|-----|-------|---------|
| $$1$$ |$\text{Light contains no components reflected by the ink}$ | Light: (1, 0, 0) (red) <br> Ink: CYAN (0, 1, 1) <br> Result: (1×0, 0×1, 0×1) = (0, 0, 0) → black (no reflection)|
| $$2$$ | $\text{Ink absorbs all light (black ink)}$ | Light: (1, 1, 1) (white) <br> Ink: Black (CMY = 1, 1, 1) → RGB = (0, 0, 0) <br> Result: (1×0, 1×0, 1×0) = (0, 0, 0)|
| $$3$$ | $\text{White light on partial reflecting ink}$ | Light: (1, 1, 1) (white) <br> Ink: Magenta (CMY = 0, 1, 0) → RGB = (1, 0, 1) <br> Result: (1×1, 1×0, 1×1) = (1, 0, 1) → magenta |
| $$4$$ | $\text{No ink (white paper)}$ | CMY = (0, 0, 0) → RGB = (1, 1, 1) <br> Result = light color unchanged |
