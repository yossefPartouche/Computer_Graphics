## Legacy OpenGL (Immediate Mode) Summary

Legacy OpenGL (immediate mode) is a deprecated but educational graphics API. It allows you to render objects by issuing commands between `glBegin` and `glEnd`.

---

### 1. Rendering Primitives

Used to draw geometry directly.

- `glBegin(GL_TRIANGLES)`, `GL_QUADS`, etc. — Start primitive drawing.
- `glVertex*` — Specify vertex coordinates.
- `glColor*` — Set vertex color.
- `glNormal*` — Specify normal vector (for lighting).
- `glTexCoord*` — Set texture coordinate.
- `glEnd()` — Finish primitive.

---

### 2. Transformations

Used to change object placement.

- `glMatrixMode(GL_MODELVIEW)`
- `glLoadIdentity()`
- `glTranslatef(x, y, z)`
- `glRotatef(angle, x, y, z)`
- `glScalef(x, y, z)`
- `glPushMatrix()` / `glPopMatrix()`

---

### 3. Camera Setup

Controls how the scene is projected and viewed.

- `gluPerspective(...)` — Perspective projection.
- `gluOrtho2D(...)` — Orthographic projection.
- `gluLookAt(...)` — Position the camera.

---

### 4. Coloring and Materials

How surfaces look.

- `glColor3f(r, g, b)`
- `glMaterialfv(...)`
- `glEnable(GL_COLOR_MATERIAL)`

---

### 5. Lighting

Simulates realistic shading.

- `glEnable(GL_LIGHTING)`
- `glEnable(GL_LIGHT0)`
- `glLightfv(...)`
- `glNormal3f(...)`

---

### 6. Texture Mapping

Applies images to geometry.

- `glEnable(GL_TEXTURE_2D)`
- `glBindTexture(...)`
- `glTexCoord2f(u, v)`
- `glTexImage2D(...)`

---

### 7. State Management

OpenGL is state-based.

- `glEnable(...)` / `glDisable(...)`
- `glDepthFunc(...)`
- `glBlendFunc(...)`
- `glShadeModel(GL_SMOOTH)`

---

### 8. Display Management

Controls screen output.

- `glClear(GL_COLOR_BUFFER_BIT | GL_DEPTH_BUFFER_BIT)`
- `glFlush()`
- `glutSwapBuffers()`

---

### 9. GLUT Toolkit

Handles window creation and events.

- `glutInit(...)`
- `glutCreateWindow("Title")`
- `glutDisplayFunc(...)`
- `glutKeyboardFunc(...)`
- `glutMainLoop()`
