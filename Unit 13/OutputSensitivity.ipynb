### Output Sensitive Algorithm


$\text{Output Sensitive Algorithm := This is an algorithm which is more dependant on the output rather than the input}$
- $\text{In our case most of the algorithm we've learnt discuss rendering an image} \\ \text{thus the output is the pixels hxw, and the input will vary depending on the algorithm used.}$

$\text{The general goal in optimizing an algorithm to be more output sensitive is to reduce the complexity of the input factors.}$

### Ray Tracing
$\text{Ray Tracing (basic Form) isn't output sensitive, since it's complexity is largly dependent on:}$
- Number of Light sources
- Number of Triangles (Representing the 3D object)
- Depth of recursion 
- Dimension of Image (HxW)

$\text{Most of the components of the algorithm is Input dependent rather than output dependant}$

### Optimize for greater Sensitivity

$\text{For every ray that is shot from the camera we must check for an intersection, this demands iterating thorugh the set of triangles.}$

$\text{We can optimize this process by creating bounding box in the scene,} \\ \text{which eliminates areas of the scene completely based on the intersections of these "boxes"} \\ \text{reduces the complexity of the algorithm from } O(T) \text{ to } O(log(T))$

$\text{This reduces the dependency of the input, thus making it more output sensitive}$


### Rendering Pipeline

$\text{The rendering pipeline, sends in the input a set of triangles and the output is an image, thus the algorithm is indeed input dependant}$

### Optimize for greater Sensitivity

$\text{We can define a hierarchial data structure, which allows to choose different versions (Level of Detail) of an object.} \\ \text{This is done by adjusting the number of triangles (Triangle density) that is represented by the object (a.k.a Multi-Resolution)} \\ \text{Indeed the simpler the object the lower the resolution of the object thus making the algorithm more output sensitive}$
