### Projection Comparison

### $\text{Projection Types Summary}$

---

#### $\text{1. Orthographic Projection}$

- $\text{Projection lines are parallel and perpendicular to the viewing plane.}$
- $\text{No foreshortening — object size remains constant regardless of depth.}$
- $\text{Used for CAD, engineering, architectural blueprints.}$

<img src="image_U8/Screenshot 2025-05-29 at 15.21.47.png" height="325" width="400"/>
<img src="image_U8/Screenshot 2025-05-29 at 15.21.29.png" height="325" width="400"/>

---




#### $\text{2. Oblique Projection}$

- $\text{Projection lines remain parallel but are not perpendicular to the viewing plane.}$
- $\text{Causes foreshortening along one axis (usually z), creating depth illusion.}$
- $\text{Common types: Cabinet (scaled), Cavalier (unscaled).}$

<img src="image_U8/Screenshot 2025-05-29 at 15.22.23.png" height="325" width="400"/>
<img src="image_U8/Screenshot 2025-05-29 at 15.22.07.png" height="325" width="400"/>

---




#### $\text{3. Perspective Projection}$

- $\text{Projection lines converge to a center of projection (COP), creating realistic depth.}$
- $\text{Objects farther from viewer appear smaller (foreshortening).}$
- $\text{Used in photography, games, film.}$

### $\text{Moving the Viewing Plane vs. Moving the COP (Perspective)}$

- $\textbf{Changing the viewing plane}$:
  - $\text{Alters which part of the scene is visible without changing perspective.}$
  - $\text{Distance between projected points may stay the same.}$

- $\textbf{Changing the center of projection (COP)}$:
  - $\text{Affects depth and vanishing point behavior.}$
  - $\text{Can increase or decrease perspective distortion.}$

### $\text{Viewing Volume and Clipping}$

- $\textbf{Viewing volume}$ is the region of space that gets projected into the final image.

<img src="image_U8/Screenshot 2025-05-29 at 15.24.51.png" height="325" width="400"/>

#### $\text{Orthographic}$:
- $\text{Viewing volume = rectangular box (viewing window + front/back clipping planes).}$

<img src="image_U8/Screenshot 2025-05-29 at 15.28.44.png" height="325" width="400"/>

#### $\text{Oblique}$:
- $\text{Similar to orthographic, but skewed in direction of projection.}$

<img src="image_U8/Screenshot 2025-05-29 at 15.29.02.png" height="325" width="400"/>

#### $\text{Perspective}$:
- $\text{Viewing volume = frustum (truncated pyramid), defined by:}$
  - $\text{View window (image plane edges)}$
  - $\text{Near and far clipping planes}$

$\text{Objects outside this volume are clipped and not rendered.}$  
$\text{This prevents rendering infinite geometry or visual artifacts.}$

<img src="image_U8/Screenshot 2025-05-29 at 15.29.57.png" height="325" width="400"/>

---

### $\text{Effects of Clipping}$

- $\text{Removes geometry too close/far or outside the field of view.}$
- $\text{Ensures efficient rendering and avoids distortion.}$
- $\text{Near clipping plane avoids divide-by-zero in perspective.}$

### $\text{Summary of Planar Geometric Projections and Vanishing Point Classifications}$

| $\text{Projection}$           | $\text{Type}$         | $\text{Projection Lines}$                        | $\text{Foreshortening}$           | $\text{View Plane Orientation}$             | $\text{Vanishing Points}$        | $\text{Geometric Properties Preserved}$                     |
|-------------------------------|-----------------------|--------------------------------------------------|------------------------------------|----------------------------------------------|-------------------------------|--------------------------------------------------------------|
| $\text{Parallel}$             | $\text{Parallel}$     | $\text{Parallel to each other}$                  | $\text{Possible (if oblique)}$     | $\text{Any orientation}$                     | $\text{Zero-point}$            | $\text{Parallelism}$                                         |
| $\text{Perspective}$          | $\text{Central}$      | $\text{Converge to center of projection}$        | $\text{Yes (realistic)}$           | $\text{Depends on viewpoint}$                | $\text{1, 2, 3 or k points}$    | $\text{Straight lines, not parallelism or ratios}$         |
| $\text{Orthographic}$         | $\text{Parallel}$     | $\text{Perpendicular to view plane}$             | $\text{None}$                      | $\text{Frontal (Top, Side, Front)}$          | $\text{Zero-point}$            | $\text{True dimensions, parallelism}$                       |
| $\text{Top}$                  | $\text{Orthographic}$ | $\text{Perpendicular to xy-plane}$               | $\text{None}$                      | $\text{From above}$                          | $\text{Zero-point}$            | $\text{Distances and angles in xy-plane}$                   |
| $\text{Front}$                | $\text{Orthographic}$ | $\text{Perpendicular to xz-plane}$               | $\text{None}$                      | $\text{From front}$                          | $\text{Zero-point}$            | $\text{Distances and angles in xz-plane}$                   |
| $\text{Side}$                 | $\text{Orthographic}$ | $\text{Perpendicular to yz-plane}$               | $\text{None}$                      | $\text{From left/right}$                     | $\text{Zero-point}$            | $\text{Distances and angles in yz-plane}$                   |
| $\text{Axonometric}$          | $\text{Parallel}$     | $\text{Tilted parallel lines}$                   | $\text{Yes (uniform)}$             | $\text{Axes equally skewed}$                 | $\text{Zero-point}$            | $\text{Parallelism, scale along axes, but not angles}$      |
| $\text{Oblique}$              | $\text{Parallel}$     | $\text{Not perpendicular to plane}$              | $\text{Yes (z-axis only)}$         | $\text{Slanted lines for depth}$             | $\text{Zero-point}$            | $\text{Front face true scale, depth distorted}$             |
| $\text{Cabinet}$              | $\text{Oblique}$      | $\text{Z-axis scaled by } \frac{1}{2}$           | $\text{Reduced foreshortening}$    | $\text{Z-axis appears halved}$               | $\text{Zero-point}$            | $\text{Parallelism, partial depth scaling}$                 |
| $\text{Cavalier}$             | $\text{Oblique}$      | $\text{Full-scale depth (z-axis)}$               | $\text{Yes (full)}$                | $\text{Z-axis projected at angle}$           | $\text{Zero-point}$            | $\text{Parallelism, depth not distorted}$                   |
| $\text{1-Point Perspective}$  | $\text{Perspective}$  | $\text{Lines converge along 1 axis}$             | $\text{Yes}$                       | $\text{Frontal plane facing viewer}$         | $\text{1 vanishing point}$     | $\text{Lines perpendicular to viewer foreshortened}$        |
| $\text{2-Point Perspective}$  | $\text{Perspective}$  | $\text{Lines converge along 2 axes}$             | $\text{Yes}$                       | $\text{Edges not parallel to view plane}$    | $\text{2 vanishing points}$    | $\text{Parallel lines in depth converge}$                   |
| $\text{3-Point Perspective}$  | $\text{Perspective}$  | $\text{Converge in all 3 axes}$                  | $\text{Yes (strongest effect)}$    | $\text{Tilted or high/low angles}$           | $\text{3 vanishing points}$    | $\text{No true parallelism or preserved scale} $            |
| $\text{k-Point Perspective}$  | $\text{Perspective}$  | $\text{Generalized to k axes}$                   | $\text{Yes}$                       | $\text{Extreme or curved scenes}$            | $\text{k vanishing points}$    | $\text{Only straightness preserved (not parallelism)}$      |
