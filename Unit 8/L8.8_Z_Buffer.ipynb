### Z-Buffer

### Review of the 3D rendering pipline, specifically focus on the Visibility stage and present the z-buffer Algorithm

$\text{Suppose we just project all points of all objects onto the vewing plane we will not get a correct image of the world, this is becasue objects may occlud each other.}$

- $\text{It's evident now that any object that are not with the space of the vewing plane will be clipped from the image and won't be seen.}$
- $\text{In the ray tracing (casting) we shot rays and via the parametric equation chose the smallest value of t that intersected with an object.}$
- $\text{In the current scenario we aren't shooting any rays - So how do we know which object should be rendered.}$


#### $\text{Naive Solution}$ 
1. $\text{Sort the objects by finding the LARGEST Z coordinate (BEFORE Projecting), it's largest since the direction of the camera is in the negative direction.}$
2. $\text{And render object from far to near (minimum z- value to maximum z-values).}$


#### $\text{Problem}$
$\text{The issue with this method is that, everytime we change the camera's perspective (position) we must resort (O(nlog(n))) and overlapping objects may not be sorted properly.}$

## $\text{Z-Buffer Algorithm}$

### $\text{Main Idea}$

$\text{The Z-buffer algorithm (or depth buffering) is used in 3D rendering to determine visible surfaces. For each pixel, it stores the z-depth of the closest surface.}$

---

### $\text{Formal Algorithm}$

$\text{Let the image resolution be } w \times h.$

$\text{For each pixel } (i, j):$

$$
Z(i,j) := \infty, \quad C(i,j) := \text{background color}
$$

$\text{For each polygon in the scene:}$

- $\text{For each pixel } (i,j) \text{ the polygon covers:}$
  - $\text{Compute interpolated depth } z \text{ at } (i,j)$
  - $\text{If } z < Z(i,j):$

$$
Z(i,j) := z, \quad C(i,j) := \text{polygon color at } (i,j)
$$

---


### $\text{Pseudocode}$

```python
for each pixel (i, j):
    Z[i][j] = ∞
    C[i][j] = background_color

for each polygon in scene:
    for each pixel (i, j) the polygon covers:
        z = depth_at(i, j)
        if z < Z[i][j]:
            Z[i][j] = z
            C[i][j] = color_at(i, j)
```
---

$\text{Suppose two overlapping triangles:}$

- $\text{Triangle A (red) is in front of Triangle B (blue)}$
- $\text{At pixel } (i,j): \quad z_A = 2.0,\ z_B = 5.0$

$\text{Then:}$

$$
Z(i,j) = 2.0, \quad C(i,j) = \text{red}
$$

$\text{Triangle A is visible at that pixel.}$

---

### $\text{Properties}$

- $\text{Time Complexity: } O(n)$ per polygon ($n =$ number of pixels it covers)
- $\text{Space Complexity: } O(w \cdot h)$
- $\text{Works in both hardware and software implementations}$
- $\text{Supports arbitrary overlapping geometry}$

---

### $\text{Limitations}$

- $\text{Memory usage increases with image resolution}$
- $\text{Transparency requires additional handling (e.g., sorting)}$
- $\text{Z-fighting occurs when two surfaces are extremely close in depth}$

---

### $\text{Conclusion}$

$\text{The Z-buffer algorithm is a foundational method for visibility determination in 3D rasterization.}$

$\text{It is efficient, GPU-accelerated, and widely used in real-time rendering pipelines.}$
