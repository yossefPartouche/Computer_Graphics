## Perspective Projection Properties

#### objective:
- See what properties of images is created with perspective projection

- All points after projection will lie on the projection plane, and as such all z-values will be the same and no information can be determined from this. So we focus on the x and y coordinates.
- We also notice that point that lie on the veiwing plane don't change after projection, this is because their z-value is exactly $-f$,  and their X and Y coordinates will be multiplied by the factor $\frac{f}{z} = 1$.

## Points on a plane that's Parallel to the veiwing plane

Such points are characterised in camera coordinate system by the equation $ z = k_i$ for some scalar value k. since all points on one of these planes have the same Z-coordinates their X and Y values will be multiplied by the same facor $\frac{-f}{k}$.

When discussing $\text{forshortening}$ the large the value of k is, the smaller the scaling factor $\frac{-f}{k}$ will be, thus will look smaller as k gets larger. 

At the same time angles and gradients will remain the same. Meaning lines that are parallel to the veiwing plane will remain parallel after the perspective pojection.

#### What about line that are not parallel to the veiwing plane

- These line will intersect the veiwing plane denote this point of intersection as $P_0 = (x_0, y_0, 0)$
- We can define the parametric representation of the line as: $$ P = P_0 + t\vec{v}$$ where $\vec{v}$ is the normalised direction of the line.
- As t goes from 0 to infinity, each point $P$ will projected on a different point on the viewing plane.
- As distances are foreshortened as t grows the distances between the projected point becomes smaller and smaller on the veiwing plane. 
- if we take the limit, they will all converge to a single point on the viewing plane $\text{Vanishing Point}$
- We can find the vanishing point bt connnecting the direction vector $\vec{v}$, to the centre of projection, and finding the intersection with the veiwing plane

#### $\text{Lines Not Parallel to the Viewing Plane}$

- $\text{Consider a line that intersects the viewing plane at } P_0 = (x_0, y_0, 0).$
- $\text{The parametric form of the line is:}$
$$
P(t) = P_0 + t \vec{v}
$$
where $\vec{v} = (v_x, v_y, v_z)$ is the normalized direction vector of the line.

- $\text{As } t \to \infty, \text{ the projected points } P(t) \text{ approach a single point on the viewing plane called the Vanishing Point.}$
- $\text{The vanishing point can be found by intersecting the line defined by } \vec{v} \\ \text{ starting at the center of projection (usually the origin) with the viewing plane } z=0.$

---

#### $\text{Example}$

Given:

- $\quad P_0 = (1, 2, 0)$
- $\quad \vec{v} = \left(\frac{1}{\sqrt{3}}, \frac{1}{\sqrt{3}}, \frac{1}{\sqrt{3}}\right)$

$\text{Parametric line:}$
$$
P(t) = \left(1, 2, 0\right) + t \left(\frac{1}{\sqrt{3}}, \frac{1}{\sqrt{3}}, \frac{1}{\sqrt{3}}\right)
$$

$\text{Vanishing point}$ is found by solving for $t$ such that $z=0$ when line starts at the origin with direction $\vec{v}$:
$$
P_v = t_v \vec{v} = (x_v, y_v, 0)
$$
Since $z_v = t_v \frac{1}{\sqrt{3}} = 0 \implies t_v = 0,$

$\text{which means the vanishing point is at the origin } (0,0,0).$





```python
import numpy as np

# Define direction vector (normalized)
v = np.array([1,1,1]) / np.sqrt(3)

# Define point on viewing plane
P0 = np.array([1, 2, 0])

# Define parametric function of the line
def P(t):
    return P0 + t * v

# Let's find t when the line starting at origin with direction v hits z=0
# Equation: z = t * v_z = 0 => t = 0 (vanishing point at origin)
t_vanish = 0
vanishing_point = t_vanish * v

print("Vanishing point (starting at origin):", vanishing_point)

# Show some points on the line for t > 0
for t in [0, 1, 2, 5, 10]:
    print(f"P({t}) =", P(t))
```

### $\text{Types of Perspective Projection}$

1. **One-Point Perspective**  
- $\text{There is a single vanishing point on the horizon line.}$  
- $\text{Used when objects face directly towards the viewer (e.g., looking down a hallway).}$  
- $\text{Parallel lines perpendicular to the viewing plane converge at one point.}$


<img src="image_U8/Screenshot 2025-05-29 at 14.58.40.png" height="325" width="400"/>


2. **Two-Point Perspective**  
- $\text{There are two vanishing points on the horizon line.}$  
- $\text{Used when objects are rotated so that edges recede in two directions.}$  
- $\text{Lines parallel to two different axes converge separately to two points.}$


<img src="image_U8/Screenshot 2025-05-29 at 14.59.17.png" height="325" width="1500"/>

3. **Three-Point Perspective**  
- $\text{There are three vanishing points: two on the horizon and one above or below.}$  
- $\text{Used for dramatic views like looking up at a tall building or down from above.}$  
- $\text{Lines parallel to three axes converge to three different vanishing points.}$

<img src="image_U8/Screenshot 2025-05-29 at 15.00.08.png" height="325" width="400"/>

---

$\text{Summary table:}$

| $\text{Perspective}$ | $\text{Vanishing Points}$ | $\text{Typical Use}$                         |
|---------------------|--------------------------|---------------------------------------------|
| One-Point           | 1 (on horizon)            | Facing objects head-on                       |
| Two-Point           | 2 (both on horizon)       | Objects rotated with two receding directions|
| Three-Point         | 3 (two on horizon, one vertical) | Extreme angles, tall buildings, aerial views|
