### Coordinate Systems in the Rendering Pipline
##### Objective
1. Learn about the main tasks of each transformation that occurs in the Rendering pipline process

- The rendering pipline operate of the **Vertices** of the triangles which represent the objects.  

<img src="image_U8/Screenshot 2025-05-29 at 09.09.05.png" height="600" width="400"/>


-  Every triangle of a 3D object in a 3D scene undergoes a series of transformations in the graphical pipeline.
    - We start with $p = (x, y, z, 1)$  in the 3D Object Coordinate System
    1. Modelling Transformation (First Stage of Pipeline)
        - Converts it to 3D world Coordinate System
    2. Veiwing Transformation (Third Stage of Pipeline)
        - Converts a vertex of 3D camera Coordinate System, where the origin in the camera coordinate system, is at the centre of the projection of the camera, and the axes are aligned witht the camera direction.
    3. Projection Transformation (Forth Stage of Pipeline)
        - The conversion to camera coordinate system simplifies the mathematical definition of the projection, because the origin of the 3D world in the new camera coordinate system lies at the centre of projectgion and the direction that the camera is facing is perpendicular to the projection prane (Normal to it).
        - Thus this process converts 3D coordinate to 2D coordiate System, which is defined in the screen coordinate system. 
    4. Window-to-Viewpoint Transformation (After Clipping Stage and Before Rasterisation Stage)
        - Map the 2D screen coordinate of the triangles to 2D image coordinates which are pixels in the final image.
- All of these transformation use the Affine tranformations that we've learnt in Uni 6 and Unit 7.
