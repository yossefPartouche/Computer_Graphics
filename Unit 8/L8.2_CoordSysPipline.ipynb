### Coordinate Systems in the Rendering Pipline
##### Objective
1. Learn about the main tasks of each transformation that occurs in the Rendering pipline process

$\boxed{\text{The rendering pipline operate on the } \bold{Vertices} \text{ of the triangles which represent the objects.}  }$

<img src="image_U8/Screenshot 2025-05-29 at 09.09.05.png" height="600" width="400"/>

|Stage| Explanantion | 
|-----|--------------|
|$$\text{Input} \\ \Downarrow$$ | $\text{We start with } p = (x, y, z, 1) \text{ Local axes/origin and possibly normals, texture coordinates in the 3D Object Coordinate System} \\ \text{Note the pipeline, doesn't really "care" about the mesh representation since it's only dealing with a set of vertices BUT} \\ \text{since it's a data structure used to organise vertices into surfaces - it helps with the efficiency of rendering and for us to understand.}$ |
|$$\text{Modelling Transformation} \\ \Downarrow$$ | $\text{Converts it to 3D World Coordinate system, where we compose our scene.} \\ \text{This is typically done by Scaling, Rotation, Translation} \\ \text{The 1st stage in the Pipeline process}$|
|$$\text{Veiwing Transformation} \\ \Downarrow$$ | $\text{Input: scene goemetry in } \bold{World Coordinates} \text{ and Camera Position} \\ \text{This stage moves and rotates the entire world/scene so that the camera is at the origin (0,0,0)} \\ \text{In standard practice the camera is facing the -Z direction (hence the world scene will be transformed in that direction as well)}$|
|$$\text{Projection Transformation} \\ \Downarrow$$| $\text{Input: 3D Coordinates in camera space} \\ \text{Transform 3D coordinates into 2D coordinates for the scene to simulate how the "human eye" percieves depth and perspective}$|
|$$\text{Window-Viewpoing Transformation}$$ | $\text{Input: 2D coordinates after projection (Usually in normalised device coordinates (NDC)} \\ \text{Convert 2D coordinates to viewport (device screen) coordinates using simple affine Transformations}$|
