## Finding angle $\alpha$ and $\phi$.

### $\alpha$
$\text{Goal: Find the angle}\ \alpha \ \text{between the direction of projection and the projection plane (i.e., the angle of obliqueness)}$.

<img src="image_U8/Screenshot 2025-05-29 at 10.39.32.png" height="325" width="400"/>


$\text{Method:}$

1. Let the original point be:
$$
C_0 = (x, y, z)
$$

2. The orthographic projection onto the plane $z = 0$ is:
$$
P_0 = (x, y, 0)
$$

3. Choose the obliquely projected point $P$ on the plane $z = 0$:
$$
P = (x_1, y_1, 0)
$$

4. Construct the following vectors:
- Projection direction vector:
$$
\vec{d} = P - C_0 = (x_1 - x, y_1 - y, -z)
$$

- Plane normal vector (since the plane is the xy-plane):
$$
\vec{n} = (0, 0, 1)
$$

5. Compute the angle $\alpha$ between the projection direction $\vec{d}$ and the plane using:
$$
\cos \alpha = \frac{ \vec{d} \cdot \vec{n} }{ \| \vec{d} \| \cdot \| \vec{n} \| }
$$

6. Compute each component:
- Dot product:
$$
\vec{d} \cdot \vec{n} = (x_1 - x)(0) + (y_1 - y)(0) + (-z)(1) = -z
$$

- Norms:
$$
\| \vec{d} \| = \sqrt{ (x_1 - x)^2 + (y_1 - y)^2 + z^2 }, \quad \| \vec{n} \| = 1
$$

7. Final formula:
$$
\cos \alpha = \frac{ -z }{ \sqrt{ (x_1 - x)^2 + (y_1 - y)^2 + z^2 } }
$$

8. Therefore:
$$
\alpha = \cos^{-1} \left( \frac{ -z }{ \sqrt{ (x_1 - x)^2 + (y_1 - y)^2 + z^2 } } \right)
$$

$\text{Note: The negative sign is due to the direction from the point down to the plane. If you want the acute angle, you can take the absolute value in the numerator}$.




### $\phi$

$\text{Goal: Find the angle of obliqueness } \phi \text{ between the projection direction } \vec{d} \text{ and its projection onto the plane (i.e., the shadow of } \vec{d} \text{ on the plane).}$

<img src="image_U8/Screenshot 2025-05-29 at 13.17.27.png" height="325" width="400"/>

$\text{Setup:}$

$\text{Let:}$
$$
\vec{d} = P - C_0 = (x_1 - x,\ y_1 - y,\ -z)
$$

$\text{The projection of } \vec{d} \text{ onto the } xy\text{-plane is:}$
$$
\vec{d}_{\text{xy}} = (x_1 - x,\ y_1 - y,\ 0)
$$

$\text{We compute the angle } \phi \text{ between } \vec{d} \text{ and } \vec{d}_{\text{xy}}.$

$\text{Step 1: Use the dot product formula:}$
$$
\cos \phi = \frac{ \vec{d} \cdot \vec{d}_{\text{xy}} }{ \| \vec{d} \| \cdot \| \vec{d}_{\text{xy}} \| }
$$

$\text{Step 2: Compute the dot product:}$
$$
\vec{d} \cdot \vec{d}_{\text{xy}} = (x_1 - x)^2 + (y_1 - y)^2 + (-z)(0) = (x_1 - x)^2 + (y_1 - y)^2
$$

$\text{Step 3: Compute the norms:}$
$$
\| \vec{d} \| = \sqrt{ (x_1 - x)^2 + (y_1 - y)^2 + z^2 }
$$
$$
\| \vec{d}_{\text{xy}} \| = \sqrt{ (x_1 - x)^2 + (y_1 - y)^2 }
$$

$\text{Step 4: Plug into the cosine formula:}$
$$
\cos \phi = \frac{ (x_1 - x)^2 + (y_1 - y)^2 }{ \sqrt{ (x_1 - x)^2 + (y_1 - y)^2 + z^2 } \cdot \sqrt{ (x_1 - x)^2 + (y_1 - y)^2 } }
$$

$\text{Step 5: Simplify:}$

$\text{Let:}$
$$
r^2 = (x_1 - x)^2 + (y_1 - y)^2
$$

$\text{Then:}$
$$
\cos \phi = \frac{ r }{ \sqrt{ r^2 + z^2 } }
$$

$\text{So:}$
$$
\phi = \cos^{-1} \left( \frac{ r }{ \sqrt{ r^2 + z^2 } } \right)
$$

$\text{Conclusion:}$

$\text{The angle } \phi \text{ is between the projection direction } \vec{d} \text{ and its shadow } \vec{d}_{\text{xy}} \text{ on the plane.}$
$\text{ The angle is small if the projection is nearly in-plane (low obliqueness), and large if there's a steep component in } z\text{.}$


## Finding Point of Projection given $(\alpha, \ \phi, \ C_0)$

$\text{Goal: Given } C_0 = (x, y, z),\ \alpha,\ \phi,\ \text{find the point of oblique projection } P \text{ onto the plane } z = 0.$

$\text{Step 1: Interpret the angles.}$

$\text{Let } \vec{d} \text{ be the unit direction vector of projection.}$

$\alpha = \text{angle between } \vec{d} \text{ and the plane normal } (0,0,1)$

$\phi = \text{angle between } \vec{d} \text{ and its projection onto the } xy\text{-plane}$

$\text{Then:}$
$$
\cos \alpha = \frac{ |d_z| }{ \|\vec{d}\| } = |d_z| \Rightarrow d_z = -\cos \alpha
$$

$\text{So the length of the shadow is:}$
$$
\| \vec{d}_{\text{xy}} \| = \sin \alpha
$$

$\text{Use } \phi \text{ to determine the direction of the shadow in the } xy\text{-plane. Let the shadow vector be:}$

$$
\vec{d}_{\text{xy}} = (\cos \phi,\ \sin \phi,\ 0)
$$

$\text{Then the full (unit) projection direction vector is:}$

$$
\vec{d} = \sin \alpha \cdot (\cos \phi,\ \sin \phi,\ 0) + (-\cos \alpha) \cdot (0, 0, 1) = (\sin \alpha \cos \phi,\ \sin \alpha \sin \phi,\ -\cos \alpha)
$$

$\text{Step 2: Define the parametric ray from } C_0 \text{ in direction } \vec{d}:$
$$
C(t) = C_0 + t \vec{d} = (x, y, z) + t (\sin \alpha \cos \phi,\ \sin \alpha \sin \phi,\ -\cos \alpha)
$$

$\text{Step 3: Find the value of } t \text{ such that } C(t) \text{ lies on the plane } z = 0.$

$\text{Set the third component to 0:}$

$$
z - t \cos \alpha = 0 \Rightarrow t = \frac{z}{\cos \alpha}
$$

$\text{Step 4: Plug } t \text{ back into } C(t) \text{ to get the oblique projection point } P:$

$$
P = C\left(\frac{z}{\cos \alpha}\right) = (x, y, z) + \frac{z}{\cos \alpha} (\sin \alpha \cos \phi,\ \sin \alpha \sin \phi,\ -\cos \alpha)
$$

$\text{Simplify:}$

$$
P = \left( x + z \tan \alpha \cos \phi,\ y + z \tan \alpha \sin \phi,\ 0 \right)
$$

$\text{Final Answer:}$

$$
P = \left( x + z \tan \alpha \cos \phi,\ y + z \tan \alpha \sin \phi,\ 0 \right)
$$




$\text{This is the oblique projection of } C_0 \text{ in the direction defined by } \alpha \text{ and } \phi.$

$\text{Oblique projection matrix onto } z=0 \text{ plane:}$
$$
P =
\begin{bmatrix}
1 & 0 & \tan \alpha \cos \phi & 0 \\
0 & 1 & \tan \alpha \sin \phi & 0 \\
0 & 0 & 0 & 0 \\
0 & 0 & 1 & 1
\end{bmatrix}
$$

$\text{Given a point in homogeneous coordinates:}$
$$
\mathbf{P}_{\text{orig}} =
\begin{bmatrix}
x \\
y \\
z \\
1
\end{bmatrix}
$$

$\text{Multiply the matrix by the point:}$

$$
\mathbf{P}_{\text{proj}} = P \cdot \mathbf{P}_{\text{orig}} =
\begin{bmatrix}
1 & 0 & \tan \alpha \cos \phi & 0 \\
0 & 1 & \tan \alpha \sin \phi & 0 \\
0 & 0 & 0 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix}
\cdot
\begin{bmatrix}
x \\
y \\
z \\
1
\end{bmatrix}
=
\begin{bmatrix}
x + z \tan \alpha \cos \phi \\
y + z \tan \alpha \sin \phi \\
0 \\
1
\end{bmatrix}
$$

$\text{Ignoring the homogeneous scale (since this is an affine projection), the Cartesian coordinates of the projected point are:}$

$$
\left( x + z \tan \alpha \cos \phi,\quad y + z \tan \alpha \sin \phi,\quad 0 \right)
$$

$\text{The following matrix defines the oblique transformation using scaling factor a}$

\begin{bmatrix}
1 & 0 & a\cos \phi & 0 \\
0 & 1 & a\sin \phi & 0 \\
0 & 0 & 0 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix}

### Distance between $P$ and $P_0$

$$d = \frac{z}{\tan{\alpha}}$$

#### Extra
$\text{If you have two points} \ P_1 \ and \ P_0 \ \text{whose distance is} \ d \ \text{then the distance between these two points when projected obliquely is} \frac{d}{\tan{\alpha}}$

### Special Types of Onlique projections

#### Cavalier Projection

- $a=1$ so $\alpha = 45$


\begin{bmatrix}
1 & 0 & \cos \phi & 0 \\
0 & 1 & \sin \phi & 0 \\
0 & 0 & 0 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix}

#### Cabinet Projection

- a = 1/2 so $\alpha = 26.5 (roughly)$
\begin{bmatrix}
1 & 0 & \frac{1}{2}\cos \phi & 0 \\
0 & 1 & \frac{1}{2}\sin \phi & 0 \\
0 & 0 & 0 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix}
