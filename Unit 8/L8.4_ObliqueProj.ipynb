### Oblique Projections

##### Objective:
1. Discuss Oblique projections (Sub-Parallel projection) but not perpendicular to the viewing plane.

$\text{Oblique Projection} := \text{A parallel prjection when the direction of projection is not Perpendicular to the viewing plane}$

### What we need:
2 Angles relative to the veiwing plane

### What to do: 
1. You have a starting point $C_0 = (x,y,z)$
2. Apply Orthograpic Projection $P_0 = (x,y,0)$. (This brings you to a point on the plane) 
3. Choose any other point which satisfy the equation of the plane $P = (x_1, y_1, 0)$ (This is usually the point we wanted to project).
