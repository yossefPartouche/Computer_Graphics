### 3D Orthographic Projections

#### Objective:
1. Discuss 3D projection in the context of Graphics

$\text{Projections}$ := $\text{Linear Transformation P from a vector space to itself such that} \ P^2 = P$
- A projection that is not the $I_2$ is **NOT INVERTIBLE**
- The geometric operation of a non-trivial projection isn't invertible, so once you apply a projection you can go back and information is lost.
    - Given a 2D point that is a projection of 3D point, you wouldn't be able to determine which 3D point was projected on the 2D point (because there are infinite number of points that could be projeccted on this point)


- We will be focusing on the projections from 3D to 2D

#### Assumption: 
1. The camera Lies at the origin of the coordinate system.
2. The 2D plane of projection is always parallel to the camera veiwing plane
3. The veiwing plane is placed in front of the camera
4. Any point in the world coordinate system is given to us in the camera coordinate system.
