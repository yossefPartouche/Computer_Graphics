- We need 2 component to define a projection 
    1. Plane of projection 
    2. Direction of Projection

### Question 
The 3D point A=(1,1,-3) is projected using a parallel projection into an image plane parallel to the xy plane. If the projected point is B=(2,3,-1), what is the direction of projection?

$\textbf{Given:}$

$$
A = (1, 1, -3)
$$

$$
B = (2, 3, -1)
$$

Parallel projection onto a plane parallel to the \( xy \)-plane.

$\text{Goal: Find the direction of projection}$.

In a parallel projection, all projection lines are parallel and point in a constant direction.

The direction of projection is the vector from point \( A \) to point \( B \):

$$
\vec{v} = \vec{B} - \vec{A}
$$

$$
\vec{v} = 
\begin{pmatrix}
2 \\
3 \\
-1
\end{pmatrix}
-
\begin{pmatrix}
1 \\
1 \\
-3
\end{pmatrix}
=
\begin{pmatrix}
1 \\
2 \\
2
\end{pmatrix}
$$

$\text{Answer:}$

$$
\text{Direction of projection} = \boxed{(1,\ 2,\ 2)}
$$

#### Axonometrix Projection
$ \text{Axonometrix Projection :=A parallel projection where the object is rotated along one or more of its axes relative to the plane of projection}$
- This will reveal more than one side of the object, thus will be able to show more of the 3D structure than the Side/Front/Top projections.
