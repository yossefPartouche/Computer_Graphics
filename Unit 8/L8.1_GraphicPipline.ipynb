#### Renderring 
- $\text{This is the process of converting a 3D virtual scene in some representation to a 2D image}$.
- $\text{The transformation Responsible for renderring is Projection}$

#### Component of Rendering Scene: 
1. Camera Model
2. Veiwing plane (Where the 2D image is create)
3. Light 

#### Reminder
1. Ray tracing places the camera at a certain place in the scene
2. Positions the image canvas in front of it on the veiwing plane
3. Shoots rays from the camera, through the pixel of the image into the scene.

## Graphical Rendering Pipline - Background
- This algorithm is usually supported by a hardware called GPU.
- $ \text{GPU} = \text{A hardware component that implement that Graphical Rendering Pipeline}$

We can rerpresent objects in a 3D scene via:
- Implicit Representation.
- Parametric Representation.
- Model the object using Triangles
We can the convert the 3D representation of the object into a mesh of triangles (**Triangulation**).


We use triangles since the GPU can be viewed as a "Triangle Rendering Machine", where it converts triangles to pixels qucikly and efficiently.
#### The reason Triangles are used by the GPU
1. Simplest Polygon
2. Every other polygon can be triangulated. 
3. Higher degree surfaces can be approximated to any extent by triangles.
4. They are alway planar - The have a single normal vector.
5. Bilinear interpolation gives reasonable results (for Color, texture, normal evaluation).
6. Simplifies hardware design and implementation of GPU.

## Graphical Rendering Pipeine Process


| **Stage**                    | **Description**                                                                 | **Relevant Sub-Processes**                                                                 |
|-----------------------------|----------------------------------------------------------------------------------|---------------------------------------------------------------------------------------------|
| **1. Modeling Transformation** | Convert from **local (object)** space to **world** space                        | - Object geometry definition  <br> - Transformations: Translation, Rotation, Scaling       |
| **2. Lighting**             | Compute **illumination** for each vertex or fragment                            | - Surface normals calculation  <br> - Light source modeling (ambient, diffuse, specular)   |
|                             |                                                                                  | - Material properties <br> - Shading models: Phong, Blinn-Phong, etc.                      |
| **3. Viewing Transformation** | Change from **world** coordinates to **camera (view)** coordinates               | - Camera positioning and orientation  <br> - View matrix construction                      |
|                             |                                                                                  | - Right-handed to left-handed conversions (if applicable)                                 |
| **4. Projection Transformation** | Project 3D points onto a 2D screen plane                                    | - Perspective or orthographic projection  <br> - Projection matrix setup <br> - Depth (z) normalization                                                                  
| **5. Clipping**             | Remove parts of primitives outside **view volume**                               | - Frustum culling <br> - Near/far plane clipping <br> - Triangle splitting if necessary    |
| **6. Rasterization**        | Convert primitives into **pixel fragments**                                      | - Scan conversion <br> - Texture mapping <br> - Interpolation (color, depth, normals, UV)  |
|                             |                                                                                  | - Fragment shading <br> - Occlusion handling (z-buffer)                                    |
