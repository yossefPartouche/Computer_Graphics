#### Renderring 
- $\text{This is the process of converting a 3D virtual scene in some representation to a 2D image}$.
- The transformation Responsible for renderring is **Projection**

#### Component of Rendering Scene: 
1. Camera Model
2. Veiwing plane (Where the 2D image is create)
3. Light 

#### Reminder
1. Ray tracing places the camera at a certain place in the scene
2. Positions the image canvas in fron of it on the veiwing plane
3. Shoots rays from the camera, through the pixel of the image into the scene.

## Graphical Rendering Pipline - Background
- This algorithm is usually supported by a hardware called GPU.
- $ \text{GPU} = \text{A hardware component that implement that Graphical Rendering Pipeline}$

We can rerpresent objects in a 3D scene via:
- Implicit Representation.
- Parametric Representation.
- Model the object using Triangles
We can the convert the 3D representation of the object into a mesh of triangles (**Triangulation**).


We use triangles since the GPU can be viewed as a "Triangle Rendering Machine", where it converts triangles to pixels qucikly and efficiently.
#### The reason Triangles are used by the GPU
1. Simplest Polygon
2. Every other polygon can be triangulated. 
3. Higher degree surfaces can be approximated to any extent by triangles.
4. They are alway planar - The have a single normal vector.
5. Bilinear interpolation gives reasonable results (for Color, texture, normal evaluation).
6. Siplifies hardware design and implementation of GPU.

## Graphical Rendering Pipline Process

#### 1. Modeling Tranformation
- Transform an object from Local to Global Coordinate system
#### 2. Lighting 
- Computing the colours of the vertices of the triangles in the scene (Very similar to Ray-Casting)
#### 3. Viewing Transformation
- Affine transformation changing the coordinate systems, changing the vertices from the $\text{World coordinate system}$ to the $\text{camera coordinate system}$.
#### 4. Prjection Transformation
- Transform from a 3D world to a 2D world
#### 5. Clipping 
- Clip primitives outside the camera's view
#### 6. Rasterisation 
- Converting the 2D scene into pixels (including texturing, occlusion etc)
