#### Families of Transformation

#### Objective:
To discuss different freedoms of different subsets of transformation families

Transformation Family := A subset of transformation type that is closed under composition



#### Rotation Family
| **Component**             | **Explanation**                                  | **Degrees of Freedom (DoF)** |
|---------------------------|--------------------------------------------------|------------------------------|
| Rotation Axis Direction   | Normalized vector (2 DoF)                        | 2                            |
| Rotation Angle            | Angle of rotation around the axis                | 1                            |
| **Total**                 |                                                  | **3**                        |

<br>

| **Property**                            | **Holds?** | **Notes**                                              |
|----------------------------------------|------------|--------------------------------------------------------|
| Preserves distances                    | ✅ Yes     | Rotation is an isometry                                |
| Preserves angles                       | ✅ Yes     | Pure rotation doesn't alter angles                     |
| Preserves ratios along a line          | ✅ Yes     | Proportions along rotated objects stay the same        |
| Closed under composition               | ✅ Yes     | Composition of rotations yields another rotation       |
| Maps lines to lines                    | ✅ Yes     | No curvature introduced                                |
| Keeps parallel lines parallel          | ❌ No      | Can change orientation of parallel lines               |
| Maps origin to origin                  | ✅ Yes     | No translation involved                                |


#### Translation Family
| **Component**   | **Explanation**                   | **Degrees of Freedom (DoF)** |
|-----------------|-----------------------------------|------------------------------|
| Translation     | Shifts in x, y, z directions      | 3                            |
| **Total**       |                                   | **3**                        |

<br>

| **Property**                            | **Holds?** | **Notes**                                         |
|----------------------------------------|------------|--------------------------------------------------|
| Preserves distances                    | ✅ Yes     | Pure translations are isometries                |
| Preserves angles                       | ✅ Yes     | No rotation or scaling to alter angles          |
| Preserves ratios along a line          | ✅ Yes     | Relative positions are unchanged                |
| Closed under composition               | ✅ Yes     | Combining translations gives another translation |
| Maps lines to lines                    | ✅ Yes     | No distortion to linearity                      |
| Keeps parallel lines parallel          | ✅ Yes     | Entire geometry shifts uniformly                |
| Maps origin to origin                  | ❌ No      | Origin is shifted unless translation is zero    |


#### Rigid Transformations
| **Component**          | **Explanation**           | **Degrees of Freedom (DoF)** |
|------------------------|----------------------------|------------------------------|
| Rotation               | Change in orientation      | 3                            |
| Translation            | Shift in position          | 3                            |
| **Total**              |                            | **6**                        |

<br>

| **Property**                            | **Holds?** | **Notes**                                           |
|----------------------------------------|------------|----------------------------------------------------|
| Preserves distances                    | ✅ Yes     | Rigid transformations are isometries               |
| Preserves angles                       | ✅ Yes     | Angles remain unchanged                            |
| Preserves ratios along a line          | ✅ Yes     | Proportions along lines are preserved              |
| Closed under composition               | ✅ Yes     | Composition of two rigid transforms is rigid       |
| Maps lines to lines                    | ✅ Yes     | No curves introduced                               |
| Keeps parallel lines parallel          | ✅ Yes     | Parallelism is preserved                           |
| Maps origin to origin                  | ❌ Not necessarily | Depends on translation vector               |

#### Similarity Transformations
| **Component**          | **Explanation**                          | **Degrees of Freedom (DoF)** |
|------------------------|-------------------------------------------|------------------------------|
| Rotation               | Orientation change                        | 3                            |
| Translation            | Shift in space                            | 3                            |
| Uniform Scale          | Same scale factor along all axes          | 1                            |
| **Total**              |                                           | **7**                        |

<br>

| **Property**                            | **Holds?** | **Notes**                                            |
|----------------------------------------|------------|-----------------------------------------------------|
| Preserves distances                    | ❌ No      | Scales all lengths equally, but changes absolute size |
| Preserves angles                       | ✅ Yes     | Because scale is uniform, angles stay the same        |
| Preserves ratios along a line          | ✅ Yes     | Midpoints and divisions on a line stay consistent     |
| Closed under composition               | ✅ Yes     | Combining two similarity transforms gives another     |
| Maps lines to lines                    | ✅ Yes     | Straight lines remain straight                        |
| Keeps parallel lines parallel          | ✅ Yes     | Parallelism is preserved under uniform scaling        |
| Maps origin to origin                  | ❌ Not necessarily | Depends on the translation part               |


#### General 3D Affine Transformation
| **Component**          | **Explanation**                         | **Degrees of Freedom (DoF)** |
|------------------------|------------------------------------------|------------------------------|
| Linear Transformation  | Rotation, scaling, and shearing          | 9                            |
| Translation Vector     | Shifts along x, y, z                     | 3                            |
| **Total**              |                                          | **12**                       |

<br>

| **Property**                            | **Holds?** | **Notes**                                         |
|----------------------------------------|------------|--------------------------------------------------|
| Preserves distances                    | ❌ No      | Lengths can change                               |
| Preserves angles                       | ❌ No      | Can distort angles                               |
| Preserves ratios along a line          | ✅ Yes     | E.g., midpoint stays midpoint                    |
| Closed under composition               | ✅ Yes     | Applying two affine transforms gives another     |
| Maps lines to lines                    | ✅ Yes     | Curves aren't introduced                         |
| Keeps parallel lines parallel          | ✅ Yes     | Affine maps preserve parallelism                 |
| Maps origin to origin                  | ❌ Not necessarily | Depends on translation vector               |


#### Projective Transformation

| **Component**                 | **Explanation**                                           | **Degrees of Freedom (DoF)** |
|-------------------------------|-----------------------------------------------------------|------------------------------|
| General 4×4 Matrix            | 16 parameters, minus 1 due to scale (homogeneous coords)  | 15                           |
| **Total**                     |                                                           | **15**                       |

<br>

| **Property**                            | **Holds?** | **Notes**                                                    |
|----------------------------------------|------------|--------------------------------------------------------------|
| Preserves distances                    | ❌ No      | Lengths can change arbitrarily                               |
| Preserves angles                       | ❌ No      | Angles are not preserved                                     |
| Preserves ratios along a line          | ❌ No      | Cross-ratios are preserved, but not simple linear ratios     |
| Closed under composition               | ✅ Yes     | Composition yields another projective transformation         |
| Maps lines to lines                    | ✅ Yes     | Straight lines stay straight                                 |
| Keeps parallel lines parallel          | ❌ No      | Parallel lines may converge (e.g. perspective)               |
| Maps origin to origin                  | ❌ Not necessarily | Depends on the matrix used                             |

#### Linear Transformation

| **Component**        | **Explanation**                             | **Degrees of Freedom (DoF)** |
|----------------------|---------------------------------------------|------------------------------|
| Linear 3×3 Matrix     | Includes rotation, scaling, shearing        | 9                            |
| **Total**            |                                             | **9**                        |

<br>

| **Property**                            | **Holds?** | **Notes**                                                    |
|----------------------------------------|------------|--------------------------------------------------------------|
| Preserves distances                    | ❌ No      | Only preserved in special cases (e.g., rotation only)        |
| Preserves angles                       | ❌ No      | Shearing and non-uniform scaling distort angles              |
| Preserves ratios along a line          | ✅ Yes     | Linear maps preserve ratios of segments on a line            |
| Closed under composition               | ✅ Yes     | Two linear maps composed yield another linear transformation |
| Maps lines to lines                    | ✅ Yes     | Always maps lines to lines                                   |
| Keeps parallel lines parallel          | ✅ Yes     | Parallelism is preserved                                     |
| Maps origin to origin                  | ✅ Yes     | Linear maps always map (0,0,0) to (0,0,0)                     |
