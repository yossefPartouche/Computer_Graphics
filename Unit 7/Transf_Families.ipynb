#### Families of Transformation

#### Objective:
To discuss different freedoms of different subsets of transformation families

Transformation Family := A subset of transformation type that is closed under composition

#### Rotation Family
- They remain a rotation transformation under composition
- A general vector is defined by three parameter: (x,yz)  
if the vector is normalised then $x^2 + y^2 + z^2 = 1$ 
and this means we constrain one DoF and we need just two parameters to define a normalised vector or direction.
- 3 DoF
    1) 2 DoF to define the axis/direction of rotation
    2) 1 DoF to define the angle of rotation 
- They preseve distances (Isometries)

#### Translation Family
- They remain translation transformation under composition
- 3 DoF: One DoF for each direction in which we translate.
- They preseve distances (Isometries)

#### Rigid Transformations
- Composed of Rotation and Translation
- 6 DoF 
- Preseve distances (Isometries)

#### Similarity Transformations
- Composed of Rotation, Translation and **Uniform** Scale
- Distance **No Longer Preserved**
- Closed under composition
- 7 DoF (1 more DoF on the scale)
- They preseve angle

#### General 3D Affine Transformation
- 12 DoF
- Don't Preseve Distances
- Don't preseve angles in General
- The preseve ratios on lines
- Closed under composition
- Lines are mapped to lines
- Parallel lines remain paralle.
- The origin does not necessarily map to Origin

#### Projective Transformation
- Any invertible 4x4 matrix define projective trasnformation
- 15 DoF
- We need the mapping of 5 points to define a projective transformation
