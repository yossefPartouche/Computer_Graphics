#### Animation with Transformation
##### Objective: 
How we represent a model hierarchy and show the advantages of this method for movement and animation

Suppose we want to animate a person walking: 
- This is defined by a set of rigid limbs
- Each limb is connected to another limb at a joint. 
- Each joint has a certain number of DoF that allow different types of movements in a specific direction.

<img src="BodyHi.png" height="400" width="400"/>

We obtain a hierarchy of limbs connected by joints.
- Each limb in the hierarchy is a node, with their own local coordinate system.
- Each edge represent a transformation from a child coordinate system, to the parent coordinate system.

For Example: To find the position fo the palm fo the hand, we need to follow the tranformations from the palm through the rist elbow shoulders etc etc, until we reach the root and multiply all the transformation matrices along the way.

With this method we can animate a persons movement since, to move the palm of teh hand we **modify the wrist transformation** and the palm will move. But if we want to move the whole arm then we only need to modify **Shoulders transformation**

We maintain this by holding a **STACK Data Structure**  of transformations, we place the identity matrix (No transformation applied). When you travel down from the root, we push the transformation matrix of the edge, by multiplying it with the top matrix in the stack. This means that the transformation of each node is always represented by the matrix at the top of the stack, also being the product of all the ansestors matrices from the root. To travel up the hierarchy we only need to pop every level going up, and the correct matrix will represent the correct subset of limbs we wish to move.
