#### Affine Maps And Homogeneous Coordinates
##### Objectives: 
To learn about 3D Affine Transformations and 3D Coordinates

3D Objects in Graphics are defined in their own, local coordinate System (i.e. the image on their own), and then placed in the global coorddinate system (i.e. the scene containing many objects).

### Example of 3D Chair

To construct the chair we use a "mesh" Representation, and all other objects for that matter. The mesh is composed of triangles that create the shape of the chair.  

To define the position of these triangles we need to define the position of the vertices of each triangle. We use affine points, in the affine space to do so. This means we need the define an affine space first, we'll use the cartesian coordinate system, let $u, \ v, \ w$ for the axises. We'll set the origin of the coordinate system as one of legs of the chairs. This will be our **Local Coordinate System** for the chair.  

Suppose we want to place this chair in a room scene, this room will be defined in another coordinate system, the **Global coordinate system** and let $x, \ y, \ z$ for the axises, and we'll set the origin to be one of the corners of the room.

To place the chair in any position in the room, we must **Map** the origin point of the local system in desired spot in the room and also map to scale the chair to appropriate size.

#### How is this done?

There are 2 parts:
1. Translation map
2. Scaling map

The only difference (currently) is that we've increased the dimension:
$$ x = x_0 + u•S_x$$  
$$ y = y_0 + v•S_y$$
$$ z = z_0 + w•S_z$$  

Where $S_i$ are the scaling factors for the maped coordinates.   
When we scale uniformly, $S_x=S_y=S_z$


We can make present this in a more compact form:

$$\begin{pmatrix} x \\ y \\ z \end{pmatrix} = \begin{pmatrix} x_0 \\ y_0 \\ z_0 \end{pmatrix} + \begin{pmatrix} S_x & 0 & 0 \\ 0 & S_y & 0 \\ 0 & 0 & S_z \end{pmatrix} \begin{pmatrix} u \\ v \\ w \end{pmatrix}$$


This is indeed an **Affine Mapping** (just in 3D) as we have the translation part: $\begin{pmatrix} x \\ y \\ z \end{pmatrix}$

and then scaling part:  $\begin{pmatrix} S_x & 0 & 0 \\ 0 & S_y & 0 \\ 0 & 0 & S_z \end{pmatrix} \begin{pmatrix} u \\ v \\ w \end{pmatrix}$

To make this complete we have to use homogeneous coordinates so we generalise:  

**All homogeneous coordinates points of the form $(Wx_1, Wx_2, ..., Wx_n)$ represent the same Euclidean point $(x_1, x_2, ..., x_n)$ and in particular $(x_1, x_2, ..., x_n,1)$**  
(We start at some n+1 dimension to arrive to our dimension).
