#### Combining 3D Transformations

##### Objective:
How to deal with more complex transformation by rotating around an aribitrary axis in 3D space

- Like in 2D, 3D affine transformations are closed under compositions.
We use this fact to show that a rotation around an arbituary axis is an Affine Transformation that can be composed of simpler (Affine) transformations.  

- We do this by aligining the rotation axis with one of the major axes, once the axis is aligned we perform the rotation in $\theta$ degrees. Then apply the inverse of the alignment transforamtions, to place the axis back in its place.

##### How to align: 

Suppose we wanted to align the rotation axis with the Z-Axis: 
1) Rotate Around the X-Axis
    - This is so that the rotation axis will be on the XZ-Plane
2) We then rotate around the Y-Axis, until the line coincides with the Z-Axis
3) Rotate around the Z-Axis. 
4) Apply the inverse of the Y-Axis Rotation.
5) Apple the inverse of the X-Axis Rotation.

In Conclusion:   
We applied 5 Rotation transformations: $$ R_{final} = R_x^{-1}R_y^{-1}R_zR_yR_x$$ (Operating from the right first)

The above assume that the line of axis (the one about we wish to rotate), **Passes through the Origin**.  
##### What if it didn't?

1) Choose a point that lie of the given Axis
2) Apply Translation so that the point is at the origin
3) Apply the above Compositions of roations
4) Apply back the inverse Translation. 


In Conclusion:   
We applied 7 Transformations: $$ R_{final} = T^{-1}R_x^{-1}R_y^{-1}R_zR_yR_xT$$ (Operating from the right first)
