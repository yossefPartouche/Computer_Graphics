### Changing Coordinate Systems

Objective: 
We revisit the question, How many mappings are needed to uniquely define Affine transformations, so that we can interprete a mapping from one coordinate system to another.

#### How many mappings of points are needed to uniquely define an Affine transformation?

- An Affine Transformation has 12 DoF
$M = 
\begin{bmatrix}
\color{red}{M_{11}} & \color{red}{M_{12}} & \color{red}{M_{13}} & \color{green}{M_{14}} \\
\color{red}{M_{21}} & \color{red}{M_{22}} & \color{red}{M_{23}} & \color{green}{M_{24}} \\
\color{red}{M_{31}} & \color{red}{M_{32}} & \color{red}{M_{33}} & \color{green}{M_{34}} \\
0 & 0 & 0 & 1
\end{bmatrix}
$

- **🔴 Red entries** are the 9 **linear transformation** parameters (rotation, scaling, shearing).
- **🔵 Blue entries** are the 3 **translation** parameters.
- The bottom row $([0~0~0~1])$ ensures it is a proper affine matrix in homogeneous coordinates.

Since each mapping of a point provides 3 constraints, so we need to have 4 3D points to have 12 constraints and to uniquely deinf an affine transformation.


#### What is the geometric interpretation of this fact?
- Four points in general position in 3D define a tetrahedron
    1) we can designate 1 point as the point of origin
    2) use the other points, as 3 different vectors from the origin
    3) these vectors define a BASE for the 3D vector space
- Given a mapping of 4 source points to 4 target points, we can view this mapping as a mapping from one coordinate system to another.
- From this we can see that an affine transformation in 3D can be viewed as a change of coordinate system.

#### How Do we find the actual transformation from one coordinate system to another?
- Suppose we have 2 orthonormal Coordinate systems $(\vec{X}$, $\vec{Y}$ and $\vec{Z})$ and $(\vec{u}$, $\vec{v}$ and $\vec{w})$ 
- Assume that that both have the same origin point.
- In this case the affine transformation that maps from to the other is indeed a Rotation.

1) A rotation matrix is an orthonormal matrix.
    - So in the first row we place the entries of u (last col = 0)
    - In the second row we place the entries of v (last col = 0)
    - In the third row we place the entries of w (last col = 0)
2) With this matrix we then multiply 3 times with a different base vector:
    - onces with the u vector (last row is 1) 
    - onces with the v vector (last row is 1)
    - onces with the w vector (last row is 1)
3) for each multiplcation you will respectively obtain:
    - X-Axis
    - Y-Axis
    - Z-Axis

NOTE: If we transpose the matrix and multiply it out (i.e. run step 2 and 3) with X,Y and Z axes you will obtain:
- U
- V
- W

Suppose given the X,Y,Z coordinate system (meaning X, Y,Z are basis and (0,0,0) is the origin) we want to rotate about an Axis W (assume it's passes thorugh (0,0,0))
- Let W be one of the basis and now we wish to find 2 more vectors to be basis of W.
- Normalise the vector W.
- Use the cross product of W and one of the coordinate Axis $\vec{W}$ x $\vec{X}$ (This order)
- Then normalise this vector and Denote V.
- Then use cross product of W and W: $\vec{W}$ x $\vec{V}$ (This order)
- Then normalise this vector and Denote U.  

We now have three orthonormal basis and can apply the rotation as shown above.
