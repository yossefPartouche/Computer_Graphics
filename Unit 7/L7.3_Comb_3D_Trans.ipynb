#### Combining 3D Transformations

##### Objective:
How to deal with more complex transformation by rotating around an aribitrary axis in 3D space

- Like in 2D, 3D affine transformations are closed under compositions.
We use this fact to show that a rotation around an arbituary axis is an Affine Transformation that can be composed of simpler (Affine) transformations.  

- We do this by aligining the rotation axis with one of the major axes, once the axis is aligned we perform the rotation in $\theta$ degrees. Then apply the inverse of the alignment transforamtions, to place the axis back in its place.

$\text{The above assume that the line of axis (the one about we wish to rotate) }$, **Passes through the Origin**.

### What if the line doesn't pass through the axis 

1) $\text{Choose a point that lie of the given Axis}$
2) $\text{Apply Translation so that the point is at the origin}$
3) $\text{Apply the above Compositions of roations}$
4) $\text{Apply back the inverse Translation. }$


$\text{In conclusion, We applied 7 Transformations:}$

$$ R_{final} = \left(T^{-1}R_x^{-1}R_y^{-1}R_zR_yR_xT\right) \cdot p$$
