#### Transformation Hierarchy
##### Objective: 
To understand how the mapping from one coordinate system to another will help us create objects in scenes and animate them

Many complex objects are just compositions (in the non-mathematical sense) of simplier objects, and thus produce a complex object.
When modelling a simple part of an object it's usually defined in its own local coordinate system, with that each point of an object needs to be tranformed in the correct values in the global coordinate system.

##### The problem : Very Inefficient to do this over many points!!
- Small change leads to many many recomputations.
- Animated Scenes, have coordinates changing the whole time.
- Multiple clones of objects would need to be delt with seperately
- Change in the scene (global Coordinate System).

#### The Solution: Create a hierarchy of tranformation

##### Example: Chair in a House
- If we have the chair coordinate system, the house coordinate system and the house coordinate system etc...
- When we want to place the chair in the room - Just find the Affine transformation from chair coordinate to Room coordinate system. To then find position of each of the chairs vertices, we multiply its coordinates vector by the matrix of transformation, with that obtain the position of the vertex in the room coordinate system. and the same is applied between the room and the house etc etc.
- To get the position of a chair's vertex V in the house coordinate System:
1) multiply first by the matrix transforming from the chair coordinate system to the room coordinate system.
2) multiply by the matrix from the rooom coordinate system to the house coordinate system.
3) doing this for all poisitons of the chair, it will be positioned correctly in the house.

$$V_{new} = M_{room-house}•M_{chair-room}•V$$
4) We can go on and multiply it by the street-coordinate system:
$$V_{new} = M_{house-stret}•M_{room-house}•M_{chair-room}•V$$

**NOTE** If we want to preserve the shape of the object then we must use similarity matrix transfirmation

##### The benefit:
- If want to change a perticular scene/object etc, we just switch out the old matrix with the new one.
- If we want to animate the object, we only need to change the mapping of the car to street matrix as the car progresses (main via translation).
- We can copy/duplicate objects and scenes very easily by multiplying by another transformation (representating a different object in a different scene).
