#### Down Scaling an Image

Some few reasons to down scale:
- Fit image on camera display
- Fit image in a page on a web-page/app
- Reduce storage size
- Send faster in an email.

There are two types of scaling:
1. Uniform: "scale an image size by the same factor on the horizontal and virtical dimension
2. Non-Uniform: "Using different scale factors for each dimension.

- Indeed artifacts are more likely to occur in **Non-Uniform** scaling.

#### Uniform Scaling

- Down Scaling an image means that we must remove pixels from the image, the question is which pixels should we remove?
- We sample the pixels (that is unformily choose a subset of pixels) AKA **sub-sampling**. Aliasing can occur in this case as well as we're still sampling.
- Features of aliasing are:
    1) staircase effect of pixels
    2) color Noise

How to uniform scale down: 
- $\text{ Given a Matrix of pixel dimension  m-by-n, and we want to scale to p-by-q then: } \displaystyle \frac{p}{m} = \frac{q}{n}$

- $\text{Nearest Neighbor Scaling := choosing a color of the old image that is } \mathbf{closest} \text{ to the center of the new size of the image.}$
