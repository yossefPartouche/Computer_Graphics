#### Sampling and Reconstructing - Greyscale
- A greyscale image is a 2D grid of values that represent the light intensity at some sample points in the scene.
Image the grid of the 2D image as a 2D domain XY, and think of the Z plane as the scalar value of the XY paramater function that is $f(x,y) = z$ which shows are much light hit that specific region.

#### THhe NYQUIST - SHANNON THEOREM: A sinal can be reconstructed from its samples if the roiginal signal has no frequencies **above** half the sampling frequency.

- For us we can discuss frequency as the rate of change in space, and the theorem means that the reconstruction ability links between the rate of smapling and rate of change of the funtion.

#### Reconstruction

**Q. What is the actual reconstruction function that we use based on the pixel values to create the image**

Even though the samples are just point samples with 0 dimensions, when zooming into an image we see many regions covered by similar color
So a greyscale image over the XY domain is a constant peicewise function.

To understand the reconstruction we look again at the 1D function: 
- Given a set of sample points on the co-domain.
- We divide the domain into uniform intervals.
- Assign a specific interval a value that was sampled. (producing a step function)
Indeed the fucntion won't be continous but in large images it won't matter and should suffice as the pixels are so small.

if we wanted to reconstruct the continous function then:
- A simple method could be done via - "connect the dots", however, the min/max points are too "Sharp" and not continous.
- We can smoothen those sections via higher order functions (discussed Later).
