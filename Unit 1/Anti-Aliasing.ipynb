#### Anti Aliasing
#### Objectives
- Learn two methods to scale down an image that will reduce the anti aliasing effect
    1. Simple Average
    2. Bilinear Interpolation

- As mentioned earlier, to reduce an image we must sample (in this case unformly) points within the image. These points will be within a radius of a certain number of pixels that won't be in the image and we must choose an appropriate color value for our sample point. 
### Simple Averaging
1. Choose a certain number of pixel radius.
2. For each R, G and B compute their sum. 
3. divide by the number of pixels
$$ C_{new} = (\frac{\sum_{i \in Neighborhood}r_i}{|Neighborhood|}, \frac{\sum_{i \in Neighborhood}g_i}{|Neighborhood|},\frac{\sum_{i \in Neighborhood}b_i}{|Neighborhood|})$$

#### Bilinear Interpolation

- This method invovles complete a proceedure (Interpolating) 2 times over. we shall run the process over a general n-dimension and we can apply it to $\R^3$  
1. Let $f: \R^n \rightarrow \R$
    - The domain will refer to a specific area in the image (i.e. length and hieght)
    - The codomain refers to the colour value of one of the specific channels (R XOR G XOR B)
2. Let $p_1, p_2, p_3, p_4 \in \R^n$ such that $f(p_i) = v_i \ \forall v_i \in R$
3. Let $p \in \R^n$ be in the neighbourhood of those points.

<img src="image_U1/Hdint.png" height="300" width="300"/>

4. W.L.O.G choose $p_1, p_2$ for our first interpolation $$ y = tf(p_2) + (1-t)f(p_1)$$ where $$ t = ||\frac{p - p_1}{p_2-p_1}||$$. 
    - Note that t is the weight that will be applied to to colors base on the distance **created** between $p$ and $p_1$. 
    - Note that $p$ is on the "line" between $p_1$ and $p_2$ but we'll need to find it ourselves, it's likely our point inside the radius and not on the "line". 
        1. Define the vector $\vec{V_{12}} = p_2 - p_1$
        2. Define the vector $\vec{V_{p1}} = p - p_1$
        3. Solve for t $$ p_{prj} = p_1 + \left(\frac{\vec{V_{p1}}•\vec{V_{12}}}{\vec{V_{12}}^2}\right)\vec{V_{12}}$$ (Note we have dot product and multiplication).
        4. Once you found $p_{prj}$ it becomes the $p$ above and solve for $t$.
5. We created point $p'$ interpolated from $p_1$ and $p_2$.
6. Complete step 4 over $p_3$ and $p_4$ to produce $p''$ 
7. Use $p'$ and $p''$ to find $$f(p) = tf(p') + (1-t)f(p'')$$

#### Running Example
Given:
$$p_1 = (1,\ 1,\ 1)$$
$$p_2 = (4,\ 3,\ 2)$$
$$p = (3,\ 5,\ 4)$$

##### Step 1: Compute direction vector of the line

$$
\vec{v} = p_2 - p_1 = (4 - 1,\ 3 - 1,\ 2 - 1) = (3,\ 2,\ 1)
$$


##### Step 2: Compute vector from $p_1$ to $p$

$$
\vec{w} = p - p_1 = (3 - 1,\ 5 - 1,\ 4 - 1) = (2,\ 4,\ 3)
$$

##### Step 3: Compute the projection scalar $t$

$$
t = \frac{\vec{w} \cdot \vec{v}}{\vec{v} \cdot \vec{v}} = 
\frac{(2)(3) + (4)(2) + (3)(1)}{(3)^2 + (2)^2 + (1)^2}
= \frac{6 + 8 + 3}{9 + 4 + 1} = \frac{17}{14} \approx 1.214
$$

##### Step 4: finding the point using t
$$p_{proj} = p_1 + t\vec{v} = (1,\ 1,\ 1) + 1.214(3,\ 2,\ 1)$$


These two method fall under the same family of what's known **Weighted Averaging** for the following reason: 
1. We define a neighborhood of pixels (we choose 4).
2. We calculated the value of our pixel, using an average of the neighboring points. 
3. The weights were none negative and also summed to 1. 
4. Where the Linear combination of each pixel with repective weight created our new color pixel.

$$\sum_{i \in Neighbors}w_i = 1 $$

$$C = \sum_{i \in Neighbors}w_iC_i$$
