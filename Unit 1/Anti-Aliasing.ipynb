#### Objectives
- Learn two methods to scale down an image that will reduce the anti aliasing effect
    1. Simple Average
    2. Bilinear Interpolation

- As mentioned earlier, to reduce an image we must sample (in this case unformly) points within the image. These points will be within a radius of a certain number of pixels that won't be in the image and we must choose an appropriate color value for our sample point. 
### Simple Averaging (Nearest Neighbors)
1. Choose a certain number of pixel radius.
2. For each R, G and B compute their sum. 
3. divide by the number of pixels
$$ C_{new} = (\frac{\sum_{i \in Neighborhood}r_i}{|Neighborhood|}, \frac{\sum_{i \in Neighborhood}g_i}{|Neighborhood|},\frac{\sum_{i \in Neighborhood}b_i}{|Neighborhood|})$$
