### Multi-Size Images
##### Objective:
Learn how to enlarge an image beyond the original size

In order to increase the size we need to add new seams to the image that weren't there.
The Same Process of Seam removal occurs, where we dynamically find the optimal seam **BUT** this time we duplicate the seam (instead of removing the seam). 
#### Q. What would happen if we want to apply this method multiple times?
#### A.
Since we duplicated the best seam, the optimal path will be the exact same so if we don't take care of this scenario, we could end up with an image with an enlarged area out of proportion.

#### Solution: 
Choose the seam before expansion, if we want to increase the size by k pixels, then we choose the k-seams to remove before applying the seams.

In practice, this means that we apply seam-removal k-times but don't remove them we'll add them instead.
This will ensure we find **k-different unoticable** seams.

##### Example
Suppose we would like to convert the aspect ratio of an image who size is 640x480 pixels to 16:9 we can:
- We can insert 214 vertical seams
- We can remove 92 horizontal seams and insert 30 vertical seams
- We can remove 120 horizontal seam from the image.
- We can insert 30 vertical seams and then remove 92 horizontal seams.

### Increasing and Descreasing gradient magnitude
- If we want to avoid removing a seam we can increase it's magnitude gradient making it more unlikely to be removed, at the same time we can reduce the gradient magnitude of a region to ensure removal of the seam.

1. Mark the object
2. Remove seams (horizontal or vertical) until the object is removed.
3. Insert the same number of seams to reduced image (to return it to original size).
