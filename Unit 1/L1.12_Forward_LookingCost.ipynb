### Forward Looking Cost
##### objective:
1. Understand why aliasing still arises after retargeting an image
2. Devise a way to further reduce the effect

Artifacts are usually created when the smoothness of the image is broken because the removal of a pixels, when new pixesls that were not in the original image appear in the retargeted image. 

To try and reduce such artifacts.

$\text{Instead of looking for the optimal seem on the originl image, we look forward for the optimal seem on the resulting image.}$

#### Q. What new edges appear after a seem was removed?
#### A.These edges are pixels that became new neighbors after a seam was removed.

- the cost of the new edges will be the **difference between the colors** of the new neighboring pixels.





#### Examining Cost

1. For any pixel I(x,y) there are three possible seams passing through it. $\left\{ I(i-1,j-1),\ I(i-1, j),\ I(i-1, j+1) \right\}$
2. If the seam comes from the left, then we have two pairs of new neighbors and two new edges are created as follows: 
$$C_L(i,j) = |I(i,j+1) - I(i,j-1)| + |I(i-1, j)-I(i, j-1)|$$
3. If the seam comes from the right then:
$$C_R(i,j) = |I(i,j+1) - I(i,j-1)| + |I(i-1, j)-I(i, j+1)|$$
4. If the seam comes from above, the we'll see just one new edge: 
$$C_U(i,j) = |I(i,j+1) - I(i,j-1)|$$


With this we produce a more complete and better form of the dynamic programming equation:
$$M(i,j) = \min \left\{ M(i-1, j-1) + C_L(i,j),\ M(i-1, j) + C_U(i,j),\ M(i-1, j+1) + C_R(i,j) \right\}$$

$\text{The transition cost, indicates how disruptive removing that perticular pixel from a perticular direction will be in the resulting image}$

### The difference 

- The old formula looked at the gradients and the edges of the original image. 
- The new formula looks forward **AND** acount for the cost of the new edges being formed for the 3 possible seams coming from the previous row
- If want to add some pixel attributes such as gradient Magnitude and manual markings we can do so by adding it to the updated formula. 

the seam carving algorithm as a whole doesn't change: 
- we still use a cost matrix of the size of the image and fill from top to bottom (or left to right).
