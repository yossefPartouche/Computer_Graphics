### Noise
##### Objective
- Understand what is noise and how it's created. 
- How can we reduce noise
- The trade off reducing noise.

Noise is a phenomenon that also occurs in sampling – within the framework of sampled values, the sampled data does not reflect the true values of the signal. This can happen due to factors related to disturbances during the signal measurement process. This phenomenon also occurs in the context of image capturing.
Here some cause for noise:
1. Dirt/dust in the camera lense/sensors
2. Lack of light when capturing an image
3. Sensor in the camera is broken/faulty.

In order to understand noise better to be able to solve this problem we turn to a 1D function.

If we consider the original function and compare it to the noisy function we can see that for a specific pixel it will contain a range of values, which creates difference in the color of the neighboring pixels that weren't in the original image.

### De-Noising
This is the process of removing the noise from the function, most of the time this is possible by dividing the function into 2 sections.
1. Noise section
2. clean Section

If the function creates **small** fluctuations in the values of the original funciton we can smoothen out the noise.

<img src="image_U1/noisy.png" height="400" width="400"/>


One of the method that can be used is simple averaging, where in this case for each point of fluctuational values we take an average of all the points. 
$$f_{new} = \frac{\sum_{x' \in N}f(x')}{|N|}$$
where N is the neighborhood.

The more neighborhood we use to average the more **smooth** the resulting image will look. 

IMPORTANT: We need to be careful, not too smoothen out the picture too much as we will be removing details that define the image.

NOTE: Edge case, if we are missing values in the neighborhood, for calculation purposes we can:
1. add 0s 
2. choose the nearest neighbor
3. Assume that the image circular that is take the pixel from the of the side of the matrix.
