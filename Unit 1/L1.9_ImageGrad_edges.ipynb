#### Details
Notice that an image with a region of a lot of detail will have many different levels of color and tones etc, compared to a single coat of colour in anoter region will have less detail. 
Since we have a function that measures the change of these colors in a region $\R^2$ and this is a continous function we can fromalise details as follows:
Let $f : \R^2 \rightarrow \R$
$$ Detail :=
\nabla f(x, y) = \left( \frac{\partial f}{\partial x},\ \frac{\partial f}{\partial y} \right)
$$

The relation is indeed linearly proportional to the derivative.
Now when looking as a strip of the image we can simplify the dimension to 1D: 
$$ Detail :=
\frac{d}{dx}f(x) 
$$

Using Tailor Expansion we can approximate details in the following manner:
$$ Forward \ Differencing := \frac{d}{dx}f(x)  = \frac{f(x+\Delta x) - f(x)}{\Delta x} + O(\Delta x)$$
$$ Backward \ Differencing := \frac{d}{dx}f(x)  = \frac{f(x)- f(x - \Delta x)}{\Delta x} + O(\Delta x)$$
$$ Central \ Differencing := \frac{d}{dx}f(x)  = \frac{f(x + \Delta x) - f(x - \Delta x)}{2\Delta x} + O(\Delta x)$$

This is helpful since we're able to define details over discrete values (pixel values since the smallest $\Delta$ is 1 pixel) and these simplify to:
##### Forward Differcing

$$I(x+1, y) - I(x,y)$$
$$I(x, y+1) - I(x,y)$$

##### Backward Differcing

$$I(x, y) - I(x-1,y)$$
$$I(x, y) - I(x,y-1)$$

#### Image magnitude (Eucledian Norm Method)
$$ | \Delta I(x,y)| = \sqrt{\left ( \frac{d}{dx}I(x,y)\right)^2 + \left( \frac{d}{dy}I(x,y)\right)^2}$$
- The values in this case will range from [0, 360]
- This will only present exactly where there are details contained (no direction)

#### Gradient of Noisy Image
As we apply guassian Filtering on a noisy image, the range of the gradient and thus gradient magnitude reduces and directly proportional to when the value of sigma for the guassian kernel increases.
- **Important** If we do not scale the on each type of filtering the image would just be black.
- As smoothing gets larger the spacial (space) frequency is reduced, this means we need smaller samples to reconstruct a noisless image, with fewer aliasing artifacts
