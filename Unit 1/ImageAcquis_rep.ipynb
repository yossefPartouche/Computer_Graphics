The first camera was created by Joseph Niepce, it was discovered by the phenonmenon, that an image of an outside scene is projected through a small hole on the back side of a closed box, the image that is created is inverted left to right and upside down. Joseph Niepce, place a peice of paper coated with silver chloride on the far side of the room, and this material darkened when exposed to light and so an inverted picture of what was scene through the hole was imprinted on the paper.

Modern Day Cameras have the same principle:

#### Digital Camera
The difference in digital cameras is that instead of a physcial light sensitive material to imprint light. Instead it has digital devices such as CCDs and CMOS that are measuring light. They are made up of a matrix of small light sennsors that measure the amount of light hitting them by converting photons to electrons.

#### Amount of light Captured by Sensors (BLACK and WHITE)
To create an image, each sensor element can be converted into one pixel, we can intereoret the amount of light hitting the element as the grey Level of the pixel. 

Black = No Light hits the sensor  
White  = A lot of light hit the sensor

Indeed this can be converted into a range of values to precise the amount of light that was hit, this is done in levels of bits. 
We use 1 Byte (8 bits) to represent 256 range of greys.

So when a MATRIX of sensors in the camera captures a greyscale image it can be stored and represented as a grid of NUMBERS from 0 to 255. 

Note: That this each cell rerpesents a region and is given a single value, and if the region is very large it will still only apply a single value to represent that region (it could lead to misrepresentation of the image color). This process is call **Sampling** Taking a continous image and converting it to discrete sections.

#### Color

For now, Color is a small section of the light spectrum known as visible light, in which the primary colours are red, green and blue from which we can create any colors by adjusting the intensity of each of the primary colors.  
Digitally, the digital sesors have the three primary colors sensors and similar to the values represented in a grid via sampling, each section will contain 3 values (array of 3) ranging from 0 - 255 each.
