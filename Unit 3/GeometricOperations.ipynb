### Geometric Operations on Simple object (Implicit/Parametric) 
- Points
- Lines
- Planes

#### How to Calculate Distance of 3D point to 3D plane
Input: Implicit Representation of plane $N_xx + N_yy + N_zz + d = 0$ where $N = (N_x, N_y, N_z)$ is the unit-normal to the plane and d is the distance to origin.

Note: Any point that lies on the plane must satisfy the following: $\vec{N}•p + d = 0$

IF: you have a point $p$ not on the plane then you have to find the point $q$ on the plane which is leads to the minimal distance between another. This can be done by finding the pojection point of $p_0$ on the given plane.   
So $p_0 = q + t\vec{N}$. We find $t  = \frac{\vec{N}•p_0 + d}{\vec{N}•\vec{N}}$

### Example:
Given a paramteric representation of a plane: $$π = (1,2,3) + t(1,2,3) +s(1,1,0)$$
What is the distance between the point p = (1,1,1) and the plane π?


### Solution

Step 1: Convert the parametric form to implicit form
First, we need to find the normal vector to the plane by taking the cross product of the two direction vectors:

$\vec{n} = (1,2,3) \times (1,1,0)$

$\vec{n} = \begin{vmatrix} \vec{i} & \vec{j} & \vec{k} \\ 1 & 2 & 3 \\ 1 & 1 & 0 \end{vmatrix}$

Computing each component:

$\vec{i}$ component: $(2 \times 0) - (3 \times 1) = 0 - 3 = -3$
$\vec{j}$ component: $(3 \times 1) - (1 \times 0) = 3 - 0 = 3$
$\vec{k}$ component: $(1 \times 1) - (2 \times 1) = 1 - 2 = -1$
So the normal vector is $\vec{n} = (-3, 3, -1)$

The plane contains the point $(1,2,3)$, so the implicit form is: $-3(x-1) + 3(y-2) - (z-3) = 0$

Simplifying: $-3x + 3 + 3y - 6 - z + 3 = 0$ $-3x + 3y - z = 0$

Step 2: Calculate the distance using the formula
The distance from a point to a plane is given by:

$$d = \frac{|ax_0 + by_0 + cz_0 + d|}{\sqrt{a^2 + b^2 + c^2}}$$

Where $(a,b,c)$ is the normal vector and $d$ is the constant term in the implicit equation.

For our plane $-3x + 3y - z = 0$, we have $a = -3$, $b = 3$, $c = -1$, and $d = 0$.

For the point $p = (1,1,1)$:

$$d = \frac{|-3(1) + 3(1) - 1(1) + 0|}{\sqrt{(-3)^2 + 3^2 + (-1)^2}}$$

$$d = \frac{|-3 + 3 - 1|}{\sqrt{9 + 9 + 1}}$$

$$d = \frac{|-1|}{\sqrt{19}}$$

$$d = \frac{1}{\sqrt{19}}$$

$$d = \frac{1}{\sqrt{19}} = \frac{1}{√19} ≈ 0.2294$$

Therefore, the distance between the point $p = (1,1,1)$ and the plane $π$ is $\frac{1}{√19}$ or approximately $0.2294$ units.
