### Enumeration of surface with Paramteric

$$ F(s,t) = (f_x(s,t), f_y(s,t), f_z(s,t))$$

#### How can we find points that lie on this surface?

We choose two values for $s \ and \ t$ and map them to the geometric space using the function $F(s,t)$. The great thing about this is that any two points in the given constraint of the paramters will be on the surface, so searching is very easy.

So creating/drawing geometric objects using paramteric equations is very useful. 

### For which operations should we use parametric and for which should we use implicit?

|                         | Parametric | Implicit  |
|-------------------------|------------|-----------|
| **Enumeration**          | ✅  (Easy)        | ❌ (Hard)        |
| **Point Classification** | ❌ (Hard)         | ✅   (Easy)      |
