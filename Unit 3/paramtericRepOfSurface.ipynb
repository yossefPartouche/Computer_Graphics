### Parametric Representation of Surface

Reminder: Paramteric representation is a mapping from a parametric space U with a paramter $t \ at \ most \in \R$ to a geometric space $G$  
Depending on how we constrain the parameter t we can represent a line, ray or a line segment in the geometric space.

### Extension of paramteric representation to plane

A plane is defined using 3 points that are not collinear denote $p_0, p_1, p_2$
between any two of the points we can define a parametric equation of a line, e.g. $p(u) = (1-u)p_0 + up_1$ and this line will lie on the plane. 
Any point on this line can be represented from the paramtric equation by finding the combination with $u$.  

Suppose we want to find a line passing through $p(u)$ whilst also passing through $p_2$ this line must also lie on the plane.

To do so we'll use a different parametric equation with parameter $v \in \R$ so $q(v) = (1-v)p_2 + vp(u)$

So if we change the point $p(u)$ by changing the parameter $u$ and then adjust $q(v)$ to produce a different line that will still lie on the plane. What this really means is that by changing $v$ and $u$ or even changing the liness on the planes i.e. line between $p_0$ and $p_2$ and then have and the connecting line of $p(u)$ pass through $p_1$; we can cover every point on the plane. 

The above can be summarised: 

To define a plane parametrically: $$S(u,v) = (1-v)p_2 + vp(u)$$ Expanding some more produces $$S(u,v) = (1-v)p_2 + v((1-u)p_0 + up_1)$$ $$= v(1-u)p_0 +uvp_1+ (1-v)p_2$$  

with the general form of the equation we can represent any geometric plane.
